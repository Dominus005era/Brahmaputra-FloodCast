```python
# STEP 26B: RECONSTRUCT RAINFALL DATASET

import pandas as pd
import os

raw_path = os.path.join(os.getcwd(), "data", "raw")

# Load both rainfall files
rainfall_2021_2025 = pd.read_csv(
    os.path.join(raw_path, "assam_rainfall_2021_2025.csv")
)

rainfall_2026_2030 = pd.read_csv(
    os.path.join(raw_path, "assam_rainfall_2026_2030.csv")
)

# Combine both files
rainfall_all = pd.concat(
    [rainfall_2021_2025, rainfall_2026_2030],
    ignore_index=True
)

# Convert timestamp
rainfall_all["Data Acquisition Time"] = pd.to_datetime(
    rainfall_all["Data Acquisition Time"],
    errors="coerce"
)

print("RAINFALL DATASET RECONSTRUCTED")
print("=" * 70)

print("Total rows:", len(rainfall_all))
print("Total columns:", rainfall_all.shape[1])

print(
    "Date range:",
    rainfall_all["Data Acquisition Time"].min(),
    "to",
    rainfall_all["Data Acquisition Time"].max()
)

print(
    "Missing timestamps:",
    rainfall_all["Data Acquisition Time"].isna().sum()
)
```

```python
# STEP 26D: FILTER TARGET RAINFALL STATION

target_station = "NH15 Crossing Fakirpara Tangni"

rainfall_fakirpara = rainfall_all[
    rainfall_all["Station"] == target_station
].copy()

print("TARGET RAINFALL DATASET")
print("=" * 70)

print("Station:", target_station)
print("Rows:", len(rainfall_fakirpara))
print("Columns:", rainfall_fakirpara.shape[1])

print(
    "Date range:",
    rainfall_fakirpara["Data Acquisition Time"].min(),
    "to",
    rainfall_fakirpara["Data Acquisition Time"].max()
)

print(
    "Missing timestamps:",
    rainfall_fakirpara["Data Acquisition Time"].isna().sum()
)

print(
    "Duplicate timestamps:",
    rainfall_fakirpara["Data Acquisition Time"].duplicated().sum()
)
```

```python
# STEP 26F: COMPLETE DUPLICATE PAIR VERIFICATION

pair_check = (
    duplicate_rainfall_audit
    .groupby(time_col)[rainfall_col]
    .agg(["min", "max", "count"])
    .reset_index()
)

pair_check["difference"] = (
    pair_check["max"] - pair_check["min"]
)

pair_check["ratio"] = (
    pair_check["max"] / pair_check["min"]
)

print("DUPLICATE PAIR VERIFICATION")
print("=" * 70)

print("Total duplicate timestamps:", len(pair_check))
print("Pairs with exactly 2 records:",
      (pair_check["count"] == 2).sum())

print(
    "Exact duplicate-value pairs:",
    (pair_check["difference"] == 0).sum()
)

print(
    "Pairs with ratio exactly 2:",
    (pair_check["ratio"] == 2).sum()
)

print(
    "Pairs with ratio NOT equal to 2:",
    (pair_check["ratio"] != 2).sum()
)

print("\nNon-2× pairs:")
display(
    pair_check[
        pair_check["ratio"] != 2
    ]
)
```

```python
# STEP 27: CREATE DUPLICATE RESOLUTION TABLE

duplicate_resolution = (
    rainfall_fakirpara[
        rainfall_fakirpara[time_col].isin(duplicate_times)
    ]
    .groupby(time_col)[rainfall_col]
    .agg(["min", "max", "count"])
    .reset_index()
)

duplicate_resolution["difference"] = (
    duplicate_resolution["max"] -
    duplicate_resolution["min"]
)

duplicate_resolution["ratio"] = (
    duplicate_resolution["max"] /
    duplicate_resolution["min"]
)

duplicate_resolution["resolution_status"] = "UNRESOLVED_DUPLICATE"

# Exact same-value duplicates can be safely represented by one value
duplicate_resolution.loc[
    duplicate_resolution["difference"] == 0,
    "resolution_status"
] = "EXACT_DUPLICATE"

print("STEP 27 — DUPLICATE RESOLUTION TABLE")
print("=" * 70)

print("Total duplicate timestamps:",
      len(duplicate_resolution))

print(
    "Exact duplicates:",
    (duplicate_resolution["resolution_status"] == "EXACT_DUPLICATE").sum()
)

print(
    "Unresolved conflicting duplicates:",
    (duplicate_resolution["resolution_status"] == "UNRESOLVED_DUPLICATE").sum()
)

display(
    duplicate_resolution.head(30)
)
```

```python
# STEP 28: CREATE CLEAN RAINFALL WORKING DATASET
import numpy as np
rainfall_clean = rainfall_fakirpara.copy()

# Count observations at each timestamp
timestamp_counts = (
    rainfall_clean[time_col]
    .value_counts()
)

# Identify duplicate timestamps
duplicate_timestamp_set = set(
    timestamp_counts[timestamp_counts > 1].index
)

# Separate unique and duplicate records
unique_rows = rainfall_clean[
    ~rainfall_clean[time_col].isin(duplicate_timestamp_set)
].copy()

duplicate_rows = rainfall_clean[
    rainfall_clean[time_col].isin(duplicate_timestamp_set)
].copy()

# ---------------------------------------------------------
# Exact duplicate timestamps
# ---------------------------------------------------------

exact_duplicate_times = set(
    duplicate_resolution.loc[
        duplicate_resolution["resolution_status"] == "EXACT_DUPLICATE",
        time_col
    ]
)

exact_duplicate_rows = duplicate_rows[
    duplicate_rows[time_col].isin(exact_duplicate_times)
].copy()

# Keep one record for each exact duplicate timestamp
exact_duplicate_clean = (
    exact_duplicate_rows
    .sort_values(time_col)
    .drop_duplicates(
        subset=[time_col],
        keep="first"
    )
    .copy()
)

# ---------------------------------------------------------
# Conflicting duplicate timestamps
# ---------------------------------------------------------

conflicting_times = set(
    duplicate_resolution.loc[
        duplicate_resolution["resolution_status"] == "UNRESOLVED_DUPLICATE",
        time_col
    ]
)

conflicting_rows = duplicate_rows[
    duplicate_rows[time_col].isin(conflicting_times)
].copy()

# Create one representative row per conflicting timestamp
# but mark rainfall as NaN because true value is unknown

conflicting_clean = (
    conflicting_rows
    .sort_values(time_col)
    .drop_duplicates(
        subset=[time_col],
        keep="first"
    )
    .copy()
)

conflicting_clean[rainfall_col] = np.nan

# ---------------------------------------------------------
# Combine
# ---------------------------------------------------------

rainfall_clean = pd.concat(
    [
        unique_rows,
        exact_duplicate_clean,
        conflicting_clean
    ],
    ignore_index=True
)

rainfall_clean = (
    rainfall_clean
    .sort_values(time_col)
    .reset_index(drop=True)
)

print("STEP 28 — CLEAN RAINFALL WORKING DATASET")
print("=" * 70)

print("Original rows:", len(rainfall_fakirpara))
print("Clean rows:", len(rainfall_clean))

print(
    "Duplicate timestamps remaining:",
    rainfall_clean[time_col].duplicated().sum()
)

print(
    "Missing rainfall values:",
    rainfall_clean[rainfall_col].isna().sum()
)

print(
    "Exact duplicates safely retained:",
    len(exact_duplicate_clean)
)

print(
    "Conflicting duplicate timestamps flagged:",
    len(conflicting_clean)
)
```

```python
# STEP 38: RAINFALL RECENCY TEST

rainfall_valid = rainfall_clean[
    rainfall_clean["Telemetry Hourly Rainfall (mm)"].notna()
].copy()

rainfall_valid = rainfall_valid[
    [
        "Data Acquisition Time",
        "Telemetry Hourly Rainfall (mm)"
    ]
].copy()

rainfall_valid = rainfall_valid.rename(
    columns={
        "Data Acquisition Time": "rainfall_time",
        "Telemetry Hourly Rainfall (mm)": "rainfall_value"
    }
)

rainfall_valid = rainfall_valid.sort_values("rainfall_time")

water_base = water_level_final[
    ["Data Acquisition Time"]
].copy()

water_base = water_base.rename(
    columns={
        "Data Acquisition Time": "water_time"
    }
)

water_base = water_base.sort_values("water_time")

# Find latest rainfall observation at or before each water-level timestamp
rainfall_recency = pd.merge_asof(
    water_base,
    rainfall_valid,
    left_on="water_time",
    right_on="rainfall_time",
    direction="backward"
)

# Calculate how old the rainfall observation is
rainfall_recency["rainfall_age_hours"] = (
    rainfall_recency["water_time"]
    - rainfall_recency["rainfall_time"]
).dt.total_seconds() / 3600

print("STEP 38 — RAINFALL RECENCY TEST")
print("=" * 70)

print("Water-level timestamps:", len(rainfall_recency))

print(
    "Water-level timestamps with previous valid rainfall:",
    rainfall_recency["rainfall_time"].notna().sum()
)

print(
    "Water-level timestamps with no previous rainfall:",
    rainfall_recency["rainfall_time"].isna().sum()
)

print("\nRainfall age statistics:")

display(
    rainfall_recency["rainfall_age_hours"].describe()
)

print("\nRainfall availability by recency:")

for hours in [1, 3, 6, 12, 24, 48, 72]:
    
    count = (
        rainfall_recency["rainfall_age_hours"] <= hours
    ).sum()

    percentage = (
        count / len(rainfall_recency) * 100
    )

    print(
        f"Within {hours:>2} hours : "
        f"{count:>5} "
        f"({percentage:.2f}%)"
    )
```

```python
# STEP 39 — RAINFALL-WATER LEVEL TEMPORAL ALIGNMENT TEST

water_time = pd.to_datetime(
    water_level_final["Data Acquisition Time"]
)

rainfall_valid = rainfall_clean[
    rainfall_clean["Telemetry Hourly Rainfall (mm)"].notna()
].copy()

rainfall_valid = rainfall_valid[
    [
        "Data Acquisition Time",
        "Telemetry Hourly Rainfall (mm)"
    ]
].sort_values("Data Acquisition Time")

# ---------------------------------------------------------
# Create water-level timeline
# ---------------------------------------------------------

alignment_test = pd.DataFrame({
    "Data Acquisition Time": water_time
}).sort_values("Data Acquisition Time")

# ---------------------------------------------------------
# Find most recent rainfall observation within 72 hours
# ---------------------------------------------------------

alignment_test = pd.merge_asof(
    alignment_test,
    rainfall_valid,
    on="Data Acquisition Time",
    direction="backward",
    tolerance=pd.Timedelta(hours=72)
)

# ---------------------------------------------------------
# Find actual rainfall timestamp separately
# ---------------------------------------------------------

rainfall_timestamp = rainfall_valid[
    ["Data Acquisition Time"]
].rename(
    columns={
        "Data Acquisition Time": "rainfall_timestamp"
    }
)

alignment_test = pd.merge_asof(
    alignment_test.sort_values("Data Acquisition Time"),
    rainfall_timestamp.sort_values("rainfall_timestamp"),
    left_on="Data Acquisition Time",
    right_on="rainfall_timestamp",
    direction="backward",
    tolerance=pd.Timedelta(hours=72)
)

# ---------------------------------------------------------
# Calculate rainfall age
# ---------------------------------------------------------

alignment_test["rainfall_age_hours"] = (
    alignment_test["Data Acquisition Time"]
    - alignment_test["rainfall_timestamp"]
).dt.total_seconds() / 3600

# ---------------------------------------------------------
# Classify rainfall recency
# ---------------------------------------------------------

def classify_rainfall_age(hours):

    if pd.isna(hours):
        return "No rainfall within 72h"

    elif hours <= 1:
        return "0-1 hour"

    elif hours <= 3:
        return "1-3 hours"

    elif hours <= 6:
        return "3-6 hours"

    elif hours <= 12:
        return "6-12 hours"

    elif hours <= 24:
        return "12-24 hours"

    elif hours <= 48:
        return "24-48 hours"

    else:
        return "48-72 hours"


alignment_test["rainfall_recency_group"] = (
    alignment_test["rainfall_age_hours"]
    .apply(classify_rainfall_age)
)

# ---------------------------------------------------------
# FINAL OUTPUT
# ---------------------------------------------------------

print("STEP 39 — RAINFALL-WATER LEVEL TEMPORAL ALIGNMENT")
print("=" * 70)

print(
    "Total water-level timestamps:",
    len(alignment_test)
)

print("\nRainfall alignment groups:")

display(
    alignment_test[
        "rainfall_recency_group"
    ]
    .value_counts()
    .reindex([
        "0-1 hour",
        "1-3 hours",
        "3-6 hours",
        "6-12 hours",
        "12-24 hours",
        "24-48 hours",
        "48-72 hours",
        "No rainfall within 72h"
    ])
    .fillna(0)
    .astype(int)
)

print(
    "\nRainfall aligned within 24 hours:",
    alignment_test["rainfall_age_hours"].le(24).sum()
)

print(
    "Rainfall aligned within 72 hours:",
    alignment_test["rainfall_age_hours"].le(72).sum()
)

print(
    "No rainfall within 72 hours:",
    alignment_test["rainfall_age_hours"].isna().sum()
)
```

```python
# ================================================================
# STEP 43 — RAINFALL CUMULATIVE FEATURE QUALITY CHECK
# ================================================================

import pandas as pd
import numpy as np

print("STEP 43 — RAINFALL CUMULATIVE FEATURE QUALITY CHECK")
print("=" * 70)

# ---------------------------------------------------------------
# 1. SOURCE DATA
# ---------------------------------------------------------------

water = water_level_final.copy()
rain = rainfall_clean.copy()

time_col = "Data Acquisition Time"
rainfall_col = "Telemetry Hourly Rainfall (mm)"

water[time_col] = pd.to_datetime(water[time_col], errors="coerce")
rain[time_col] = pd.to_datetime(rain[time_col], errors="coerce")

water = (
    water
    .dropna(subset=[time_col])
    .sort_values(time_col)
    .drop_duplicates(subset=[time_col])
    .reset_index(drop=True)
)

# Only VALID rainfall observations
rain[rainfall_col] = pd.to_numeric(
    rain[rainfall_col],
    errors="coerce"
)

rain_valid = (
    rain[
        rain[time_col].notna() &
        rain[rainfall_col].notna()
    ][[time_col, rainfall_col]]
    .sort_values(time_col)
    .drop_duplicates(subset=[time_col])
    .reset_index(drop=True)
)

print("Water-level timestamps:", len(water))
print("Valid rainfall observations:", len(rain_valid))

# ---------------------------------------------------------------
# 2. CALCULATE PREVIOUS RAINFALL WINDOW FEATURES
# ---------------------------------------------------------------

water_times = water[time_col].values
rain_times = rain_valid[time_col].values
rain_values = rain_valid[rainfall_col].astype(float).values

rain_ns = rain_times.astype("datetime64[ns]").astype("int64")
water_ns = water_times.astype("datetime64[ns]").astype("int64")

# Cumulative sum of rainfall
rain_cumsum = np.concatenate(
    ([0.0], np.cumsum(rain_values))
)

def calculate_window(window_hours):

    window_ns = window_hours * 60 * 60 * 1_000_000_000

    # Include rainfall at exact water timestamp
    right_idx = np.searchsorted(
        rain_ns,
        water_ns,
        side="right"
    )

    # Include observations from (t-window, t]
    left_idx = np.searchsorted(
        rain_ns,
        water_ns - window_ns,
        side="right"
    )

    sums = (
        rain_cumsum[right_idx]
        - rain_cumsum[left_idx]
    )

    counts = right_idx - left_idx

    return sums, counts


windows = [1, 3, 6, 12, 24]

rainfall_cumulative_43 = water[
    [time_col]
].copy()

for h in windows:

    sums, counts = calculate_window(h)

    rainfall_cumulative_43[f"rainfall_{h}h_sum"] = sums
    rainfall_cumulative_43[f"rainfall_{h}h_count"] = counts


# ---------------------------------------------------------------
# 3. WHERE NO RAINFALL EXISTS, SET FEATURES TO NaN
# ---------------------------------------------------------------

for h in windows:

    count_col = f"rainfall_{h}h_count"
    sum_col = f"rainfall_{h}h_sum"

    rainfall_cumulative_43.loc[
        rainfall_cumulative_43[count_col] == 0,
        sum_col
    ] = np.nan


# ---------------------------------------------------------------
# 4. QUALITY CHECK
# ---------------------------------------------------------------

print("\nCUMULATIVE FEATURE COVERAGE")
print("-" * 70)

for h in windows:

    count_col = f"rainfall_{h}h_count"
    sum_col = f"rainfall_{h}h_sum"

    available = (
        rainfall_cumulative_43[count_col] > 0
    ).sum()

    percentage = (
        available / len(rainfall_cumulative_43) * 100
    )

    print(
        f"Previous {h:2d}h : "
        f"{available:5d} timestamps "
        f"({percentage:6.2f}%)"
    )


# ---------------------------------------------------------------
# 5. CHECK CUMULATIVE LOGIC
# ---------------------------------------------------------------

print("\nCUMULATIVE SUM LOGIC CHECK")
print("-" * 70)

logic_checks = []

for h1, h2 in [(1, 3), (3, 6), (6, 12), (12, 24)]:

    col1 = f"rainfall_{h1}h_sum"
    col2 = f"rainfall_{h2}h_sum"

    valid = (
        rainfall_cumulative_43[col1].notna() &
        rainfall_cumulative_43[col2].notna()
    )

    violations = (
        rainfall_cumulative_43.loc[valid, col2]
        <
        rainfall_cumulative_43.loc[valid, col1]
    ).sum()

    logic_checks.append(
        [f"{h1}h <= {h2}h", violations]
    )

logic_check_df = pd.DataFrame(
    logic_checks,
    columns=["Comparison", "Violations"]
)

print(logic_check_df.to_string(index=False))


# ---------------------------------------------------------------
# 6. FEATURE STATISTICS
# ---------------------------------------------------------------

print("\nRAINFALL CUMULATIVE FEATURE STATISTICS")
print("-" * 70)

sum_cols = [
    f"rainfall_{h}h_sum"
    for h in windows
]

print(
    rainfall_cumulative_43[sum_cols]
    .describe()
    .round(3)
)


# ---------------------------------------------------------------
# 7. FINAL RESULT
# ---------------------------------------------------------------

print("\nFINAL RESULT")
print("-" * 70)

print(
    "rainfall_cumulative_43 Shape:",
    rainfall_cumulative_43.shape
)

print(
    "Duplicate timestamps:",
    rainfall_cumulative_43[time_col]
    .duplicated()
    .sum()
)

print("\nSTEP 43 COMPLETED.")
```

```python
# ============================================================
# STEP 4 — ROLLING FEATURE VALIDATION
# ============================================================

print("=" * 70)
print("STEP 4 — ROLLING FEATURE VALIDATION")
print("=" * 70)

check = df.sort_values("Data Acquisition Time").reset_index(drop=True).copy()

# ------------------------------------------------------------
# Helper function
# ------------------------------------------------------------

def validate_rolling_feature(df_check, feature_col, window_hours, operation):
    
    timestamps = df_check["Data Acquisition Time"]
    values = df_check["Current_Water_Level"]

    results = []

    for i in range(len(df_check)):

        current_time = timestamps.iloc[i]

        # IMPORTANT:
        # Only data from the past/current timestamp is allowed.
        start_time = current_time - pd.Timedelta(hours=window_hours)

        window = values[
            (timestamps >= start_time) &
            (timestamps <= current_time)
        ]

        if len(window) == 0:
            results.append(np.nan)
            continue

        if operation == "mean":
            results.append(window.mean())

        elif operation == "max":
            results.append(window.max())

        elif operation == "min":
            results.append(window.min())

    expected = pd.Series(results, index=df_check.index)

    valid = expected.notna()

    difference = (
        df_check.loc[valid, feature_col].astype(float)
        - expected.loc[valid].astype(float)
    ).abs()

    matches = (difference < 1e-8).sum()
    total = valid.sum()
    mismatches = total - matches

    print(f"\n{feature_col}")
    print(f"Expected window: previous {window_hours}h + current")
    print(f"Operation: {operation}")
    print(f"Checked rows: {total}")
    print(f"Matching rows: {matches}")
    print(f"Mismatching rows: {mismatches}")

    if total > 0:
        print(f"Match rate: {matches / total * 100:.2f}%")

# ------------------------------------------------------------
# Validate rolling features
# ------------------------------------------------------------

validate_rolling_feature(
    check,
    "WaterLevel_RollingMean_6h",
    6,
    "mean"
)

validate_rolling_feature(
    check,
    "WaterLevel_RollingMean_12h",
    12,
    "mean"
)

validate_rolling_feature(
    check,
    "WaterLevel_RollingMean_24h",
    24,
    "mean"
)

validate_rolling_feature(
    check,
    "WaterLevel_RollingMax_24h",
    24,
    "max"
)

validate_rolling_feature(
    check,
    "WaterLevel_RollingMin_24h",
    24,
    "min"
)

print("\n" + "=" * 70)
print("STEP 4 COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 4.1 — ROLLING FEATURE DIAGNOSTIC (AUTO COLUMN DETECTION)
# ======================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 4.1 — ROLLING FEATURE DIAGNOSTIC")
print("=" * 70)

# ----------------------------------------------------------------------
# 1. Show all available columns
# ----------------------------------------------------------------------

print("\nAVAILABLE COLUMNS:")
for i, col in enumerate(df.columns, 1):
    print(f"{i:02d}. {col}")

# ----------------------------------------------------------------------
# 2. Automatically identify timestamp column
# ----------------------------------------------------------------------

timestamp_candidates = [
    col for col in df.columns
    if any(word in col.lower() for word in [
        "datetime",
        "timestamp",
        "date_time",
        "date",
        "time"
    ])
]

print("\n" + "-" * 70)
print("TIMESTAMP COLUMN CANDIDATES")
print("-" * 70)

for col in timestamp_candidates:
    print("-", col)

# ----------------------------------------------------------------------
# 3. Automatically identify water-level source column
# ----------------------------------------------------------------------

water_candidates = [
    col for col in df.columns
    if (
        "waterlevel" in col.lower()
        or "water_level" in col.lower()
        or "water level" in col.lower()
        or "river water level" in col.lower()
    )
    and "lag" not in col.lower()
    and "rolling" not in col.lower()
]

print("\n" + "-" * 70)
print("WATER LEVEL SOURCE COLUMN CANDIDATES")
print("-" * 70)

for col in water_candidates:
    print("-", col)

# ----------------------------------------------------------------------
# 4. Automatically identify rolling columns
# ----------------------------------------------------------------------

rolling_cols = [
    col for col in df.columns
    if "rolling" in col.lower()
]

print("\n" + "-" * 70)
print("ROLLING FEATURE COLUMNS")
print("-" * 70)

for col in rolling_cols:
    print("-", col)

# ----------------------------------------------------------------------
# 5. Safety check
# ----------------------------------------------------------------------

if len(timestamp_candidates) == 0:
    raise ValueError(
        "No timestamp/date column could be detected. "
        "See AVAILABLE COLUMNS above."
    )

if len(water_candidates) == 0:
    raise ValueError(
        "No original water-level column could be detected. "
        "See AVAILABLE COLUMNS above."
    )

if len(rolling_cols) == 0:
    raise ValueError(
        "No rolling feature columns could be detected."
    )

# ----------------------------------------------------------------------
# 6. Select columns
#
# Prefer exact known project naming if present.
# Otherwise use first detected candidate.
# ----------------------------------------------------------------------

preferred_timestamp = [
    col for col in timestamp_candidates
    if col.lower() in [
        "datetime",
        "timestamp",
        "date_time"
    ]
]

if preferred_timestamp:
    timestamp_col = preferred_timestamp[0]
else:
    timestamp_col = timestamp_candidates[0]

preferred_water = [
    col for col in water_candidates
    if "river water level telemetry hourly" in col.lower()
]

if preferred_water:
    water_col = preferred_water[0]
else:
    water_col = water_candidates[0]

print("\n" + "=" * 70)
print("SELECTED SOURCE COLUMNS")
print("=" * 70)

print("Timestamp column :", timestamp_col)
print("Water level      :", water_col)

# ----------------------------------------------------------------------
# 7. Create diagnostic dataframe
# ----------------------------------------------------------------------

diag = df.copy()

diag[timestamp_col] = pd.to_datetime(
    diag[timestamp_col],
    errors="coerce"
)

diag = (
    diag
    .sort_values(timestamp_col)
    .reset_index(drop=True)
)

# ----------------------------------------------------------------------
# 8. Timestamp validity
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("TIMESTAMP QUALITY")
print("=" * 70)

print(
    "Invalid timestamps:",
    diag[timestamp_col].isna().sum()
)

print(
    "Duplicate timestamps:",
    diag[timestamp_col].duplicated().sum()
)

# ----------------------------------------------------------------------
# 9. Calculate timestamp gaps
# ----------------------------------------------------------------------

diag["Diagnostic_TimeGap"] = (
    diag[timestamp_col].diff()
)

print("\n" + "-" * 70)
print("TIMESTAMP GAP DISTRIBUTION")
print("-" * 70)

print(
    diag["Diagnostic_TimeGap"]
    .value_counts()
    .head(15)
)

# ----------------------------------------------------------------------
# 10. Water-level missingness
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("WATER LEVEL QUALITY")
print("=" * 70)

print(
    "Water-level NaN:",
    diag[water_col].isna().sum()
)

print(
    "Water-level valid:",
    diag[water_col].notna().sum()
)

# ----------------------------------------------------------------------
# 11. Rolling feature NaN counts
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("ROLLING FEATURE NaN COUNTS")
print("=" * 70)

for col in rolling_cols:
    print(
        f"{col}: {diag[col].isna().sum()}"
    )

# ======================================================================
# 12. RECREATE ROLLING FEATURES USING ROW-BASED WINDOWS
# ======================================================================

print("\n" + "=" * 70)
print("ROW-BASED ROLLING RECREATION")
print("=" * 70)

diag["Diagnostic_Mean_6rows"] = (
    diag[water_col]
    .rolling(
        window=7,
        min_periods=7
    )
    .mean()
)

diag["Diagnostic_Mean_12rows"] = (
    diag[water_col]
    .rolling(
        window=13,
        min_periods=13
    )
    .mean()
)

diag["Diagnostic_Mean_24rows"] = (
    diag[water_col]
    .rolling(
        window=25,
        min_periods=25
    )
    .mean()
)

diag["Diagnostic_Max_24rows"] = (
    diag[water_col]
    .rolling(
        window=25,
        min_periods=25
    )
    .max()
)

diag["Diagnostic_Min_24rows"] = (
    diag[water_col]
    .rolling(
        window=25,
        min_periods=25
    )
    .min()
)

# ----------------------------------------------------------------------
# 13. Define comparisons
# ----------------------------------------------------------------------

comparison_map = {}

for col in rolling_cols:

    lower = col.lower()

    if "rollingmean_6h" in lower:
        comparison_map[col] = "Diagnostic_Mean_6rows"

    elif "rollingmean_12h" in lower:
        comparison_map[col] = "Diagnostic_Mean_12rows"

    elif "rollingmean_24h" in lower:
        comparison_map[col] = "Diagnostic_Mean_24rows"

    elif "rollingmax_24h" in lower:
        comparison_map[col] = "Diagnostic_Max_24rows"

    elif "rollingmin_24h" in lower:
        comparison_map[col] = "Diagnostic_Min_24rows"

# ----------------------------------------------------------------------
# 14. Compare original vs recreated
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("ROW-BASED RECREATION VALIDATION")
print("=" * 70)

validation_results = []

for original, recreated in comparison_map.items():

    valid = (
        diag[original].notna()
        &
        diag[recreated].notna()
    )

    checked = int(valid.sum())

    if checked == 0:

        print(f"\n{original}")
        print("No comparable rows.")

        validation_results.append({
            "Feature": original,
            "Checked": 0,
            "Matching": 0,
            "Mismatching": 0,
            "MatchRate": 0.0
        })

        continue

    difference = np.abs(
        diag.loc[valid, original]
        -
        diag.loc[valid, recreated]
    )

    matches = difference <= 1e-6

    matching = int(matches.sum())
    mismatching = int((~matches).sum())
    match_rate = matching / checked * 100

    validation_results.append({
        "Feature": original,
        "Checked": checked,
        "Matching": matching,
        "Mismatching": mismatching,
        "MatchRate": match_rate
    })

    print(f"\n{original}")
    print("-" * 50)
    print("Checked rows :", checked)
    print("Matching     :", matching)
    print("Mismatching  :", mismatching)
    print(f"Match rate   : {match_rate:.2f}%")
    print(
        f"Maximum diff : {difference.max():.10f}"
    )

# ----------------------------------------------------------------------
# 15. Validation summary table
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("VALIDATION SUMMARY")
print("=" * 70)

validation_df = pd.DataFrame(validation_results)

print(
    validation_df.to_string(index=False)
)

# ======================================================================
# 16. FIRST MISMATCHING EXAMPLES — ROLLING MEAN 6H
# ======================================================================

target_feature = None

for col in rolling_cols:
    if "rollingmean_6h" in col.lower():
        target_feature = col
        break

if target_feature is not None:

    print("\n" + "=" * 70)
    print("FIRST MISMATCHING EXAMPLES — ROLLING MEAN 6H")
    print("=" * 70)

    recreated_col = "Diagnostic_Mean_6rows"

    valid = (
        diag[target_feature].notna()
        &
        diag[recreated_col].notna()
    )

    difference = np.abs(
        diag[target_feature]
        -
        diag[recreated_col]
    )

    mismatch = (
        valid
        &
        (difference > 1e-6)
    )

    example_cols = [
        timestamp_col,
        "Diagnostic_TimeGap",
        water_col,
        target_feature,
        recreated_col
    ]

    print(
        diag.loc[
            mismatch,
            example_cols
        ]
        .head(20)
        .to_string(index=False)
    )

# ======================================================================
# 17. CHECK WHETHER MISMATCHES OCCUR AROUND TIME GAPS
# ======================================================================

print("\n" + "=" * 70)
print("MISMATCH + TIME-GAP DIAGNOSTIC")
print("=" * 70)

for original, recreated in comparison_map.items():

    valid = (
        diag[original].notna()
        &
        diag[recreated].notna()
    )

    difference = np.abs(
        diag[original]
        -
        diag[recreated]
    )

    mismatch = (
        valid
        &
        (difference > 1e-6)
    )

    mismatch_rows = diag.loc[mismatch]

    if len(mismatch_rows) == 0:
        print(f"\n{original}: No mismatches.")
        continue

    gap_rows = mismatch_rows[
        mismatch_rows["Diagnostic_TimeGap"] != pd.Timedelta(hours=1)
    ]

    print(f"\n{original}")
    print("Total mismatches              :", len(mismatch_rows))
    print("Mismatches near non-1h gaps   :", len(gap_rows))

# ======================================================================
# 18. FINAL DIAGNOSTIC STATUS
# ======================================================================

print("\n" + "=" * 70)
print("STEP 4.1 COMPLETE")
print("=" * 70)

print("\nIMPORTANT:")
print("This step ONLY diagnoses the rolling features.")
print("No existing rolling feature was modified.")
print("No rows were deleted.")
print("No model training was performed.")

print("\nWait for the diagnostic output before making any changes.")
```

```python
# ======================================================================
# STEP 4.3 — TIME-BASED ROLLING VALIDATION
# ======================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 4.3 — TIME-BASED ROLLING VALIDATION")
print("=" * 70)

# ----------------------------------------------------------------------
# SOURCE COLUMNS
# ----------------------------------------------------------------------

time_col = "Data Acquisition Time"
water_col = "Current_Water_Level"

diag = df.copy()

diag[time_col] = pd.to_datetime(
    diag[time_col],
    errors="coerce"
)

diag = (
    diag
    .sort_values(time_col)
    .reset_index(drop=True)
)

# ----------------------------------------------------------------------
# Create timestamp index
# ----------------------------------------------------------------------

time_indexed = (
    diag
    .set_index(time_col)
)

# ----------------------------------------------------------------------
# TIME-BASED ROLLING FEATURES
#
# closed="both" means:
# previous N hours + current timestamp
# ----------------------------------------------------------------------

diag["TimeMean_6h"] = (
    time_indexed[water_col]
    .rolling(
        "6h",
        closed="both",
        min_periods=1
    )
    .mean()
    .to_numpy()
)

diag["TimeMean_12h"] = (
    time_indexed[water_col]
    .rolling(
        "12h",
        closed="both",
        min_periods=1
    )
    .mean()
    .to_numpy()
)

diag["TimeMean_24h"] = (
    time_indexed[water_col]
    .rolling(
        "24h",
        closed="both",
        min_periods=1
    )
    .mean()
    .to_numpy()
)

diag["TimeMax_24h"] = (
    time_indexed[water_col]
    .rolling(
        "24h",
        closed="both",
        min_periods=1
    )
    .max()
    .to_numpy()
)

diag["TimeMin_24h"] = (
    time_indexed[water_col]
    .rolling(
        "24h",
        closed="both",
        min_periods=1
    )
    .min()
    .to_numpy()
)

# ----------------------------------------------------------------------
# Validation function
# ----------------------------------------------------------------------

def validate(original_col, calculated_col):

    valid = (
        diag[original_col].notna()
        &
        diag[calculated_col].notna()
    )

    checked = int(valid.sum())

    if checked == 0:
        return 0, 0, 0, 0

    diff = np.abs(
        diag.loc[valid, original_col]
        -
        diag.loc[valid, calculated_col]
    )

    matches = diff <= 1e-6

    return (
        checked,
        int(matches.sum()),
        int((~matches).sum()),
        float(matches.mean() * 100)
    )

# ----------------------------------------------------------------------
# Validate all features
# ----------------------------------------------------------------------

tests = [
    (
        "WaterLevel_RollingMean_6h",
        "TimeMean_6h"
    ),
    (
        "WaterLevel_RollingMean_12h",
        "TimeMean_12h"
    ),
    (
        "WaterLevel_RollingMean_24h",
        "TimeMean_24h"
    ),
    (
        "WaterLevel_RollingMax_24h",
        "TimeMax_24h"
    ),
    (
        "WaterLevel_RollingMin_24h",
        "TimeMin_24h"
    )
]

print("\n" + "=" * 70)
print("TIME-BASED VALIDATION RESULTS")
print("=" * 70)

for original, calculated in tests:

    checked, matching, mismatching, rate = validate(
        original,
        calculated
    )

    print(f"\n{original}")
    print("-" * 55)
    print("Checked rows :", checked)
    print("Matching     :", matching)
    print("Mismatching  :", mismatching)
    print(f"Match rate   : {rate:.2f}%")

# ----------------------------------------------------------------------
# Detailed example: 2022-01-04 16:00
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("EXACT TIME-WINDOW EXAMPLE — 2022-01-04 16:00")
print("=" * 70)

target_time = pd.Timestamp(
    "2022-01-04 16:00:00"
)

example = diag[
    diag[time_col] == target_time
]

if len(example) > 0:

    cols = [
        time_col,
        water_col,
        "WaterLevel_RollingMean_6h",
        "TimeMean_6h",
        "WaterLevel_RollingMean_12h",
        "TimeMean_12h",
        "WaterLevel_RollingMean_24h",
        "TimeMean_24h"
    ]

    print(
        example[cols]
        .to_string(index=False)
    )

# ----------------------------------------------------------------------
# Show all records inside the previous 6-hour window
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("RECORDS INSIDE PREVIOUS 6 HOURS")
print("=" * 70)

window_start = target_time - pd.Timedelta(hours=6)

window_rows = diag[
    (diag[time_col] >= window_start)
    &
    (diag[time_col] <= target_time)
]

print(
    window_rows[
        [time_col, water_col]
    ].to_string(index=False)
)

# ----------------------------------------------------------------------
# Manual mean
# ----------------------------------------------------------------------

if len(window_rows) > 0:

    manual_mean = window_rows[water_col].mean()

    print("\nManual 6-hour mean:", manual_mean)

# ======================================================================
# STEP 4.3 COMPLETE
# ======================================================================

print("\n" + "=" * 70)
print("STEP 4.3 COMPLETE")
print("=" * 70)

print("\nNo existing rolling feature was modified.")
print("No rows were deleted.")
print("No model was trained.")
```

```python
# ======================================================================
# STEP 4.4 — ROLLING WINDOW FORENSIC VALIDATION
# ======================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 4.4 — ROLLING WINDOW FORENSIC VALIDATION")
print("=" * 70)

# ----------------------------------------------------------------------
# SOURCE COLUMNS
# ----------------------------------------------------------------------

time_col = "Data Acquisition Time"
water_col = "Current_Water_Level"

diag = df.copy()

diag[time_col] = pd.to_datetime(
    diag[time_col],
    errors="coerce"
)

diag = (
    diag
    .sort_values(time_col)
    .reset_index(drop=True)
)

# ----------------------------------------------------------------------
# Helper function
# ----------------------------------------------------------------------

def match_rate(original, calculated):

    valid = (
        diag[original].notna()
        &
        calculated.notna()
    )

    if valid.sum() == 0:
        return 0.0

    diff = np.abs(
        diag.loc[valid, original]
        -
        calculated.loc[valid]
    )

    return float((diff <= 1e-6).mean() * 100)


# ======================================================================
# PART A — TEST TIME-BASED WINDOWS WITH ALL CLOSED OPTIONS
# ======================================================================

print("\n" + "=" * 70)
print("PART A — TIME-BASED WINDOW / CLOSED-BOUNDARY TEST")
print("=" * 70)

time_indexed = diag.set_index(time_col)

closed_options = [
    "right",
    "left",
    "both",
    "neither"
]

features = {
    "6h": "WaterLevel_RollingMean_6h",
    "12h": "WaterLevel_RollingMean_12h",
    "24h": "WaterLevel_RollingMean_24h"
}

results = []

for window_name, original_col in features.items():

    for closed in closed_options:

        calculated = (
            time_indexed[water_col]
            .rolling(
                window_name,
                closed=closed,
                min_periods=1
            )
            .mean()
        )

        rate = match_rate(
            original_col,
            calculated
        )

        results.append({
            "Feature": original_col,
            "Window": window_name,
            "Closed": closed,
            "MatchRate": rate
        })

        print(
            f"{original_col:35s} | "
            f"{window_name:4s} | "
            f"closed={closed:7s} | "
            f"{rate:6.2f}%"
        )


# ======================================================================
# PART B — TEST DIFFERENT ROW WINDOWS
# ======================================================================

print("\n" + "=" * 70)
print("PART B — ROW WINDOW TEST")
print("=" * 70)

row_tests = [
    ("WaterLevel_RollingMean_6h", "mean", range(4, 10)),
    ("WaterLevel_RollingMean_12h", "mean", range(9, 16)),
    ("WaterLevel_RollingMean_24h", "mean", range(20, 28)),
]

row_results = []

for original_col, operation, windows in row_tests:

    print(f"\n{original_col}")
    print("-" * 60)

    for w in windows:

        if operation == "mean":
            calculated = (
                diag[water_col]
                .rolling(
                    window=w,
                    min_periods=w
                )
                .mean()
            )

        rate = match_rate(
            original_col,
            calculated
        )

        row_results.append({
            "Feature": original_col,
            "Rows": w,
            "MatchRate": rate
        })

        print(
            f"Window = {w:2d} rows"
            f"   Match = {rate:6.2f}%"
        )


# ======================================================================
# PART C — TEST POSSIBLE SHIFTING
# ======================================================================

print("\n" + "=" * 70)
print("PART C — POSSIBLE FEATURE SHIFT TEST")
print("=" * 70)

shift_results = []

for window_name, original_col in features.items():

    calculated = (
        time_indexed[water_col]
        .rolling(
            window_name,
            closed="both",
            min_periods=1
        )
        .mean()
        .reset_index(drop=True)
    )

    original = diag[original_col]

    for shift in range(-3, 4):

        shifted = calculated.shift(shift)

        rate = match_rate(
            original_col,
            shifted
        )

        shift_results.append({
            "Feature": original_col,
            "Shift": shift,
            "MatchRate": rate
        })

        print(
            f"{original_col:35s} | "
            f"Shift={shift:+d} | "
            f"Match={rate:6.2f}%"
        )


# ======================================================================
# PART D — MAX / MIN WITH CLOSED OPTIONS
# ======================================================================

print("\n" + "=" * 70)
print("PART D — MAX/MIN BOUNDARY TEST")
print("=" * 70)

extreme_features = {
    "WaterLevel_RollingMax_24h": "max",
    "WaterLevel_RollingMin_24h": "min"
}

extreme_results = []

for original_col, operation in extreme_features.items():

    for closed in closed_options:

        if operation == "max":

            calculated = (
                time_indexed[water_col]
                .rolling(
                    "24h",
                    closed=closed,
                    min_periods=1
                )
                .max()
            )

        else:

            calculated = (
                time_indexed[water_col]
                .rolling(
                    "24h",
                    closed=closed,
                    min_periods=1
                )
                .min()
            )

        rate = match_rate(
            original_col,
            calculated
        )

        extreme_results.append({
            "Feature": original_col,
            "Operation": operation,
            "Closed": closed,
            "MatchRate": rate
        })

        print(
            f"{original_col:35s} | "
            f"closed={closed:7s} | "
            f"{rate:6.2f}%"
        )


# ======================================================================
# PART E — BEST CANDIDATE FOR EACH FEATURE
# ======================================================================

print("\n" + "=" * 70)
print("BEST TIME-BASED CANDIDATES")
print("=" * 70)

results_df = pd.DataFrame(results)

for feature in results_df["Feature"].unique():

    subset = (
        results_df[
            results_df["Feature"] == feature
        ]
        .sort_values(
            "MatchRate",
            ascending=False
        )
    )

    best = subset.iloc[0]

    print(
        f"\n{feature}"
    )
    print(
        f"Best window : {best['Window']}"
    )
    print(
        f"Closed      : {best['Closed']}"
    )
    print(
        f"Match rate  : {best['MatchRate']:.2f}%"
    )


# ======================================================================
# PART F — BEST ROW-BASED CANDIDATES
# ======================================================================

print("\n" + "=" * 70)
print("BEST ROW-BASED CANDIDATES")
print("=" * 70)

row_results_df = pd.DataFrame(row_results)

for feature in row_results_df["Feature"].unique():

    subset = (
        row_results_df[
            row_results_df["Feature"] == feature
        ]
        .sort_values(
            "MatchRate",
            ascending=False
        )
    )

    best = subset.iloc[0]

    print(
        f"\n{feature}"
    )

    print(
        f"Best rows   : {int(best['Rows'])}"
    )

    print(
        f"Match rate  : {best['MatchRate']:.2f}%"
    )


# ======================================================================
# PART G — BEST SHIFT CANDIDATES
# ======================================================================

print("\n" + "=" * 70)
print("BEST SHIFT CANDIDATES")
print("=" * 70)

shift_results_df = pd.DataFrame(shift_results)

for feature in shift_results_df["Feature"].unique():

    subset = (
        shift_results_df[
            shift_results_df["Feature"] == feature
        ]
        .sort_values(
            "MatchRate",
            ascending=False
        )
    )

    best = subset.iloc[0]

    print(
        f"\n{feature}"
    )

    print(
        f"Best shift  : {int(best['Shift']):+d}"
    )

    print(
        f"Match rate  : {best['MatchRate']:.2f}%"
    )


# ======================================================================
# PART H — EXACT EXAMPLE AGAIN
# ======================================================================

print("\n" + "=" * 70)
print("REFERENCE EXAMPLE — 2022-01-04 16:00")
print("=" * 70)

target_time = pd.Timestamp(
    "2022-01-04 16:00:00"
)

example = diag[
    diag[time_col] == target_time
]

if len(example) > 0:

    print(
        example[
            [
                time_col,
                water_col,
                "WaterLevel_RollingMean_6h",
                "WaterLevel_RollingMean_12h",
                "WaterLevel_RollingMean_24h",
                "WaterLevel_RollingMax_24h",
                "WaterLevel_RollingMin_24h"
            ]
        ]
        .to_string(index=False)
    )


# ======================================================================
# FINAL STATUS
# ======================================================================

print("\n" + "=" * 70)
print("STEP 4.4 COMPLETE")
print("=" * 70)

print("\nNO DATA WAS MODIFIED.")
print("NO FEATURES WERE REGENERATED.")
print("NO ROWS WERE DELETED.")
print("NO MODEL WAS TRAINED.")

print("\nSend the COMPLETE output before proceeding.")
```

```python
# ================================================================
# STEP 4.10 — STORED 6H ROLLING SEMANTICS FORENSIC TEST
# ================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 4.10 — STORED 6H ROLLING SEMANTICS FORENSIC TEST")
print("=" * 70)

# ---------------------------------------------------------------
# 0. SAFETY COPY
# ---------------------------------------------------------------

df_test = df.copy()

TIME_COL = "Data Acquisition Time"
VALUE_COL = "Current_Water_Level"
FEATURE_COL = "WaterLevel_RollingMean_6h"

df_test[TIME_COL] = pd.to_datetime(df_test[TIME_COL])

print("\nOriginal rows:", len(df_test))
print("Original columns:", len(df_test.columns))


# ---------------------------------------------------------------
# 1. VERIFY REQUIRED COLUMNS
# ---------------------------------------------------------------

required = [TIME_COL, VALUE_COL, FEATURE_COL]

missing = [c for c in required if c not in df_test.columns]

if missing:
    raise ValueError(f"Missing required columns: {missing}")

print("\nRequired columns: PASS")


# ---------------------------------------------------------------
# 2. CHECK TIMESTAMP ORDER + DUPLICATES
# ---------------------------------------------------------------

print("\n" + "-" * 70)
print("TIMESTAMP STRUCTURE")
print("-" * 70)

is_sorted = df_test[TIME_COL].is_monotonic_increasing
duplicate_count = df_test[TIME_COL].duplicated().sum()

print("Already sorted:", is_sorted)
print("Duplicate timestamps:", duplicate_count)

if duplicate_count > 0:
    print("\nDuplicate timestamp examples:")
    print(
        df_test[df_test[TIME_COL].duplicated(keep=False)]
        [[TIME_COL, VALUE_COL]]
        .sort_values(TIME_COL)
        .head(20)
    )


# ---------------------------------------------------------------
# 3. CREATE CLEAN TIME INDEX — WITHOUT MODIFYING ORIGINAL DATA
# ---------------------------------------------------------------

base = (
    df_test[[TIME_COL, VALUE_COL, FEATURE_COL]]
    .copy()
    .sort_values(TIME_COL)
)

base_indexed = base.set_index(TIME_COL)


# ---------------------------------------------------------------
# 4. COMPARISON FUNCTION
# ---------------------------------------------------------------

def compare_candidate(name, candidate):

    comparison = pd.DataFrame({
        "stored": base_indexed[FEATURE_COL],
        "candidate": candidate
    })

    valid = comparison["stored"].notna() & comparison["candidate"].notna()

    if valid.sum() == 0:
        return {
            "Configuration": name,
            "Checked": 0,
            "Matching": 0,
            "Mismatching": 0,
            "Match_Rate_%": 0,
            "Mean_Abs_Difference": np.nan,
            "Median_Abs_Difference": np.nan,
            "Maximum_Difference": np.nan
        }

    diff = (
        comparison.loc[valid, "stored"]
        - comparison.loc[valid, "candidate"]
    ).abs()

    matching = np.isclose(
        comparison.loc[valid, "stored"],
        comparison.loc[valid, "candidate"],
        rtol=1e-10,
        atol=1e-10
    )

    return {
        "Configuration": name,
        "Checked": int(valid.sum()),
        "Matching": int(matching.sum()),
        "Mismatching": int((~matching).sum()),
        "Match_Rate_%": round(matching.mean() * 100, 6),
        "Mean_Abs_Difference": diff.mean(),
        "Median_Abs_Difference": diff.median(),
        "Maximum_Difference": diff.max()
    }


# ---------------------------------------------------------------
# 5. CANDIDATE TESTS
# ---------------------------------------------------------------

results = []


# ===============================================================
# TEST GROUP A — TIME BASED ROLLING
# ===============================================================

print("\n" + "=" * 70)
print("TEST GROUP A — TIME-BASED ROLLING")
print("=" * 70)

for closed in ["right", "left", "both", "neither"]:

    try:
        candidate = (
            base_indexed[VALUE_COL]
            .rolling("6h", closed=closed)
            .mean()
        )

        results.append(
            compare_candidate(
                f"6h / closed={closed}",
                candidate
            )
        )

    except Exception as e:
        print(f"6h / closed={closed} FAILED: {e}")


# ===============================================================
# TEST GROUP B — ROW BASED ROLLING
# ===============================================================

print("\n" + "=" * 70)
print("TEST GROUP B — ROW-BASED ROLLING")
print("=" * 70)

for window in [5, 6, 7]:

    for min_periods in [1, window]:

        candidate = (
            base_indexed[VALUE_COL]
            .rolling(
                window=window,
                min_periods=min_periods
            )
            .mean()
        )

        results.append(
            compare_candidate(
                f"{window} rows / min_periods={min_periods}",
                candidate
            )
        )


# ===============================================================
# TEST GROUP C — TIME BASED + min_periods
# ===============================================================

print("\n" + "=" * 70)
print("TEST GROUP C — TIME-BASED + min_periods")
print("=" * 70)

for closed in ["right", "left", "both", "neither"]:

    for min_periods in [1, 2, 3, 4, 5, 6]:

        try:

            candidate = (
                base_indexed[VALUE_COL]
                .rolling(
                    "6h",
                    closed=closed,
                    min_periods=min_periods
                )
                .mean()
            )

            results.append(
                compare_candidate(
                    f"6h / closed={closed} / min_periods={min_periods}",
                    candidate
                )
            )

        except Exception:
            pass


# ===============================================================
# TEST GROUP D — SHIFTED TIME WINDOWS
# ===============================================================

print("\n" + "=" * 70)
print("TEST GROUP D — SHIFTED / ALTERNATIVE WINDOW ALIGNMENT")
print("=" * 70)

# These test whether the stored feature may represent
# a previous/shifted 6-hour window.

for shift in [-2, -1, 0, 1, 2]:

    candidate = (
        base_indexed[VALUE_COL]
        .rolling("6h", closed="right")
        .mean()
        .shift(shift)
    )

    results.append(
        compare_candidate(
            f"6h / closed=right / shift={shift}",
            candidate
        )
    )


# ---------------------------------------------------------------
# 6. RESULTS TABLE
# ---------------------------------------------------------------

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="Match_Rate_%",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 70)
print("ALL CANDIDATE RESULTS — RANKED")
print("=" * 70)

print(
    results_df.to_string(
        index=False
    )
)


# ---------------------------------------------------------------
# 7. BEST CANDIDATE
# ---------------------------------------------------------------

print("\n" + "=" * 70)
print("BEST CANDIDATE")
print("=" * 70)

best = results_df.iloc[0]

print("Configuration :", best["Configuration"])
print("Checked       :", best["Checked"])
print("Matching      :", best["Matching"])
print("Mismatching   :", best["Mismatching"])
print("Match rate    :", best["Match_Rate_%"], "%")
print("Mean abs diff :", best["Mean_Abs_Difference"])
print("Median diff   :", best["Median_Abs_Difference"])
print("Maximum diff  :", best["Maximum_Difference"])


# ---------------------------------------------------------------
# 8. SHOW WORST MISMATCHES FOR BEST CANDIDATE
# ---------------------------------------------------------------

best_name = best["Configuration"]

if "shift=" in best_name:

    shift_value = int(
        best_name.split("shift=")[1]
    )

    best_candidate = (
        base_indexed[VALUE_COL]
        .rolling("6h", closed="right")
        .mean()
        .shift(shift_value)
    )

elif "min_periods=" in best_name and "6h" in best_name:

    closed_value = best_name.split("closed=")[1].split(" ")[0]
    min_value = int(best_name.split("min_periods=")[1])

    best_candidate = (
        base_indexed[VALUE_COL]
        .rolling(
            "6h",
            closed=closed_value,
            min_periods=min_value
        )
        .mean()
    )

elif "6h / closed=" in best_name:

    closed_value = best_name.split("closed=")[1]

    best_candidate = (
        base_indexed[VALUE_COL]
        .rolling(
            "6h",
            closed=closed_value
        )
        .mean()
    )

else:

    # Row based fallback
    window_value = int(
        best_name.split(" rows")[0]
    )

    min_value = int(
        best_name.split("min_periods=")[1]
    )

    best_candidate = (
        base_indexed[VALUE_COL]
        .rolling(
            window=window_value,
            min_periods=min_value
        )
        .mean()
    )


comparison = pd.DataFrame({
    "Stored": base_indexed[FEATURE_COL],
    "Best_Candidate": best_candidate
})

comparison["Abs_Difference"] = (
    comparison["Stored"]
    - comparison["Best_Candidate"]
).abs()

comparison["Match"] = np.isclose(
    comparison["Stored"],
    comparison["Best_Candidate"],
    rtol=1e-10,
    atol=1e-10
)

worst = (
    comparison[
        comparison["Stored"].notna()
        & comparison["Best_Candidate"].notna()
    ]
    .sort_values(
        "Abs_Difference",
        ascending=False
    )
    .head(20)
)

print("\n" + "=" * 70)
print("TOP 20 WORST MISMATCHES — BEST CANDIDATE")
print("=" * 70)

print(worst.to_string())


# ---------------------------------------------------------------
# 9. KNOWN TIMESTAMP PROOF
# ---------------------------------------------------------------

known_timestamps = [
    "2022-01-04 16:00:00",
    "2022-01-04 17:00:00",
    "2022-01-09 01:00:00",
    "2022-01-10 21:00:00"
]

print("\n" + "=" * 70)
print("KNOWN-TIMESTAMP PROOF — BEST CANDIDATE")
print("=" * 70)

for ts in known_timestamps:

    ts = pd.Timestamp(ts)

    if ts in comparison.index:

        print("\nTimestamp:", ts)
        print(
            "Stored:",
            comparison.loc[ts, "Stored"]
        )
        print(
            "Best candidate:",
            comparison.loc[ts, "Best_Candidate"]
        )
        print(
            "Absolute difference:",
            comparison.loc[ts, "Abs_Difference"]
        )
        print(
            "Exact match:",
            comparison.loc[ts, "Match"]
        )


# ---------------------------------------------------------------
# 10. DATA SAFETY PROOF
# ---------------------------------------------------------------

print("\n" + "=" * 70)
print("DATA SAFETY CHECK")
print("=" * 70)

print("Original df rows after test :", len(df))
print("Original df columns         :", len(df.columns))

print("\nSTEP 4.10 COMPLETE")
print("NO DATA WAS MODIFIED.")
print("NO FEATURES WERE REGENERATED.")
print("NO ROWS WERE DELETED.")
```

```python
# ================================================================
# STEP 4.11 — REMAINING ROLLING FEATURES FORENSIC TEST
# ================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 4.11 — REMAINING ROLLING FEATURES FORENSIC TEST")
print("=" * 70)

TIME_COL = "Data Acquisition Time"
VALUE_COL = "Current_Water_Level"

# ---------------------------------------------------------------
# SAFETY COPY
# ---------------------------------------------------------------

work = df.copy()

work[TIME_COL] = pd.to_datetime(work[TIME_COL])

work = (
    work
    .sort_values(TIME_COL)
    .set_index(TIME_COL)
)

rolling_features = [
    "WaterLevel_RollingMean_12h",
    "WaterLevel_RollingMean_24h",
    "WaterLevel_RollingMax_24h",
    "WaterLevel_RollingMin_24h"
]

print("\nFeatures to validate:")
for feature in rolling_features:
    print(" -", feature)


# ---------------------------------------------------------------
# CANDIDATE TEST FUNCTION
# ---------------------------------------------------------------

def test_candidate(
    feature,
    window,
    aggregation,
    closed,
    min_periods
):

    rolling_obj = work[VALUE_COL].rolling(
        window=window,
        closed=closed,
        min_periods=min_periods
    )

    if aggregation == "mean":
        candidate = rolling_obj.mean()

    elif aggregation == "max":
        candidate = rolling_obj.max()

    elif aggregation == "min":
        candidate = rolling_obj.min()

    stored = work[feature]

    valid = stored.notna() & candidate.notna()

    if valid.sum() == 0:
        return None

    stored_v = stored.loc[valid]
    candidate_v = candidate.loc[valid]

    diff = (
        stored_v - candidate_v
    ).abs()

    match = np.isclose(
        stored_v,
        candidate_v,
        rtol=1e-10,
        atol=1e-10
    )

    return {
        "Feature": feature,
        "Window": window,
        "Aggregation": aggregation,
        "Closed": closed,
        "Min_Periods": min_periods,
        "Checked": int(valid.sum()),
        "Matching": int(match.sum()),
        "Mismatching": int((~match).sum()),
        "Match_Rate_%": match.mean() * 100,
        "Mean_Abs_Difference": diff.mean(),
        "Median_Abs_Difference": diff.median(),
        "Maximum_Difference": diff.max()
    }


# ---------------------------------------------------------------
# TEST CANDIDATES
# ---------------------------------------------------------------

results = []

configs = {
    "12h": ["mean"],
    "24h": ["mean", "max", "min"]
}

for feature in rolling_features:

    print("\n" + "-" * 70)
    print(feature)
    print("-" * 70)

    if "12h" in feature:
        window = "12h"
    else:
        window = "24h"

    aggregation = (
        "mean" if "Mean" in feature
        else "max" if "Max" in feature
        else "min"
    )

    # Test the same semantic family that STEP 4.10 proved:
    # time-based rolling + closed=right.
    #
    # We test min_periods from 1 through full window
    # to identify the exact stored semantics.

    hours = int(window.replace("h", ""))

    for min_periods in range(1, hours + 1):

        result = test_candidate(
            feature=feature,
            window=window,
            aggregation=aggregation,
            closed="right",
            min_periods=min_periods
        )

        if result is not None:
            results.append(result)


# ---------------------------------------------------------------
# RESULTS
# ---------------------------------------------------------------

results_df = pd.DataFrame(results)

print("\n" + "=" * 70)
print("ALL CANDIDATES — RANKED")
print("=" * 70)

results_df = results_df.sort_values(
    [
        "Feature",
        "Match_Rate_%",
        "Mean_Abs_Difference"
    ],
    ascending=[
        True,
        False,
        True
    ]
)

print(
    results_df.to_string(index=False)
)


# ---------------------------------------------------------------
# BEST CANDIDATE PER FEATURE
# ---------------------------------------------------------------

print("\n" + "=" * 70)
print("BEST CANDIDATE PER FEATURE")
print("=" * 70)

best = (
    results_df
    .sort_values(
        [
            "Feature",
            "Match_Rate_%",
            "Mean_Abs_Difference"
        ],
        ascending=[
            True,
            False,
            True
        ]
    )
    .groupby("Feature")
    .head(1)
)

print(
    best.to_string(index=False)
)


# ---------------------------------------------------------------
# SAFETY CHECK
# ---------------------------------------------------------------

print("\n" + "=" * 70)
print("DATA SAFETY CHECK")
print("=" * 70)

print("Original df shape :", df.shape)
print("Working shape     :", work.shape)

print("\nSTEP 4.11 COMPLETE")
print("NO DATA WAS MODIFIED.")
print("NO FEATURES WERE REGENERATED.")
print("NO ROWS WERE DELETED.")
```

```python
# ============================================================
# STEP 4.13 — HIGH WATER TARGET RULE FORENSIC ANALYSIS
# ============================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 4.13 — HIGH WATER 6H TARGET RULE FORENSIC ANALYSIS")
print("=" * 70)

test = df.copy()

TIME_COL = "Data Acquisition Time"
VALUE_COL = "Current_Water_Level"
TARGET_COL = "High_Water_6h"

test[TIME_COL] = pd.to_datetime(test[TIME_COL])
test = test.sort_values(TIME_COL).reset_index(drop=True)

# ------------------------------------------------------------
# BASIC TARGET / WATER-LEVEL ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PART A — TARGET GROUP WATER-LEVEL DISTRIBUTION")
print("=" * 70)

for target in [0, 1]:

    subset = test.loc[
        test[TARGET_COL] == target,
        VALUE_COL
    ]

    print(f"\nTarget = {target}")
    print("Count :", len(subset))
    print("Min   :", subset.min())
    print("Q1    :", subset.quantile(0.25))
    print("Median:", subset.median())
    print("Q3    :", subset.quantile(0.75))
    print("Max   :", subset.max())


# ------------------------------------------------------------
# CANDIDATE THRESHOLD TEST
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PART B — CURRENT WATER LEVEL THRESHOLD TEST")
print("=" * 70)

thresholds = np.arange(
    np.floor(test[VALUE_COL].min()),
    np.ceil(test[VALUE_COL].max()) + 0.1,
    0.5
)

threshold_results = []

for threshold in thresholds:

    candidate = (
        test[VALUE_COL] >= threshold
    ).astype(int)

    valid = candidate.notna() & test[TARGET_COL].notna()

    match = (
        candidate[valid].values
        == test.loc[valid, TARGET_COL].values
    )

    threshold_results.append({
        "Threshold": threshold,
        "Matching": int(match.sum()),
        "Mismatching": int((~match).sum()),
        "Match_Rate_%": match.mean() * 100
    })

threshold_df = pd.DataFrame(threshold_results)

print(
    threshold_df
    .sort_values(
        "Match_Rate_%",
        ascending=False
    )
    .head(15)
    .to_string(index=False)
)


# ------------------------------------------------------------
# PART C — TARGET BY WATER-LEVEL BINS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PART C — TARGET RATE BY WATER-LEVEL RANGE")
print("=" * 70)

test["WaterLevel_Bin"] = pd.cut(
    test[VALUE_COL],
    bins=20
)

bin_summary = (
    test
    .groupby("WaterLevel_Bin", observed=False)[TARGET_COL]
    .agg(
        Rows="count",
        Positive_Count="sum",
        Positive_Rate="mean"
    )
)

bin_summary["Positive_Rate_%"] = (
    bin_summary["Positive_Rate"] * 100
)

print(
    bin_summary[
        ["Rows", "Positive_Count", "Positive_Rate_%"]
    ].to_string()
)


# ------------------------------------------------------------
# PART D — HIGH TARGET EXAMPLES
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PART D — SAMPLE TARGET = 1 ROWS")
print("=" * 70)

print(
    test.loc[
        test[TARGET_COL] == 1,
        [
            TIME_COL,
            VALUE_COL,
            TARGET_COL
        ]
    ].head(20).to_string(index=False)
)


# ------------------------------------------------------------
# PART E — LOW TARGET EXAMPLES
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PART E — SAMPLE TARGET = 0 ROWS")
print("=" * 70)

print(
    test.loc[
        test[TARGET_COL] == 0,
        [
            TIME_COL,
            VALUE_COL,
            TARGET_COL
        ]
    ].head(20).to_string(index=False)
)


# ------------------------------------------------------------
# SAFETY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DATA SAFETY CHECK")
print("=" * 70)

print("Original df shape :", df.shape)
print("Test shape        :", test.shape)

print("\nSTEP 4.13 COMPLETE")
print("NO DATA WAS MODIFIED.")
print("NO FEATURES WERE REGENERATED.")
print("NO ROWS WERE DELETED.")
```

```python
# ============================================================
# STEP 4.14 — EXACT HIGH_WATER_6H TARGET SEMANTICS
# CORRECTED VERSION
# ============================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 4.14 — EXACT HIGH_WATER_6H TARGET SEMANTICS")
print("=" * 70)

TIME_COL = "Data Acquisition Time"
VALUE_COL = "Current_Water_Level"
TARGET_COL = "High_Water_6h"
THRESHOLD = 60.0

test = df.copy()

test[TIME_COL] = pd.to_datetime(test[TIME_COL])

test = (
    test
    .sort_values(TIME_COL)
    .reset_index(drop=True)
)

target = test[TARGET_COL].to_numpy(dtype=int)


# ------------------------------------------------------------
# CANDIDATE 1 — CURRENT LEVEL >= 60
# ------------------------------------------------------------

current_high = (
    test[VALUE_COL] >= THRESHOLD
).astype(int).to_numpy()


# ------------------------------------------------------------
# CANDIDATE 2 — NEXT 6 ROWS >= 60
# ------------------------------------------------------------

future_6rows = pd.concat(
    [
        test[VALUE_COL].shift(-i)
        for i in range(1, 7)
    ],
    axis=1
)

future_6rows_high = (
    future_6rows >= THRESHOLD
).any(axis=1).astype(int).to_numpy()


# ------------------------------------------------------------
# CANDIDATE 3 — CURRENT + NEXT 6 ROWS >= 60
# ------------------------------------------------------------

current_plus_6rows_high = (
    pd.concat(
        [
            test[VALUE_COL].shift(-i)
            for i in range(0, 7)
        ],
        axis=1
    )
    .ge(THRESHOLD)
    .any(axis=1)
    .astype(int)
    .to_numpy()
)


# ------------------------------------------------------------
# CANDIDATE 4 — FUTURE 6 HOURS >= 60
# ------------------------------------------------------------

timestamps = test[TIME_COL].to_numpy()
values = test[VALUE_COL].to_numpy()

future_6h_high = np.zeros(len(test), dtype=int)

for i in range(len(test)):

    current_time = timestamps[i]

    mask = (
        (test[TIME_COL] > current_time)
        &
        (
            test[TIME_COL]
            <= current_time + pd.Timedelta(hours=6)
        )
    )

    if mask.any():
        future_6h_high[i] = int(
            (test.loc[mask, VALUE_COL] >= THRESHOLD).any()
        )


# ------------------------------------------------------------
# CANDIDATE 5 — CURRENT + FUTURE 6 HOURS >= 60
# ------------------------------------------------------------

current_plus_future_6h = (
    (
        test[VALUE_COL].to_numpy() >= THRESHOLD
    )
    |
    (future_6h_high == 1)
).astype(int)


# ------------------------------------------------------------
# COMPARE
# ------------------------------------------------------------

candidates = {
    "Current >= 60": current_high,
    "Next 6 rows >= 60": future_6rows_high,
    "Current + next 6 rows >= 60": current_plus_6rows_high,
    "Future 6h >= 60": future_6h_high,
    "Current + future 6h >= 60": current_plus_future_6h
}

results = []

for name, candidate in candidates.items():

    match = candidate == target

    results.append({
        "Candidate": name,
        "Checked": len(test),
        "Matching": int(match.sum()),
        "Mismatching": int((~match).sum()),
        "Match_Rate_%": round(
            match.mean() * 100,
            6
        )
    })

results_df = pd.DataFrame(results)

print("\n" + "=" * 70)
print("TARGET SEMANTICS RESULTS")
print("=" * 70)

print(
    results_df
    .sort_values("Match_Rate_%", ascending=False)
    .to_string(index=False)
)


# ------------------------------------------------------------
# 39 CURRENT-THRESHOLD MISMATCHES
# ------------------------------------------------------------

threshold_mismatch = (
    current_high != target
)

mismatch_view = test.loc[
    threshold_mismatch,
    [
        TIME_COL,
        VALUE_COL,
        TARGET_COL
    ]
].copy()

mismatch_view["Future_6h_High"] = future_6h_high[
    threshold_mismatch
]

print("\n" + "=" * 70)
print("CURRENT >= 60 MISMATCHES")
print("=" * 70)

print(
    mismatch_view.to_string(index=False)
)


# ------------------------------------------------------------
# BEST CANDIDATE
# ------------------------------------------------------------

best_row = results_df.loc[
    results_df["Match_Rate_%"].idxmax()
]

print("\n" + "=" * 70)
print("BEST CANDIDATE")
print("=" * 70)

print(best_row.to_string())


# ------------------------------------------------------------
# SAFETY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DATA SAFETY CHECK")
print("=" * 70)

print("Original df shape :", df.shape)
print("Test shape        :", test.shape)

print("\nSTEP 4.14 COMPLETE")
print("NO DATA WAS MODIFIED.")
print("NO FEATURES WERE REGENERATED.")
print("NO ROWS WERE DELETED.")
```

```python
# =========================================
# STEP 4 — EXACT-TIME LAG FEATURES
# =========================================

print("=" * 70)
print("STEP 4 — EXACT-TIME LAG FEATURES")
print("=" * 70)

# Make sure time is datetime
df[TIME_COL] = pd.to_datetime(df[TIME_COL], errors="coerce")

# Sort chronologically
df = df.sort_values(TIME_COL).reset_index(drop=True)

# Create exact timestamp -> water level lookup
water_series = (
    df.set_index(TIME_COL)["Current_Water_Level"]
    .sort_index()
)

# Required lag periods
lag_hours = [1, 3, 6, 12, 24]

for hours in lag_hours:

    target_times = df[TIME_COL] - pd.Timedelta(hours=hours)

    df[f"WaterLevel_Lag_{hours}h"] = (
        water_series.reindex(target_times).to_numpy()
    )

print("\nCreated lag features:")

for hours in lag_hours:
    col = f"WaterLevel_Lag_{hours}h"
    print(f"✓ {col}")

print("\nLatest rows with lag features:")

print(
    df[
        [
            TIME_COL,
            "Current_Water_Level",
            "WaterLevel_Lag_1h",
            "WaterLevel_Lag_3h",
            "WaterLevel_Lag_6h",
            "WaterLevel_Lag_12h",
            "WaterLevel_Lag_24h"
        ]
    ].tail(10).to_string(index=False)
)

print("\nCurrent shape:", df.shape)
```

```python
# =========================================
# STEP 5 — ROLLING WATER-LEVEL FEATURES
# =========================================

print("=" * 70)
print("STEP 5 — ROLLING WATER-LEVEL FEATURES")
print("=" * 70)

# Make sure data is sorted by exact time
df = df.sort_values(TIME_COL).reset_index(drop=True)

# Rolling windows
df["WaterLevel_RollingMean_6h"] = (
    df["Current_Water_Level"]
    .rolling(window=6, min_periods=6)
    .mean()
)

df["WaterLevel_RollingMean_12h"] = (
    df["Current_Water_Level"]
    .rolling(window=12, min_periods=12)
    .mean()
)

df["WaterLevel_RollingMean_24h"] = (
    df["Current_Water_Level"]
    .rolling(window=24, min_periods=24)
    .mean()
)

df["WaterLevel_RollingMax_24h"] = (
    df["Current_Water_Level"]
    .rolling(window=24, min_periods=24)
    .max()
)

df["WaterLevel_RollingMin_24h"] = (
    df["Current_Water_Level"]
    .rolling(window=24, min_periods=24)
    .min()
)

print("\nCreated rolling features:")

rolling_features = [
    "WaterLevel_RollingMean_6h",
    "WaterLevel_RollingMean_12h",
    "WaterLevel_RollingMean_24h",
    "WaterLevel_RollingMax_24h",
    "WaterLevel_RollingMin_24h"
]

for feature in rolling_features:
    print(f"✓ {feature}")

print("\nLatest rows:")

print(
    df[
        [
            TIME_COL,
            "Current_Water_Level",
            "WaterLevel_RollingMean_6h",
            "WaterLevel_RollingMean_12h",
            "WaterLevel_RollingMean_24h",
            "WaterLevel_RollingMax_24h",
            "WaterLevel_RollingMin_24h"
        ]
    ].tail(10).to_string(index=False)
)

print("\nCurrent shape:", df.shape)

print("\nMissing values in rolling features:")

print(
    df[rolling_features]
    .isna()
    .sum()
)
```

```python
# =========================================
# STEP 6 — WATER-LEVEL CHANGE / TREND FEATURES
# =========================================

print("=" * 70)
print("STEP 6 — WATER-LEVEL CHANGE / TREND FEATURES")
print("=" * 70)

# Exact-time lookup series
water_series = (
    df.set_index(TIME_COL)["Current_Water_Level"]
    .sort_index()
)

change_hours = [1, 3, 6, 12, 24]

for hours in change_hours:

    target_times = df[TIME_COL] - pd.Timedelta(hours=hours)

    past_values = (
        water_series
        .reindex(target_times)
        .to_numpy()
    )

    df[f"WaterLevel_Change_{hours}h"] = (
        df["Current_Water_Level"].to_numpy()
        - past_values
    )

print("\nCreated change/trend features:")

change_features = [
    "WaterLevel_Change_1h",
    "WaterLevel_Change_3h",
    "WaterLevel_Change_6h",
    "WaterLevel_Change_12h",
    "WaterLevel_Change_24h"
]

for feature in change_features:
    print(f"✓ {feature}")

print("\nLatest rows:")

print(
    df[
        [
            TIME_COL,
            "Current_Water_Level",
            "WaterLevel_Change_1h",
            "WaterLevel_Change_3h",
            "WaterLevel_Change_6h",
            "WaterLevel_Change_12h",
            "WaterLevel_Change_24h"
        ]
    ].tail(10).to_string(index=False)
)

print("\nCurrent shape:", df.shape)

print("\nMissing values in change features:")

print(
    df[change_features]
    .isna()
    .sum()
)
```
