```python
# ================================================================
# STEP 51 — DISCHARGE API DISCOVERY
# NWDP Dataset API
# ================================================================

import requests
import pandas as pd
import json

print("=" * 70)
print("STEP 51 — DISCHARGE API DISCOVERY")
print("=" * 70)

# Official NWDP Dataset API page
NWDP_API_PAGE = (
    "https://www.nwdp.nwic.gov.in/dataset_api/home_api_page"
)

print("\nNWDP API PAGE")
print(NWDP_API_PAGE)

# ---------------------------------------------------------------
# Search the NWDP API catalogue page
# ---------------------------------------------------------------

response = requests.get(
    NWDP_API_PAGE,
    timeout=30
)

print("\nHTTP STATUS:", response.status_code)

if response.status_code != 200:
    raise RuntimeError(
        f"NWDP API page could not be accessed. "
        f"HTTP status = {response.status_code}"
    )

html = response.text

print("NWDP API page successfully loaded.")
print("Page size:", len(html), "characters")

# ---------------------------------------------------------------
# Look for discharge-related API information
# ---------------------------------------------------------------

keywords = [
    "discharge",
    "river discharge",
    "telemetry",
    "hourly",
    "central water commission",
    "cwc"
]

html_lower = html.lower()

print("\nKEYWORD CHECK")
print("-" * 70)

for keyword in keywords:
    count = html_lower.count(keyword.lower())

    print(f"{keyword:<35} : {count}")

# ---------------------------------------------------------------
# Save the catalogue page locally for inspection
# ---------------------------------------------------------------

catalogue_file = "nwdp_discharge_api_catalogue.html"

with open(
    catalogue_file,
    "w",
    encoding="utf-8"
) as f:
    f.write(html)

print("\nCatalogue page saved as:")
print(catalogue_file)

# ---------------------------------------------------------------
# Final status
# ---------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 51 COMPLETED")
print("=" * 70)

print(
    "\nNext step: identify the exact CWC Assam discharge API/resource."
)
```

```python
# ================================================================
# STEP 52 — FIND NWDP API ENDPOINT
# ================================================================

import re
import requests

print("=" * 70)
print("STEP 52 — FIND NWDP API ENDPOINT")
print("=" * 70)

html = response.text

# ---------------------------------------------------------------
# Find URLs present in the NWDP API page
# ---------------------------------------------------------------

urls = re.findall(
    r'https?://[^"\'>\s]+',
    html
)

urls = list(dict.fromkeys(urls))

print("\nURLs found in NWDP API page:")
print("-" * 70)

for url in urls:
    print(url)

# ---------------------------------------------------------------
# Find JavaScript files
# ---------------------------------------------------------------

scripts = re.findall(
    r'<script[^>]+src=["\']([^"\']+)["\']',
    html,
    flags=re.IGNORECASE
)

scripts = list(dict.fromkeys(scripts))

print("\nJavaScript files found:")
print("-" * 70)

for script in scripts:
    print(script)

# ---------------------------------------------------------------
# Search page source for API-related strings
# ---------------------------------------------------------------

api_terms = [
    "api/",
    "/api",
    "ajax",
    "fetch(",
    "$.ajax",
    "$.get",
    "axios",
    "dataset_api",
    "api_url",
    "apiurl"
]

print("\nAPI-related source patterns:")
print("-" * 70)

for term in api_terms:
    count = html.lower().count(term.lower())
    print(f"{term:<20}: {count}")

print("\n" + "=" * 70)
print("STEP 52 COMPLETED")
print("=" * 70)
```

```python
# ================================================================
# STEP 53 — READ NWDP API PAGE JAVASCRIPT
# ================================================================

import requests
import re

print("=" * 70)
print("STEP 53 — READ NWDP API PAGE JAVASCRIPT")
print("=" * 70)

JS_URL = "https://www.nwdp.nwic.gov.in/js/home_api_page.js"

js_response = requests.get(
    JS_URL,
    timeout=30
)

print("\nJavaScript URL:")
print(JS_URL)

print("\nHTTP STATUS:", js_response.status_code)

if js_response.status_code != 200:
    raise RuntimeError(
        f"Could not load NWDP JavaScript. "
        f"HTTP status = {js_response.status_code}"
    )

js = js_response.text

print("JavaScript loaded successfully.")
print("JavaScript size:", len(js), "characters")

# ---------------------------------------------------------------
# Print API-related lines
# ---------------------------------------------------------------

print("\nAPI-RELATED LINES")
print("-" * 70)

lines = js.splitlines()

found = 0

for i, line in enumerate(lines, start=1):

    line_lower = line.lower()

    if any(term in line_lower for term in [
        "api",
        "dataset",
        "ajax",
        "url",
        "endpoint",
        "fetch",
        "getjson",
        "post"
    ]):

        print(f"{i}: {line.strip()}")
        found += 1

print("\nTotal relevant lines found:", found)

# ---------------------------------------------------------------
# Extract URLs / paths from JavaScript
# ---------------------------------------------------------------

print("\nPOSSIBLE API PATHS / URLS")
print("-" * 70)

patterns = re.findall(
    r'["\']([^"\']*(?:api|dataset)[^"\']*)["\']',
    js,
    flags=re.IGNORECASE
)

patterns = list(dict.fromkeys(patterns))

for item in patterns:
    print(item)

# ---------------------------------------------------------------
# Save JavaScript for inspection
# ---------------------------------------------------------------

with open(
    "nwdp_home_api_page.js",
    "w",
    encoding="utf-8"
) as f:
    f.write(js)

print("\nJavaScript saved as:")
print("nwdp_home_api_page.js")

print("\n" + "=" * 70)
print("STEP 53 COMPLETED")
print("=" * 70)
```

```python
# ================================================================
# STEP 54 — FIND EXACT CWC DISCHARGE DATASET
# ================================================================

import requests
import pandas as pd

print("=" * 70)
print("STEP 54 — FIND EXACT CWC DISCHARGE DATASET")
print("=" * 70)

BASE_URL = "https://www.nwdp.nwic.gov.in"

API_URL = f"{BASE_URL}/api/3/action/package_search"

# Search specifically for CWC hourly river discharge
params = {
    "q": "River Discharge Telemetry Hourly Central Water Commission CWC",
    "rows": 100
}

response = requests.get(
    API_URL,
    params=params,
    timeout=30
)

print("\nHTTP STATUS:", response.status_code)

if response.status_code != 200:
    raise RuntimeError(
        f"NWDP CKAN API failed. HTTP status = {response.status_code}"
    )

data = response.json()

if not data.get("success", False):
    raise RuntimeError("NWDP API returned success=False")

results = data["result"]["results"]

print("Total matching datasets:", len(results))

print("\n" + "-" * 70)
print("MATCHING DATASETS")
print("-" * 70)

for i, dataset in enumerate(results, start=1):

    print(f"\nDATASET {i}")
    print("Title :", dataset.get("title"))
    print("ID    :", dataset.get("id"))
    print("Name  :", dataset.get("name"))
    print("Org   :", dataset.get("organization", {}).get("title"))

    resources = dataset.get("resources", [])

    print("Resources:", len(resources))

    for j, resource in enumerate(resources, start=1):

        print(
            f"  Resource {j}: "
            f"{resource.get('name')} | "
            f"format={resource.get('format')} | "
            f"id={resource.get('id')}"
        )

print("\n" + "=" * 70)
print("STEP 54 COMPLETED")
print("=" * 70)
```
