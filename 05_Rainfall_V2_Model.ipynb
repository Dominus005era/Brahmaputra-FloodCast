```python
# ==============================================================
# V1 TARGET vs EXACT +6H FUTURE WATER LEVEL
# ==============================================================

import pandas as pd
import numpy as np

V1_PATH = "data/cleaned/final_ml_dataset.csv"
WATER_PATH = "data/cleaned/darrang_fakipara_cleaned.csv"

# --------------------------------------------------------------
# Load V1
# --------------------------------------------------------------

v1 = pd.read_csv(V1_PATH)

v1["Data Acquisition Time"] = pd.to_datetime(
    v1["Data Acquisition Time"]
)

v1 = (
    v1[
        [
            "Data Acquisition Time",
            "High_Water_6h"
        ]
    ]
    .drop_duplicates("Data Acquisition Time")
    .sort_values("Data Acquisition Time")
    .reset_index(drop=True)
)

# --------------------------------------------------------------
# Load original water data
# --------------------------------------------------------------

water = pd.read_csv(WATER_PATH)

water["Data Acquisition Time"] = pd.to_datetime(
    water["Data Acquisition Time"],
    errors="coerce"
)

water = (
    water[
        [
            "Data Acquisition Time",
            "River Water Level Telemetry Hourly (meter)"
        ]
    ]
    .dropna(subset=["Data Acquisition Time"])
    .drop_duplicates("Data Acquisition Time")
    .sort_values("Data Acquisition Time")
    .reset_index(drop=True)
)

# --------------------------------------------------------------
# Exact t + 6 hour timestamp
# --------------------------------------------------------------

v1["Future_Time_6h"] = (
    v1["Data Acquisition Time"]
    + pd.Timedelta(hours=6)
)

# --------------------------------------------------------------
# Lookup exact future water level
# --------------------------------------------------------------

future_lookup = water.rename(
    columns={
        "Data Acquisition Time": "Future_Time_6h",
        "River Water Level Telemetry Hourly (meter)":
            "Future_Water_Level_6h"
    }
)

audit = v1.merge(
    future_lookup,
    on="Future_Time_6h",
    how="left"
)

# --------------------------------------------------------------
# Results
# --------------------------------------------------------------

print("=" * 70)
print("V1 TARGET vs EXACT +6H FUTURE WATER LEVEL")
print("=" * 70)

print("V1 rows:", len(v1))

print(
    "Exact +6h future water-level matches:",
    audit["Future_Water_Level_6h"].notna().sum()
)

print(
    "Missing +6h future water-level matches:",
    audit["Future_Water_Level_6h"].isna().sum()
)

print("\nFuture water level by target:")
print(
    audit
    .groupby("High_Water_6h")["Future_Water_Level_6h"]
    .describe()
)
```

```python
# ==============================================================
# VERIFY V1 TARGET THRESHOLD
# ==============================================================

THRESHOLD = 60.0

audit["Expected_Target_60m"] = (
    audit["Future_Water_Level_6h"] >= THRESHOLD
).astype(float)

audit["Target_Match"] = (
    audit["High_Water_6h"]
    == audit["Expected_Target_60m"]
)

print("=" * 70)
print("V1 TARGET THRESHOLD VERIFICATION")
print("=" * 70)

print("Threshold tested:", THRESHOLD)

print("\nTotal rows:", len(audit))

print(
    "Matching rows:",
    audit["Target_Match"].sum()
)

print(
    "Mismatching rows:",
    (~audit["Target_Match"]).sum()
)

print(
    "\nMatch percentage:",
    audit["Target_Match"].mean() * 100
)

print("\nMismatching examples:")

print(
    audit.loc[
        ~audit["Target_Match"],
        [
            "Data Acquisition Time",
            "Future_Time_6h",
            "Future_Water_Level_6h",
            "High_Water_6h",
            "Expected_Target_60m"
        ]
    ].head(20)
)
```

```python
import pandas as pd
import numpy as np

# ==============================================================
# V2 TARGET GENERATION
# SAME VERIFIED V1 TARGET DEFINITION
# ==============================================================

INPUT_PATH = "data/cleaned/water_rainfall_integrated_clean.csv"

TIME_COL = "Data Acquisition Time"
WATER_COL = "River Water Level Telemetry Hourly (meter)"

TARGET = "High_Water_6h"

THRESHOLD = 60.0
FORECAST_HORIZON = pd.Timedelta(hours=6)

# --------------------------------------------------------------
# Load integrated water + rainfall dataset
# --------------------------------------------------------------

v2 = pd.read_csv(INPUT_PATH)

# --------------------------------------------------------------
# Timestamp
# --------------------------------------------------------------

v2[TIME_COL] = pd.to_datetime(
    v2[TIME_COL],
    errors="coerce"
)

# --------------------------------------------------------------
# Water level numeric
# --------------------------------------------------------------

v2[WATER_COL] = pd.to_numeric(
    v2[WATER_COL],
    errors="coerce"
)

# --------------------------------------------------------------
# Sort chronologically
# --------------------------------------------------------------

v2 = (
    v2
    .dropna(subset=[TIME_COL])
    .sort_values(TIME_COL)
    .reset_index(drop=True)
)

print("=" * 70)
print("V2 BASE DATA PREPARED")
print("=" * 70)

print("Shape:", v2.shape)

print(
    "Date range:",
    v2[TIME_COL].min(),
    "→",
    v2[TIME_COL].max()
)

# --------------------------------------------------------------
# Create exact future timestamp
# --------------------------------------------------------------

v2["Future_Time_6h"] = (
    v2[TIME_COL] + FORECAST_HORIZON
)

# --------------------------------------------------------------
# Build future water-level lookup
# --------------------------------------------------------------

future_water = (
    v2[
        [
            TIME_COL,
            WATER_COL
        ]
    ]
    .drop_duplicates(TIME_COL)
    .rename(
        columns={
            TIME_COL: "Future_Time_6h",
            WATER_COL: "Future_Water_Level_6h"
        }
    )
)

# --------------------------------------------------------------
# Match exact t+6h water level
# --------------------------------------------------------------

v2 = v2.merge(
    future_water,
    on="Future_Time_6h",
    how="left"
)

# --------------------------------------------------------------
# Generate SAME verified target
# --------------------------------------------------------------

v2[TARGET] = np.where(
    v2["Future_Water_Level_6h"].notna(),
    (
        v2["Future_Water_Level_6h"] >= THRESHOLD
    ).astype(float),
    np.nan
)

# --------------------------------------------------------------
# Audit
# --------------------------------------------------------------

print("\n" + "=" * 70)
print("V2 TARGET GENERATED")
print("=" * 70)

print("Total rows:", len(v2))

print(
    "Exact +6h future matches:",
    v2["Future_Water_Level_6h"].notna().sum()
)

print(
    "Missing +6h future matches:",
    v2["Future_Water_Level_6h"].isna().sum()
)

print("\nTarget distribution:")
print(
    v2[TARGET].value_counts(dropna=False)
)

print("\nFuture water-level statistics:")
print(
    v2.groupby(TARGET)["Future_Water_Level_6h"].describe()
)
```

```python
import json

# ==============================================================
# V2 EXISTING SCRIPTS AUDIT2
# PERSON 1 — RAINFALL INTEGRATION
# ==============================================================

NOTEBOOK_PATH = "Person1_Rainfall_Integration.ipynb"

with open(NOTEBOOK_PATH, "r", encoding="utf-8") as f:
    nb = json.load(f)

KEYWORDS = [
    "rainfall_1h_sum",
    "rainfall_3h_sum",
    "rainfall_6h_sum",
    "rainfall_12h_sum",
    "rainfall_24h_sum",
    "resample",
    "merge",
    "groupby",
    "rolling"
]

print("=" * 70)
print("V2 EXISTING SCRIPTS AUDIT2")
print("PERSON 1 — RAINFALL INTEGRATION")
print("=" * 70)

relevant_cells = []

for i, cell in enumerate(nb["cells"]):

    source = "".join(cell.get("source", []))
    source_lower = source.lower()

    matched_keywords = [
        keyword
        for keyword in KEYWORDS
        if keyword in source_lower
    ]

    if matched_keywords:

        relevant_cells.append({
            "cell_index": i,
            "cell_type": cell["cell_type"],
            "keywords": matched_keywords,
            "source": source
        })


# --------------------------------------------------------------
# Summary
# --------------------------------------------------------------

print(f"\nTotal notebook cells    : {len(nb['cells'])}")
print(f"Relevant cells found    : {len(relevant_cells)}")

print("\n" + "=" * 70)
print("RELEVANT CELL SUMMARY")
print("=" * 70)

for item in relevant_cells:

    print(
        f"Cell {item['cell_index']} | "
        f"Type: {item['cell_type']} | "
        f"Keywords: {', '.join(item['keywords'])}"
    )


# --------------------------------------------------------------
# Print relevant cells
# --------------------------------------------------------------

print("\n" + "=" * 70)
print("RELEVANT CELL CODE")
print("=" * 70)

for item in relevant_cells:

    print("\n" + "-" * 70)
    print(
        f"CELL {item['cell_index']} "
        f"| TYPE: {item['cell_type']}"
    )
    print(
        f"KEYWORDS: {', '.join(item['keywords'])}"
    )
    print("-" * 70)

    print(item["source"])
```

```python
# ============================================================
# STEP 4.17 — VERIFIED DATASET LOAD + ROLLING FEATURE AUDIT
# ============================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 4.17 — VERIFIED DATASET LOAD + ROLLING FEATURE AUDIT")
print("=" * 70)


# ------------------------------------------------------------
# 1. LOAD VERIFIED FINAL ML DATASET
# ------------------------------------------------------------

FILE = "data/cleaned/final_ml_dataset.csv"

ml_df = pd.read_csv(FILE)

ml_df["Data Acquisition Time"] = pd.to_datetime(
    ml_df["Data Acquisition Time"],
    errors="coerce"
)

ml_df = (
    ml_df
    .sort_values("Data Acquisition Time")
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 2. BASIC VERIFICATION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("VERIFIED DATASET")
print("=" * 70)

print("File              :", FILE)
print("Shape             :", ml_df.shape)
print(
    "Date range        :",
    ml_df["Data Acquisition Time"].min(),
    "→",
    ml_df["Data Acquisition Time"].max()
)

print(
    "Duplicate times   :",
    ml_df["Data Acquisition Time"].duplicated().sum()
)

print(
    "Sorted            :",
    ml_df["Data Acquisition Time"].is_monotonic_increasing
)


# ------------------------------------------------------------
# 3. SAFETY COPY
# ------------------------------------------------------------

audit = ml_df.copy()

TIME_COL = "Data Acquisition Time"
VALUE_COL = "Current_Water_Level"


# ------------------------------------------------------------
# 4. ROLLING FEATURES TO AUDIT
# ------------------------------------------------------------

rolling_features = [
    "WaterLevel_RollingMean_6h",
    "WaterLevel_RollingMean_12h",
    "WaterLevel_RollingMean_24h",
    "WaterLevel_RollingMax_24h",
    "WaterLevel_RollingMin_24h"
]

print("\nFeatures to audit:")

for feature in rolling_features:
    print(" -", feature)


# ------------------------------------------------------------
# 5. CANDIDATE TEST FUNCTION
# ------------------------------------------------------------

def test_candidate(
    feature,
    window,
    aggregation,
    min_periods
):

    rolling_obj = audit[VALUE_COL].rolling(
        window=window,
        min_periods=min_periods
    )

    if aggregation == "mean":
        candidate = rolling_obj.mean()

    elif aggregation == "max":
        candidate = rolling_obj.max()

    elif aggregation == "min":
        candidate = rolling_obj.min()

    stored = audit[feature]

    valid = (
        stored.notna()
        &
        candidate.notna()
    )

    if valid.sum() == 0:
        return None

    stored_v = stored.loc[valid]
    candidate_v = candidate.loc[valid]

    diff = (
        stored_v - candidate_v
    ).abs()

    match = np.isclose(
        stored_v,
        candidate_v,
        rtol=1e-10,
        atol=1e-10
    )

    return {
        "Feature": feature,
        "Window": window,
        "Aggregation": aggregation,
        "Min_Periods": min_periods,
        "Checked": int(valid.sum()),
        "Matching": int(match.sum()),
        "Mismatching": int((~match).sum()),
        "Match_Rate_%": match.mean() * 100,
        "Mean_Abs_Difference": diff.mean(),
        "Median_Abs_Difference": diff.median(),
        "Maximum_Difference": diff.max()
    }


# ------------------------------------------------------------
# 6. TEST ACTUAL ROW-BASED ROLLING SEMANTICS
# ------------------------------------------------------------

results = []

configs = {
    "WaterLevel_RollingMean_6h":  (6, "mean"),
    "WaterLevel_RollingMean_12h": (12, "mean"),
    "WaterLevel_RollingMean_24h": (24, "mean"),
    "WaterLevel_RollingMax_24h":  (24, "max"),
    "WaterLevel_RollingMin_24h":  (24, "min")
}

for feature, (window, aggregation) in configs.items():

    print("\n" + "-" * 70)
    print(feature)
    print("-" * 70)

    for min_periods in range(1, window + 1):

        result = test_candidate(
            feature=feature,
            window=window,
            aggregation=aggregation,
            min_periods=min_periods
        )

        if result is not None:
            results.append(result)


# ------------------------------------------------------------
# 7. ALL CANDIDATES
# ------------------------------------------------------------

results_df = pd.DataFrame(results)

print("\n" + "=" * 70)
print("ALL CANDIDATES — RANKED")
print("=" * 70)

results_df = results_df.sort_values(
    [
        "Feature",
        "Match_Rate_%",
        "Mean_Abs_Difference"
    ],
    ascending=[
        True,
        False,
        True
    ]
)

print(
    results_df.to_string(index=False)
)


# ------------------------------------------------------------
# 8. BEST CANDIDATE PER FEATURE
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("BEST CANDIDATE PER FEATURE")
print("=" * 70)

best = (
    results_df
    .sort_values(
        [
            "Feature",
            "Match_Rate_%",
            "Mean_Abs_Difference"
        ],
        ascending=[
            True,
            False,
            True
        ]
    )
    .groupby("Feature")
    .head(1)
)

print(
    best.to_string(index=False)
)


# ------------------------------------------------------------
# 9. SAFETY CHECK
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DATA SAFETY CHECK")
print("=" * 70)

print("Original ml_df shape :", ml_df.shape)
print("Audit shape          :", audit.shape)

print("\nNO DATA WAS MODIFIED.")
print("NO FEATURES WERE REGENERATED.")
print("NO ROWS WERE DELETED.")

print("\n" + "=" * 70)
print("STEP 4.17 COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# STEP 4.17 — ACTUAL ROW-BASED ROLLING FEATURE GENERATION AUDIT
# ============================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 4.17 — ACTUAL ROW-BASED ROLLING FEATURE GENERATION AUDIT")
print("=" * 70)


# ------------------------------------------------------------
# 1. LOAD VERIFIED DATASET
# ------------------------------------------------------------

FILE = "data/cleaned/final_ml_dataset.csv"

audit = pd.read_csv(FILE)

TIME_COL = "Data Acquisition Time"
VALUE_COL = "Current_Water_Level"

audit[TIME_COL] = pd.to_datetime(
    audit[TIME_COL],
    errors="coerce"
)

audit = (
    audit
    .sort_values(TIME_COL)
    .reset_index(drop=True)
)

print("\n" + "=" * 70)
print("VERIFIED DATASET")
print("=" * 70)

print("File              :", FILE)
print("Shape             :", audit.shape)
print(
    "Date range        :",
    audit[TIME_COL].min(),
    "→",
    audit[TIME_COL].max()
)

print(
    "Duplicate times   :",
    audit[TIME_COL].duplicated().sum()
)

print(
    "Sorted            :",
    audit[TIME_COL].is_monotonic_increasing
)


# ------------------------------------------------------------
# 2. FEATURES TO AUDIT
# ------------------------------------------------------------

rolling_specs = {
    "WaterLevel_RollingMean_6h":  ("mean", 6),
    "WaterLevel_RollingMean_12h": ("mean", 12),
    "WaterLevel_RollingMean_24h": ("mean", 24),
    "WaterLevel_RollingMax_24h":  ("max", 24),
    "WaterLevel_RollingMin_24h":  ("min", 24)
}

print("\nFeatures to audit:")

for feature in rolling_specs:
    print(" -", feature)


# ------------------------------------------------------------
# 3. ROW-BASED CANDIDATE TEST
# ------------------------------------------------------------

def test_row_based_candidate(
    feature,
    window,
    aggregation,
    min_periods
):

    rolling_obj = (
        audit[VALUE_COL]
        .rolling(
            window=window,
            min_periods=min_periods
        )
    )

    if aggregation == "mean":
        candidate = rolling_obj.mean()

    elif aggregation == "max":
        candidate = rolling_obj.max()

    elif aggregation == "min":
        candidate = rolling_obj.min()

    else:
        raise ValueError(
            f"Unknown aggregation: {aggregation}"
        )

    stored = audit[feature]

    valid = (
        stored.notna()
        &
        candidate.notna()
    )

    if valid.sum() == 0:
        return None

    stored_v = stored.loc[valid]
    candidate_v = candidate.loc[valid]

    match = np.isclose(
        stored_v,
        candidate_v,
        rtol=1e-10,
        atol=1e-10
    )

    diff = (
        stored_v - candidate_v
    ).abs()

    return {
        "Feature": feature,
        "Window_Rows": window,
        "Aggregation": aggregation,
        "Min_Periods": min_periods,
        "Checked": int(valid.sum()),
        "Matching": int(match.sum()),
        "Mismatching": int((~match).sum()),
        "Match_Rate_%": match.mean() * 100,
        "Mean_Abs_Difference": diff.mean(),
        "Median_Abs_Difference": diff.median(),
        "Maximum_Difference": diff.max()
    }


# ------------------------------------------------------------
# 4. TEST ACTUAL ROW-BASED SEMANTICS
# ------------------------------------------------------------

results = []

for feature, (aggregation, window) in rolling_specs.items():

    print("\n" + "-" * 70)
    print(feature)
    print("-" * 70)

    # Test the actual generation rule first:
    # window=N, min_periods=N

    result = test_row_based_candidate(
        feature=feature,
        window=window,
        aggregation=aggregation,
        min_periods=window
    )

    if result is not None:
        results.append(result)


# ------------------------------------------------------------
# 5. RESULTS
# ------------------------------------------------------------

results_df = pd.DataFrame(results)

print("\n" + "=" * 70)
print("ACTUAL ROW-BASED GENERATION RESULTS")
print("=" * 70)

print(
    results_df.to_string(index=False)
)


# ------------------------------------------------------------
# 6. EXACT MATCH CHECK
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("EXACT GENERATION CHECK")
print("=" * 70)

all_exact = True

for _, row in results_df.iterrows():

    feature = row["Feature"]

    exact = (
        row["Mismatching"] == 0
        and
        row["Match_Rate_%"] == 100.0
    )

    print(
        f"{feature:<40} : "
        f"{'PASS' if exact else 'FAIL'}"
    )

    if not exact:
        all_exact = False


# ------------------------------------------------------------
# 7. FINAL FORENSIC DECISION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 4.17 FORENSIC DECISION")
print("=" * 70)

if all_exact:

    print("STATUS: PASS")
    print()
    print(
        "The stored rolling features are exactly reproducible"
    )
    print(
        "using row-based pandas rolling windows."
    )
    print()
    print(
        "Verified semantics:"
    )
    print(
        " - 6h  = 6 rows"
    )
    print(
        " - 12h = 12 rows"
    )
    print(
        " - 24h = 24 rows"
    )
    print(
        " - min_periods = full window"
    )
    print(
        " - current row included"
    )
    print(
        " - pandas default rolling behavior"
    )

else:

    print("STATUS: INVESTIGATION REQUIRED")
    print()
    print(
        "The stored rolling features do not exactly reproduce"
    )
    print(
        "under the actual row-based generation semantics."
    )


# ------------------------------------------------------------
# 8. DATA SAFETY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DATA SAFETY CHECK")
print("=" * 70)

print("Original file :", FILE)
print("Audit shape   :", audit.shape)

print("\nNO DATA WAS MODIFIED.")
print("NO FEATURES WERE REGENERATED.")
print("NO ROWS WERE DELETED.")

print("\n" + "=" * 70)
print("STEP 4.17 COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# STEP 4.18B — SOURCE-LEVEL ROLLING FEATURE REPRODUCTION
# ============================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 4.18B — SOURCE-LEVEL ROLLING FEATURE REPRODUCTION")
print("=" * 70)


# ------------------------------------------------------------
# 1. LOAD FINAL ML DATASET
# ------------------------------------------------------------

FINAL_FILE = "data/cleaned/final_ml_dataset.csv"

final_df = pd.read_csv(FINAL_FILE)

TIME_COL = "Data Acquisition Time"
FINAL_VALUE_COL = "Current_Water_Level"

final_df[TIME_COL] = pd.to_datetime(
    final_df[TIME_COL],
    errors="coerce"
)

final_df = (
    final_df
    .sort_values(TIME_COL)
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 2. LOAD ORIGINAL INTEGRATED SOURCE
# ------------------------------------------------------------

SOURCE_FILE = (
    "data/cleaned/"
    "water_rainfall_integrated_clean.csv"
)

source_df = pd.read_csv(SOURCE_FILE)

SOURCE_VALUE_COL = (
    "River Water Level Telemetry Hourly (meter)"
)

source_df[TIME_COL] = pd.to_datetime(
    source_df[TIME_COL],
    errors="coerce"
)

source_df = (
    source_df
    .sort_values(TIME_COL)
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 3. BASIC CHECK
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DATASETS")
print("=" * 70)

print(
    "Final ML shape    :",
    final_df.shape
)

print(
    "Source shape      :",
    source_df.shape
)

print(
    "Final date range  :",
    final_df[TIME_COL].min(),
    "→",
    final_df[TIME_COL].max()
)

print(
    "Source date range :",
    source_df[TIME_COL].min(),
    "→",
    source_df[TIME_COL].max()
)


# ------------------------------------------------------------
# 4. SOURCE WATER-LEVEL QUALITY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("SOURCE WATER-LEVEL CHECK")
print("=" * 70)

print(
    "Source water missing:",
    source_df[SOURCE_VALUE_COL].isna().sum()
)

print(
    "Source duplicate timestamps:",
    source_df[TIME_COL].duplicated().sum()
)

print(
    "Source sorted:",
    source_df[TIME_COL].is_monotonic_increasing
)


# ------------------------------------------------------------
# 5. CREATE SOURCE ROLLING FEATURES
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("SOURCE ROLLING GENERATION")
print("=" * 70)

source_calc = source_df.copy()

source_calc[
    "CALC_RollingMean_6h"
] = (
    source_calc[SOURCE_VALUE_COL]
    .rolling(
        window=6,
        min_periods=6
    )
    .mean()
)

source_calc[
    "CALC_RollingMean_12h"
] = (
    source_calc[SOURCE_VALUE_COL]
    .rolling(
        window=12,
        min_periods=12
    )
    .mean()
)

source_calc[
    "CALC_RollingMean_24h"
] = (
    source_calc[SOURCE_VALUE_COL]
    .rolling(
        window=24,
        min_periods=24
    )
    .mean()
)

source_calc[
    "CALC_RollingMax_24h"
] = (
    source_calc[SOURCE_VALUE_COL]
    .rolling(
        window=24,
        min_periods=24
    )
    .max()
)

source_calc[
    "CALC_RollingMin_24h"
] = (
    source_calc[SOURCE_VALUE_COL]
    .rolling(
        window=24,
        min_periods=24
    )
    .min()
)


# ------------------------------------------------------------
# 6. CHECK CURRENT WATER LEVEL MAPPING
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CURRENT WATER LEVEL MAPPING")
print("=" * 70)

source_water_lookup = (
    source_calc[
        [
            TIME_COL,
            SOURCE_VALUE_COL
        ]
    ]
    .drop_duplicates(TIME_COL)
    .set_index(TIME_COL)[SOURCE_VALUE_COL]
)

mapped_source_water = (
    final_df[TIME_COL]
    .map(source_water_lookup)
)

valid_water = (
    final_df[FINAL_VALUE_COL].notna()
    &
    mapped_source_water.notna()
)

water_match = np.isclose(
    final_df.loc[
        valid_water,
        FINAL_VALUE_COL
    ],
    mapped_source_water.loc[
        valid_water
    ],
    rtol=1e-10,
    atol=1e-10
)

print(
    "Final rows checked:",
    int(valid_water.sum())
)

print(
    "Water-level matching:",
    int(water_match.sum())
)

print(
    "Water-level mismatching:",
    int((~water_match).sum())
)

if valid_water.sum() > 0:
    print(
        "Water-level match rate:",
        water_match.mean() * 100,
        "%"
    )


# ------------------------------------------------------------
# 7. MAP SOURCE ROLLING VALUES TO FINAL TIMESTAMPS
# ------------------------------------------------------------

source_calc_lookup = (
    source_calc
    .drop_duplicates(TIME_COL)
    .set_index(TIME_COL)
)


feature_map = {

    "WaterLevel_RollingMean_6h":
        "CALC_RollingMean_6h",

    "WaterLevel_RollingMean_12h":
        "CALC_RollingMean_12h",

    "WaterLevel_RollingMean_24h":
        "CALC_RollingMean_24h",

    "WaterLevel_RollingMax_24h":
        "CALC_RollingMax_24h",

    "WaterLevel_RollingMin_24h":
        "CALC_RollingMin_24h"
}


# ------------------------------------------------------------
# 8. COMPARE STORED VS SOURCE-RECONSTRUCTED
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("ROLLING FEATURE COMPARISON")
print("=" * 70)

results = []

for stored_feature, calc_feature in feature_map.items():

    source_values = (
        final_df[TIME_COL]
        .map(
            source_calc_lookup[calc_feature]
        )
    )

    stored_values = final_df[
        stored_feature
    ]

    valid = (
        stored_values.notna()
        &
        source_values.notna()
    )

    if valid.sum() == 0:

        print("\n", stored_feature)
        print("No comparable rows.")

        continue

    stored_v = stored_values.loc[valid]
    source_v = source_values.loc[valid]

    match = np.isclose(
        stored_v,
        source_v,
        rtol=1e-10,
        atol=1e-10
    )

    diff = (
        stored_v - source_v
    ).abs()

    result = {
        "Feature": stored_feature,
        "Checked": int(valid.sum()),
        "Matching": int(match.sum()),
        "Mismatching": int((~match).sum()),
        "Match_Rate_%": match.mean() * 100,
        "Mean_Abs_Difference": diff.mean(),
        "Median_Abs_Difference": diff.median(),
        "Maximum_Difference": diff.max()
    }

    results.append(result)

    print("\n" + "-" * 70)
    print(stored_feature)
    print("-" * 70)

    print(
        "Checked             :",
        result["Checked"]
    )

    print(
        "Matching            :",
        result["Matching"]
    )

    print(
        "Mismatching         :",
        result["Mismatching"]
    )

    print(
        "Match rate          :",
        result["Match_Rate_%"],
        "%"
    )

    print(
        "Mean abs difference :",
        result["Mean_Abs_Difference"]
    )

    print(
        "Maximum difference  :",
        result["Maximum_Difference"]
    )


# ------------------------------------------------------------
# 9. SUMMARY TABLE
# ------------------------------------------------------------

results_df = pd.DataFrame(results)

print("\n" + "=" * 70)
print("SOURCE-LEVEL ROLLING AUDIT SUMMARY")
print("=" * 70)

print(
    results_df.to_string(index=False)
)


# ------------------------------------------------------------
# 10. DATA SAFETY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DATA SAFETY CHECK")
print("=" * 70)

print(
    "Final ML shape :",
    final_df.shape
)

print(
    "Source shape   :",
    source_df.shape
)

print("\nNO DATA WAS MODIFIED.")
print("NO FEATURES WERE REGENERATED IN ANY FILE.")
print("NO ROWS WERE DELETED.")

print("\n" + "=" * 70)
print("STEP 4.18B COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# STEP 4.18C — ROLLING MISMATCH CAUSE FORENSIC
# ============================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 4.18C — ROLLING MISMATCH CAUSE FORENSIC")
print("=" * 70)


# ------------------------------------------------------------
# 1. LOAD FINAL ML DATASET
# ------------------------------------------------------------

FINAL_FILE = "data/cleaned/final_ml_dataset.csv"
SOURCE_FILE = "data/cleaned/water_rainfall_integrated_clean.csv"

final_df = pd.read_csv(FINAL_FILE)
source_df = pd.read_csv(SOURCE_FILE)

TIME_COL = "Data Acquisition Time"
VALUE_COL = "Current_Water_Level"
SOURCE_VALUE_COL = "River Water Level Telemetry Hourly (meter)"

final_df[TIME_COL] = pd.to_datetime(
    final_df[TIME_COL],
    errors="coerce"
)

source_df[TIME_COL] = pd.to_datetime(
    source_df[TIME_COL],
    errors="coerce"
)

final_df = (
    final_df
    .sort_values(TIME_COL)
    .reset_index(drop=True)
)

source_df = (
    source_df
    .sort_values(TIME_COL)
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# 2. CREATE SOURCE WATER LEVEL LOOKUP
# ------------------------------------------------------------

source_lookup = (
    source_df[
        [TIME_COL, SOURCE_VALUE_COL]
    ]
    .drop_duplicates(TIME_COL)
    .set_index(TIME_COL)[SOURCE_VALUE_COL]
)


# ------------------------------------------------------------
# 3. MAP SOURCE WATER LEVEL TO FINAL TIMESTAMPS
# ------------------------------------------------------------

audit = final_df[
    [
        TIME_COL,
        VALUE_COL,
        "WaterLevel_RollingMean_6h",
        "WaterLevel_RollingMean_12h",
        "WaterLevel_RollingMean_24h",
        "WaterLevel_RollingMax_24h",
        "WaterLevel_RollingMin_24h"
    ]
].copy()

audit["Source_Water_Level"] = (
    audit[TIME_COL]
    .map(source_lookup)
)


# ------------------------------------------------------------
# 4. SOURCE TIMESTAMP GAP ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("SOURCE TIMESTAMP GAP ANALYSIS")
print("=" * 70)

source_time_diff = (
    source_df[TIME_COL]
    .diff()
)

print("\nMost common timestamp gaps:")

print(
    source_time_diff
    .value_counts()
    .head(15)
    .to_string()
)

print("\nLargest timestamp gaps:")

largest_gaps = (
    source_time_diff
    .sort_values(
        ascending=False
    )
    .head(20)
)

print(
    largest_gaps
    .to_string()
)


# ------------------------------------------------------------
# 5. SOURCE MISSING WATER LEVEL ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("SOURCE MISSING WATER LEVEL ANALYSIS")
print("=" * 70)

print(
    "Total source rows:",
    len(source_df)
)

print(
    "Missing source water levels:",
    source_df[SOURCE_VALUE_COL].isna().sum()
)

missing_water_times = source_df.loc[
    source_df[SOURCE_VALUE_COL].isna(),
    TIME_COL
]

print(
    "\nFirst 30 missing-water timestamps:"
)

print(
    missing_water_times
    .head(30)
    .to_string(index=False)
)


# ------------------------------------------------------------
# 6. REGENERATE SOURCE ROLLING FEATURES
#    ONLY FOR FORENSIC COMPARISON
# ------------------------------------------------------------

source_work = source_df[
    [
        TIME_COL,
        SOURCE_VALUE_COL
    ]
].copy()

source_work = (
    source_work
    .set_index(TIME_COL)
)

source_work["candidate_6h"] = (
    source_work[SOURCE_VALUE_COL]
    .rolling(
        window=6,
        min_periods=6
    )
    .mean()
)

source_work["candidate_12h"] = (
    source_work[SOURCE_VALUE_COL]
    .rolling(
        window=12,
        min_periods=12
    )
    .mean()
)

source_work["candidate_24h_mean"] = (
    source_work[SOURCE_VALUE_COL]
    .rolling(
        window=24,
        min_periods=24
    )
    .mean()
)

source_work["candidate_24h_max"] = (
    source_work[SOURCE_VALUE_COL]
    .rolling(
        window=24,
        min_periods=24
    )
    .max()
)

source_work["candidate_24h_min"] = (
    source_work[SOURCE_VALUE_COL]
    .rolling(
        window=24,
        min_periods=24
    )
    .min()
)


# ------------------------------------------------------------
# 7. MAP SOURCE CANDIDATES TO FINAL DATA
# ------------------------------------------------------------

for feature, candidate in {
    "WaterLevel_RollingMean_6h": "candidate_6h",
    "WaterLevel_RollingMean_12h": "candidate_12h",
    "WaterLevel_RollingMean_24h": "candidate_24h_mean",
    "WaterLevel_RollingMax_24h": "candidate_24h_max",
    "WaterLevel_RollingMin_24h": "candidate_24h_min"
}.items():

    audit[
        "Source_" + feature
    ] = audit[TIME_COL].map(
        source_work[candidate]
    )


# ------------------------------------------------------------
# 8. IDENTIFY MISMATCH ROWS
# ------------------------------------------------------------

features = [
    "WaterLevel_RollingMean_6h",
    "WaterLevel_RollingMean_12h",
    "WaterLevel_RollingMean_24h",
    "WaterLevel_RollingMax_24h",
    "WaterLevel_RollingMin_24h"
]

print("\n" + "=" * 70)
print("MISMATCH CAUSE ANALYSIS")
print("=" * 70)

summary = []

for feature in features:

    source_feature = "Source_" + feature

    valid = (
        audit[feature].notna()
        &
        audit[source_feature].notna()
    )

    diff = (
        audit.loc[valid, feature]
        -
        audit.loc[valid, source_feature]
    ).abs()

    mismatch_mask = (
        valid
        &
        ~np.isclose(
            audit[feature],
            audit[source_feature],
            rtol=1e-10,
            atol=1e-10
        )
    )

    mismatch_times = audit.loc[
        mismatch_mask,
        TIME_COL
    ]

    # Check whether mismatch timestamps are close
    # to missing source water-level timestamps
    mismatch_count = int(
        mismatch_mask.sum()
    )

    summary.append({
        "Feature": feature,
        "Mismatches": mismatch_count,
        "Mismatch_Near_Missing_Source": 0
    })

    print("\n" + "-" * 70)
    print(feature)
    print("-" * 70)

    print(
        "Mismatch rows:",
        mismatch_count
    )

    if mismatch_count > 0:

        sample = audit.loc[
            mismatch_mask,
            [
                TIME_COL,
                VALUE_COL,
                feature,
                source_feature
            ]
        ].copy()

        sample["Absolute_Difference"] = (
            sample[feature]
            -
            sample[source_feature]
        ).abs()

        print("\nLargest mismatch examples:")

        print(
            sample
            .sort_values(
                "Absolute_Difference",
                ascending=False
            )
            .head(20)
            .to_string(index=False)
        )


# ------------------------------------------------------------
# 9. INSPECT SOURCE WINDOW AROUND LARGEST MISMATCH
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("LARGEST MISMATCH WINDOW")
print("=" * 70)

all_mismatches = []

for feature in features:

    source_feature = "Source_" + feature

    valid = (
        audit[feature].notna()
        &
        audit[source_feature].notna()
    )

    mismatch_mask = (
        valid
        &
        ~np.isclose(
            audit[feature],
            audit[source_feature],
            rtol=1e-10,
            atol=1e-10
        )
    )

    temp = audit.loc[
        mismatch_mask,
        [
            TIME_COL,
            feature,
            source_feature
        ]
    ].copy()

    temp["Feature"] = feature

    temp["Difference"] = (
        temp[feature]
        -
        temp[source_feature]
    ).abs()

    all_mismatches.append(temp)


if all_mismatches:

    mismatch_df = pd.concat(
        all_mismatches,
        ignore_index=True
    )

    largest = (
        mismatch_df
        .sort_values(
            "Difference",
            ascending=False
        )
        .head(1)
    )

    if len(largest) > 0:

        ts = largest.iloc[0][TIME_COL]

        print(
            "\nLargest mismatch timestamp:",
            ts
        )

        window = source_df[
            (
                source_df[TIME_COL]
                >= ts - pd.Timedelta(hours=30)
            )
            &
            (
                source_df[TIME_COL]
                <= ts + pd.Timedelta(hours=30)
            )
        ][
            [
                TIME_COL,
                SOURCE_VALUE_COL
            ]
        ]

        print(
            "\nSource rows around largest mismatch:"
        )

        print(
            window.to_string(index=False)
        )


# ------------------------------------------------------------
# 10. FINAL DECISION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 4.18C FORENSIC DECISION")
print("=" * 70)

print(
    "Source rolling features have been reproduced "
    "for investigation only."
)

print(
    "No final ML features were changed."
)

print(
    "No rows were deleted."
)

print(
    "No dataset was overwritten."
)

print(
    "\nSTATUS: INVESTIGATION CONTINUES"
)

print(
    "Next decision will be based on timestamp gaps, "
    "missing source water levels, and mismatch windows."
)

print("\n" + "=" * 70)
print("STEP 4.18C COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 4.18D — ORIGINAL FEATURE SOURCE IDENTIFICATION
# ======================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 4.18D — ORIGINAL FEATURE SOURCE IDENTIFICATION")
print("=" * 70)

# ----------------------------------------------------------------------
# FILES
# ----------------------------------------------------------------------

FINAL_FILE = "data/cleaned/final_ml_dataset.csv"

SOURCE_FILES = {
    "water_rainfall_integrated_clean":
        "data/cleaned/water_rainfall_integrated_clean.csv",

    "darrang_fakipara_cleaned":
        "data/cleaned/darrang_fakipara_cleaned.csv"
}

TIME_COL = "Data Acquisition Time"

FINAL_WATER = "Current_Water_Level"

SOURCE_WATER = "River Water Level Telemetry Hourly (meter)"

ROLLING_FEATURES = {
    "WaterLevel_RollingMean_6h": (6, "mean"),
    "WaterLevel_RollingMean_12h": (12, "mean"),
    "WaterLevel_RollingMean_24h": (24, "mean"),
    "WaterLevel_RollingMax_24h": (24, "max"),
    "WaterLevel_RollingMin_24h": (24, "min"),
}

# ----------------------------------------------------------------------
# LOAD FINAL DATASET
# ----------------------------------------------------------------------

final_df = pd.read_csv(FINAL_FILE)

final_df[TIME_COL] = pd.to_datetime(
    final_df[TIME_COL],
    errors="coerce"
)

final_df = final_df.sort_values(TIME_COL).reset_index(drop=True)

print("\n" + "=" * 70)
print("FINAL DATASET")
print("=" * 70)

print(f"Rows    : {len(final_df)}")
print(f"Columns : {len(final_df.columns)}")
print(
    f"Range   : {final_df[TIME_COL].min()} → "
    f"{final_df[TIME_COL].max()}"
)

# ----------------------------------------------------------------------
# FUNCTION
# ----------------------------------------------------------------------

def audit_candidate(
    final_df,
    candidate_df,
    candidate_name,
    water_col
):

    print("\n" + "-" * 70)
    print(f"CANDIDATE: {candidate_name}")
    print("-" * 70)

    candidate = candidate_df.copy()

    candidate[TIME_COL] = pd.to_datetime(
        candidate[TIME_COL],
        errors="coerce"
    )

    candidate[water_col] = pd.to_numeric(
        candidate[water_col],
        errors="coerce"
    )

    candidate = (
        candidate[
            [TIME_COL, water_col]
        ]
        .dropna(subset=[TIME_COL])
        .drop_duplicates(TIME_COL)
        .sort_values(TIME_COL)
        .reset_index(drop=True)
    )

    print(f"Candidate rows: {len(candidate)}")
    print(f"Water column  : {water_col}")
    print(
        f"Range         : {candidate[TIME_COL].min()} → "
        f"{candidate[TIME_COL].max()}"
    )

    # --------------------------------------------------------------
    # Generate rolling features on candidate
    # --------------------------------------------------------------

    candidate_roll = candidate.copy()

    for feature, (window, aggregation) in ROLLING_FEATURES.items():

        rolling_obj = (
            candidate_roll[water_col]
            .rolling(
                window=window,
                min_periods=window
            )
        )

        if aggregation == "mean":
            candidate_roll[feature] = rolling_obj.mean()

        elif aggregation == "max":
            candidate_roll[feature] = rolling_obj.max()

        elif aggregation == "min":
            candidate_roll[feature] = rolling_obj.min()

    # --------------------------------------------------------------
    # Merge with final
    # --------------------------------------------------------------

    merged = final_df[
        [
            TIME_COL,
            FINAL_WATER,
            *ROLLING_FEATURES.keys()
        ]
    ].merge(
        candidate_roll[
            [
                TIME_COL,
                water_col,
                *ROLLING_FEATURES.keys()
            ]
        ],
        on=TIME_COL,
        how="inner",
        suffixes=("_final", "_candidate")
    )

    print(
        f"Timestamp matches: "
        f"{len(merged)} / {len(final_df)}"
    )

    # --------------------------------------------------------------
    # Current water consistency
    # --------------------------------------------------------------

    water_diff = (
        merged[f"{FINAL_WATER}"]
        - merged[water_col]
    ).abs()

    water_valid = water_diff.notna()

    water_match = (
        water_diff[water_valid] <= 1e-9
    ).sum()

    water_checked = water_valid.sum()

    print("\nCurrent water-level consistency:")
    print(f"Checked    : {water_checked}")
    print(f"Matching   : {water_match}")
    print(f"Mismatching: {water_checked - water_match}")

    # --------------------------------------------------------------
    # Rolling comparison
    # --------------------------------------------------------------

    results = []

    for feature in ROLLING_FEATURES:

        final_col = f"{feature}_final"
        cand_col = f"{feature}_candidate"

        valid = (
            merged[final_col].notna()
            &
            merged[cand_col].notna()
        )

        diff = (
            merged.loc[valid, final_col]
            -
            merged.loc[valid, cand_col]
        ).abs()

        matching = (diff <= 1e-9).sum()

        checked = len(diff)

        mismatching = checked - matching

        match_rate = (
            matching / checked * 100
            if checked > 0
            else 0
        )

        mean_diff = (
            diff.mean()
            if checked > 0
            else np.nan
        )

        max_diff = (
            diff.max()
            if checked > 0
            else np.nan
        )

        results.append({
            "Candidate": candidate_name,
            "Feature": feature,
            "Checked": checked,
            "Matching": matching,
            "Mismatching": mismatching,
            "Match_Rate_%": match_rate,
            "Mean_Abs_Difference": mean_diff,
            "Maximum_Difference": max_diff
        })

    result_df = pd.DataFrame(results)

    print("\nRolling reproduction:")
    print(result_df.to_string(index=False))

    return result_df


# ======================================================================
# RUN CANDIDATES
# ======================================================================

all_results = []

for candidate_name, file_path in SOURCE_FILES.items():

    try:

        candidate_df = pd.read_csv(file_path)

        print("\n" + "=" * 70)
        print(f"LOADING: {candidate_name}")
        print("=" * 70)

        print(f"File: {file_path}")
        print(f"Shape: {candidate_df.shape}")

        if TIME_COL not in candidate_df.columns:
            print("TIME COLUMN: MISSING")
            continue

        if SOURCE_WATER not in candidate_df.columns:

            print(
                f"WATER COLUMN '{SOURCE_WATER}' NOT FOUND"
            )

            print("\nAvailable numeric columns:")

            print(
                candidate_df.select_dtypes(
                    include=np.number
                ).columns.tolist()
            )

            continue

        result = audit_candidate(
            final_df,
            candidate_df,
            candidate_name,
            SOURCE_WATER
        )

        all_results.append(result)

    except Exception as e:

        print(
            f"\nERROR while loading {candidate_name}:"
        )

        print(type(e).__name__, str(e))


# ======================================================================
# FINAL SUMMARY
# ======================================================================

print("\n" + "=" * 70)
print("STEP 4.18D — SUMMARY")
print("=" * 70)

if all_results:

    summary = pd.concat(
        all_results,
        ignore_index=True
    )

    print(
        summary.to_string(index=False)
    )

else:

    print("No candidate could be audited.")


print("\n" + "=" * 70)
print("DATA SAFETY")
print("=" * 70)

print("NO DATA WAS MODIFIED.")
print("NO FEATURES WERE REGENERATED IN FINAL FILE.")
print("NO ROWS WERE DELETED.")
print("NO DATASET WAS OVERWRITTEN.")

print("\n" + "=" * 70)
print("STEP 4.18D COMPLETE")
print("=" * 70)
```

```python
import pandas as pd
import numpy as np

# ==============================================================
# STEP 4.18E — ROLLING SEMANTICS FORENSIC
# SAFE / READ-ONLY AUDIT
# ==============================================================

FINAL_FILE = "data/cleaned/final_ml_dataset.csv"
CANDIDATE_FILES = {
    "water_rainfall_integrated_clean":
        "data/cleaned/water_rainfall_integrated_clean.csv",

    "darrang_fakipara_cleaned":
        "data/cleaned/darrang_fakipara_cleaned.csv"
}

TIME_COL_FINAL = "Data Acquisition Time"
TIME_COL_SOURCE = "Data Acquisition Time"

ROLLING_FEATURES = {
    "WaterLevel_RollingMean_6h":  ("mean", 6),
    "WaterLevel_RollingMean_12h": ("mean", 12),
    "WaterLevel_RollingMean_24h": ("mean", 24),
    "WaterLevel_RollingMax_24h":  ("max", 24),
    "WaterLevel_RollingMin_24h":  ("min", 24),
}

# ==============================================================
# LOAD FINAL DATASET
# ==============================================================

final_df = pd.read_csv(FINAL_FILE)

final_df[TIME_COL_FINAL] = pd.to_datetime(
    final_df[TIME_COL_FINAL],
    errors="coerce"
)

final_df = final_df.sort_values(
    TIME_COL_FINAL
).reset_index(drop=True)

print("=" * 70)
print("STEP 4.18E — ROLLING SEMANTICS FORENSIC")
print("=" * 70)

print("\nFINAL DATASET")
print("=" * 70)
print("Rows    :", len(final_df))
print("Columns :", len(final_df.columns))
print(
    "Range   :",
    final_df[TIME_COL_FINAL].min(),
    "→",
    final_df[TIME_COL_FINAL].max()
)

# ==============================================================
# FINAL WATER COLUMN
# ==============================================================

FINAL_WATER_COL = "Current_Water_Level"

if FINAL_WATER_COL not in final_df.columns:
    raise KeyError(
        f"Final dataset does not contain '{FINAL_WATER_COL}'.\n"
        f"Available columns:\n{list(final_df.columns)}"
    )

final_df[FINAL_WATER_COL] = pd.to_numeric(
    final_df[FINAL_WATER_COL],
    errors="coerce"
)

# ==============================================================
# HELPER — DETECT SOURCE WATER COLUMN
# ==============================================================

def detect_water_column(df):

    candidates = [
        "River Water Level Telemetry Hourly (meter)",
        "Current_Water_Level",
        "Water_Level",
        "WaterLevel"
    ]

    for col in candidates:
        if col in df.columns:
            return col

    raise KeyError(
        "No recognized water-level column found.\n"
        f"Available columns:\n{list(df.columns)}"
    )


# ==============================================================
# HELPER — EXACT ROW-BASED ROLLING
# ==============================================================

def generate_row_based_rolling(series, aggregation, window):

    if aggregation == "mean":
        return series.rolling(
            window=window,
            min_periods=window
        ).mean()

    elif aggregation == "max":
        return series.rolling(
            window=window,
            min_periods=window
        ).max()

    elif aggregation == "min":
        return series.rolling(
            window=window,
            min_periods=window
        ).min()

    else:
        raise ValueError(
            f"Unknown aggregation: {aggregation}"
        )


# ==============================================================
# HELPER — COMPARE
# ==============================================================

def compare_series(stored, reproduced):

    comparison = pd.DataFrame({
        "stored": stored,
        "reproduced": reproduced
    })

    comparison["difference"] = (
        comparison["stored"] -
        comparison["reproduced"]
    ).abs()

    valid = comparison.dropna(
        subset=["stored", "reproduced"]
    )

    if len(valid) == 0:
        return {
            "Checked": 0,
            "Matching": 0,
            "Mismatching": 0,
            "Match_Rate_%": 0,
            "Mean_Abs_Difference": np.nan,
            "Median_Abs_Difference": np.nan,
            "Maximum_Difference": np.nan
        }

    tolerance = 1e-9

    matching = (
        valid["difference"] <= tolerance
    )

    return {
        "Checked": len(valid),
        "Matching": int(matching.sum()),
        "Mismatching": int((~matching).sum()),
        "Match_Rate_%": matching.mean() * 100,
        "Mean_Abs_Difference": valid["difference"].mean(),
        "Median_Abs_Difference": valid["difference"].median(),
        "Maximum_Difference": valid["difference"].max()
    }


# ==============================================================
# TEST DIFFERENT ROLLING SEMANTICS ON FINAL DATASET
# ==============================================================

print("\n" + "=" * 70)
print("FINAL DATASET — SEMANTICS TEST")
print("=" * 70)

semantics_results = []

for feature, (agg, window) in ROLLING_FEATURES.items():

    print("\n" + "-" * 70)
    print(feature)
    print("-" * 70)

    stored = final_df[feature]

    # ----------------------------------------------------------
    # SEMANTIC A
    # Current row INCLUDED
    # ----------------------------------------------------------

    reproduced_current = generate_row_based_rolling(
        final_df[FINAL_WATER_COL],
        agg,
        window
    )

    result_current = compare_series(
        stored,
        reproduced_current
    )

    # ----------------------------------------------------------
    # SEMANTIC B
    # Current row EXCLUDED
    # ----------------------------------------------------------

    reproduced_previous = (
        generate_row_based_rolling(
            final_df[FINAL_WATER_COL],
            agg,
            window
        )
        .shift(1)
    )

    result_previous = compare_series(
        stored,
        reproduced_previous
    )

    # ----------------------------------------------------------
    # SEMANTIC C
    # Window of previous N rows + current row
    # with min_periods=1
    # ----------------------------------------------------------

    if agg == "mean":
        reproduced_min1 = final_df[
            FINAL_WATER_COL
        ].rolling(
            window=window,
            min_periods=1
        ).mean()

    elif agg == "max":
        reproduced_min1 = final_df[
            FINAL_WATER_COL
        ].rolling(
            window=window,
            min_periods=1
        ).max()

    elif agg == "min":
        reproduced_min1 = final_df[
            FINAL_WATER_COL
        ].rolling(
            window=window,
            min_periods=1
        ).min()

    result_min1 = compare_series(
        stored,
        reproduced_min1
    )

    print("\nCurrent row INCLUDED:")
    print(
        f"  Match rate : "
        f"{result_current['Match_Rate_%']:.6f}%"
    )
    print(
        f"  Mismatch   : "
        f"{result_current['Mismatching']}"
    )
    print(
        f"  Mean diff  : "
        f"{result_current['Mean_Abs_Difference']}"
    )

    print("\nCurrent row EXCLUDED:")
    print(
        f"  Match rate : "
        f"{result_previous['Match_Rate_%']:.6f}%"
    )
    print(
        f"  Mismatch   : "
        f"{result_previous['Mismatching']}"
    )
    print(
        f"  Mean diff  : "
        f"{result_previous['Mean_Abs_Difference']}"
    )

    print("\nmin_periods=1:")
    print(
        f"  Match rate : "
        f"{result_min1['Match_Rate_%']:.6f}%"
    )
    print(
        f"  Mismatch   : "
        f"{result_min1['Mismatching']}"
    )
    print(
        f"  Mean diff  : "
        f"{result_min1['Mean_Abs_Difference']}"
    )

    semantics_results.append({
        "Feature": feature,

        "Included_Match_%":
            result_current["Match_Rate_%"],

        "Excluded_Match_%":
            result_previous["Match_Rate_%"],

        "Min1_Match_%":
            result_min1["Match_Rate_%"]
    })


# ==============================================================
# SOURCE DATASETS
# ==============================================================

print("\n" + "=" * 70)
print("SOURCE DATASET SEMANTICS CHECK")
print("=" * 70)

source_results = []

for candidate_name, candidate_file in CANDIDATE_FILES.items():

    print("\n" + "-" * 70)
    print("CANDIDATE:", candidate_name)
    print("-" * 70)

    df = pd.read_csv(candidate_file)

    # ----------------------------------------------------------
    # AUTOMATIC COLUMN DETECTION
    # ----------------------------------------------------------

    if TIME_COL_SOURCE not in df.columns:

        print(
            "SKIPPED — timestamp column missing"
        )
        print("Available columns:")
        print(list(df.columns))
        continue

    water_col = detect_water_column(df)

    print("Water column detected:")
    print(" ", water_col)

    # ----------------------------------------------------------
    # PREPARE
    # ----------------------------------------------------------

    df[TIME_COL_SOURCE] = pd.to_datetime(
        df[TIME_COL_SOURCE],
        errors="coerce"
    )

    df[water_col] = pd.to_numeric(
        df[water_col],
        errors="coerce"
    )

    df = df.sort_values(
        TIME_COL_SOURCE
    ).reset_index(drop=True)

    print("Rows :", len(df))
    print(
        "Range:",
        df[TIME_COL_SOURCE].min(),
        "→",
        df[TIME_COL_SOURCE].max()
    )

    # ----------------------------------------------------------
    # TEST ALL FEATURES
    # ----------------------------------------------------------

    for feature, (agg, window) in ROLLING_FEATURES.items():

        if feature not in final_df.columns:
            continue

        # Generate from source
        reproduced = generate_row_based_rolling(
            df[water_col],
            agg,
            window
        )

        source_lookup = pd.DataFrame({
            TIME_COL_SOURCE:
                df[TIME_COL_SOURCE],
            "reproduced":
                reproduced
        })

        # Remove duplicate timestamps defensively
        source_lookup = source_lookup.drop_duplicates(
            TIME_COL_SOURCE
        )

        # Map source reproduction to final timestamps
        merged = final_df[
            [TIME_COL_FINAL, feature]
        ].merge(
            source_lookup,
            left_on=TIME_COL_FINAL,
            right_on=TIME_COL_SOURCE,
            how="inner"
        )

        result = compare_series(
            merged[feature],
            merged["reproduced"]
        )

        source_results.append({
            "Candidate": candidate_name,
            "Feature": feature,
            **result
        })


# ==============================================================
# FINAL SUMMARY
# ==============================================================

print("\n" + "=" * 70)
print("STEP 4.18E — SEMANTICS SUMMARY")
print("=" * 70)

semantics_df = pd.DataFrame(
    semantics_results
)

print("\nFINAL DATASET SEMANTICS:")
print(
    semantics_df.to_string(
        index=False
    )
)

if source_results:

    source_summary = pd.DataFrame(
        source_results
    )

    print("\nSOURCE DATASET COMPARISON:")
    print(
        source_summary.to_string(
            index=False
        )
    )

# ==============================================================
# SAFETY
# ==============================================================

print("\n" + "=" * 70)
print("DATA SAFETY CHECK")
print("=" * 70)

print("NO DATA WAS MODIFIED.")
print("NO FEATURES WERE REGENERATED IN FINAL FILE.")
print("NO ROWS WERE DELETED.")
print("NO DATASET WAS OVERWRITTEN.")

print("\n" + "=" * 70)
print("STEP 4.18E COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 4.18F — EXACT ROLLING WINDOW / GAP SEMANTICS FORENSIC
# ======================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 4.18F — EXACT ROLLING WINDOW / GAP SEMANTICS FORENSIC")
print("=" * 70)


# ======================================================================
# CONFIGURATION
# ======================================================================

FINAL_FILE = "data/cleaned/final_ml_dataset.csv"

SOURCE_FILE = "data/cleaned/darrang_fakipara_cleaned.csv"

TIME_COL = "Data Acquisition Time"

WATER_COL = "River Water Level Telemetry Hourly (meter)"

FINAL_WATER_COL = "Current_Water_Level"

ROLLING_FEATURES = {
    "WaterLevel_RollingMean_6h":  ("mean", 6),
    "WaterLevel_RollingMean_12h": ("mean", 12),
    "WaterLevel_RollingMean_24h": ("mean", 24),
    "WaterLevel_RollingMax_24h":  ("max", 24),
    "WaterLevel_RollingMin_24h":  ("min", 24),
}


# ======================================================================
# LOAD DATA
# ======================================================================

final_df = pd.read_csv(FINAL_FILE)
source_df = pd.read_csv(SOURCE_FILE)

print("\n" + "=" * 70)
print("DATA LOADING")
print("=" * 70)

print(f"Final shape  : {final_df.shape}")
print(f"Source shape : {source_df.shape}")


# ======================================================================
# PREPARE FINAL DATA
# ======================================================================

final_df[TIME_COL] = pd.to_datetime(
    final_df[TIME_COL],
    errors="coerce"
)

final_df = final_df.sort_values(TIME_COL).reset_index(drop=True)

final_df[FINAL_WATER_COL] = pd.to_numeric(
    final_df[FINAL_WATER_COL],
    errors="coerce"
)


# ======================================================================
# PREPARE SOURCE DATA
# ======================================================================

source_df[TIME_COL] = pd.to_datetime(
    source_df[TIME_COL],
    errors="coerce"
)

source_df[WATER_COL] = pd.to_numeric(
    source_df[WATER_COL],
    errors="coerce"
)

source_df = (
    source_df
    .sort_values(TIME_COL)
    .drop_duplicates(TIME_COL)
    .reset_index(drop=True)
)


print("\nFinal date range:")
print(final_df[TIME_COL].min(), "→", final_df[TIME_COL].max())

print("\nSource date range:")
print(source_df[TIME_COL].min(), "→", source_df[TIME_COL].max())


# ======================================================================
# CREATE SOURCE LOOKUP
# ======================================================================

source_lookup = source_df.set_index(TIME_COL)[WATER_COL]


# ======================================================================
# MAP SOURCE WATER LEVEL TO FINAL TIMESTAMPS
# ======================================================================

final_df["_SOURCE_WATER"] = final_df[TIME_COL].map(source_lookup)


# ======================================================================
# VERIFY WATER LEVEL FIRST
# ======================================================================

water_check = final_df[
    final_df[FINAL_WATER_COL].notna()
    & final_df["_SOURCE_WATER"].notna()
].copy()

water_check["water_diff"] = (
    water_check[FINAL_WATER_COL]
    - water_check["_SOURCE_WATER"]
).abs()

print("\n" + "=" * 70)
print("WATER LEVEL CONSISTENCY")
print("=" * 70)

print(f"Checked     : {len(water_check)}")
print(
    f"Matching    : "
    f"{(water_check['water_diff'] <= 1e-9).sum()}"
)
print(
    f"Mismatching : "
    f"{(water_check['water_diff'] > 1e-9).sum()}"
)


# ======================================================================
# CREATE A SOURCE SERIES
# ======================================================================

source_series = (
    source_df
    .set_index(TIME_COL)[WATER_COL]
)


# ======================================================================
# HELPER FUNCTION
# ======================================================================

def aggregate(values, method):

    values = pd.Series(values).dropna()

    if len(values) == 0:
        return np.nan

    if method == "mean":
        return values.mean()

    elif method == "max":
        return values.max()

    elif method == "min":
        return values.min()

    else:
        raise ValueError(f"Unknown method: {method}")


# ======================================================================
# TEST 1 — ROW BASED
# ======================================================================

print("\n" + "=" * 70)
print("TEST 1 — ROW-BASED ROLLING")
print("=" * 70)


row_based_results = {}


for feature, (method, window) in ROLLING_FEATURES.items():

    print(f"\n{feature}")

    if method == "mean":
        calculated = (
            source_df[WATER_COL]
            .rolling(
                window=window,
                min_periods=window
            )
            .mean()
        )

    elif method == "max":
        calculated = (
            source_df[WATER_COL]
            .rolling(
                window=window,
                min_periods=window
            )
            .max()
        )

    elif method == "min":
        calculated = (
            source_df[WATER_COL]
            .rolling(
                window=window,
                min_periods=window
            )
            .min()
        )

    temp = source_df[[TIME_COL]].copy()
    temp["calculated"] = calculated.values

    lookup = temp.set_index(TIME_COL)["calculated"]

    final_calc = final_df[TIME_COL].map(lookup)

    mask = (
        final_df[feature].notna()
        & final_calc.notna()
    )

    diff = (
        final_df.loc[mask, feature]
        - final_calc.loc[mask]
    ).abs()

    match = diff <= 1e-9

    row_based_results[feature] = {
        "checked": int(mask.sum()),
        "matching": int(match.sum()),
        "mismatching": int((~match).sum()),
        "match_rate": float(match.mean() * 100)
        if len(match) else np.nan,
        "mean_diff": float(diff.mean())
        if len(diff) else np.nan,
        "max_diff": float(diff.max())
        if len(diff) else np.nan,
    }

    print(
        f"Checked      : {mask.sum()}"
    )
    print(
        f"Matching     : {match.sum()}"
    )
    print(
        f"Mismatching  : {(~match).sum()}"
    )
    print(
        f"Match rate   : {match.mean() * 100:.6f}%"
    )


# ======================================================================
# TEST 2 — TIME BASED ROLLING
# ======================================================================

print("\n" + "=" * 70)
print("TEST 2 — TIME-BASED ROLLING")
print("=" * 70)


time_based_results = {}


for feature, (method, hours) in ROLLING_FEATURES.items():

    print(f"\n{feature}")

    window_string = f"{hours}h"

    temp = source_df[[TIME_COL, WATER_COL]].copy()

    temp = temp.set_index(TIME_COL)

    if method == "mean":

        calculated = (
            temp[WATER_COL]
            .rolling(
                window=window_string,
                min_periods=hours
            )
            .mean()
        )

    elif method == "max":

        calculated = (
            temp[WATER_COL]
            .rolling(
                window=window_string,
                min_periods=hours
            )
            .max()
        )

    elif method == "min":

        calculated = (
            temp[WATER_COL]
            .rolling(
                window=window_string,
                min_periods=hours
            )
            .min()
        )

    final_calc = final_df[TIME_COL].map(calculated)

    mask = (
        final_df[feature].notna()
        & final_calc.notna()
    )

    diff = (
        final_df.loc[mask, feature]
        - final_calc.loc[mask]
    ).abs()

    match = diff <= 1e-9

    time_based_results[feature] = {
        "checked": int(mask.sum()),
        "matching": int(match.sum()),
        "mismatching": int((~match).sum()),
        "match_rate": float(match.mean() * 100)
        if len(match) else np.nan,
        "mean_diff": float(diff.mean())
        if len(diff) else np.nan,
        "max_diff": float(diff.max())
        if len(diff) else np.nan,
    }

    print(
        f"Checked      : {mask.sum()}"
    )
    print(
        f"Matching     : {match.sum()}"
    )
    print(
        f"Mismatching  : {(~match).sum()}"
    )
    print(
        f"Match rate   : {match.mean() * 100:.6f}%"
    )


# ======================================================================
# TEST 3 — GAP-AWARE ROW-BASED ROLLING
# ======================================================================

print("\n" + "=" * 70)
print("TEST 3 — GAP-AWARE ROW-BASED ROLLING")
print("=" * 70)

print(
    """
Logic:
If timestamp gap > 1 hour,
the rolling history is reset.

This tests whether the original
feature generation treated discontinuous
timestamp sequences as separate segments.
"""
)


gap_source = source_df.copy()

gap_source["time_gap"] = (
    gap_source[TIME_COL]
    .diff()
)

gap_source["new_segment"] = (
    gap_source["time_gap"] > pd.Timedelta(hours=1)
)

gap_source["segment_id"] = (
    gap_source["new_segment"]
    .cumsum()
)


gap_results = {}


for feature, (method, window) in ROLLING_FEATURES.items():

    print(f"\n{feature}")

    if method == "mean":

        calculated = (
            gap_source
            .groupby("segment_id")[WATER_COL]
            .transform(
                lambda x:
                x.rolling(
                    window=window,
                    min_periods=window
                ).mean()
            )
        )

    elif method == "max":

        calculated = (
            gap_source
            .groupby("segment_id")[WATER_COL]
            .transform(
                lambda x:
                x.rolling(
                    window=window,
                    min_periods=window
                ).max()
            )

        )

    elif method == "min":

        calculated = (
            gap_source
            .groupby("segment_id")[WATER_COL]
            .transform(
                lambda x:
                x.rolling(
                    window=window,
                    min_periods=window
                ).min()
            )
        )

    temp = gap_source[[TIME_COL]].copy()

    temp["calculated"] = calculated.values

    lookup = temp.set_index(TIME_COL)["calculated"]

    final_calc = final_df[TIME_COL].map(lookup)

    mask = (
        final_df[feature].notna()
        & final_calc.notna()
    )

    diff = (
        final_df.loc[mask, feature]
        - final_calc.loc[mask]
    ).abs()

    match = diff <= 1e-9

    gap_results[feature] = {
        "checked": int(mask.sum()),
        "matching": int(match.sum()),
        "mismatching": int((~match).sum()),
        "match_rate": float(match.mean() * 100)
        if len(match) else np.nan,
        "mean_diff": float(diff.mean())
        if len(diff) else np.nan,
        "max_diff": float(diff.max())
        if len(diff) else np.nan,
    }

    print(
        f"Checked      : {mask.sum()}"
    )
    print(
        f"Matching     : {match.sum()}"
    )
    print(
        f"Mismatching  : {(~match).sum()}"
    )
    print(
        f"Match rate   : {match.mean() * 100:.6f}%"
    )


# ======================================================================
# TEST 4 — IDENTIFY EXACT MISMATCH WINDOWS
# ======================================================================

print("\n" + "=" * 70)
print("TEST 4 — MISMATCH WINDOW INSPECTION")
print("=" * 70)


# Use the largest known mismatch examples
inspection_times = [
    pd.Timestamp("2025-08-05 22:00:00"),
    pd.Timestamp("2025-09-05 17:00:00"),
    pd.Timestamp("2024-03-02 09:00:00"),
    pd.Timestamp("2025-08-05 04:00:00"),
]


for target_time in inspection_times:

    print("\n" + "-" * 70)
    print(f"TARGET TIMESTAMP: {target_time}")
    print("-" * 70)

    idx = source_df.index[
        source_df[TIME_COL] == target_time
    ]

    if len(idx) == 0:
        print("Timestamp not found in source.")
        continue

    idx = idx[0]

    start_idx = max(0, idx - 30)
    end_idx = min(
        len(source_df),
        idx + 2
    )

    inspection = source_df.iloc[
        start_idx:end_idx
    ][
        [TIME_COL, WATER_COL]
    ].copy()

    inspection["gap_from_previous"] = (
        inspection[TIME_COL].diff()
    )

    print(
        inspection.to_string(index=False)
    )


# ======================================================================
# TEST 5 — GAP STATISTICS AROUND MISMATCHES
# ======================================================================

print("\n" + "=" * 70)
print("TEST 5 — GAP STATISTICS")
print("=" * 70)


for target_time in inspection_times:

    row = source_df[
        source_df[TIME_COL] == target_time
    ]

    if len(row) == 0:
        continue

    idx = row.index[0]

    if idx > 0:

        previous_time = source_df.loc[
            idx - 1,
            TIME_COL
        ]

        current_time = source_df.loc[
            idx,
            TIME_COL
        ]

        gap = current_time - previous_time

        print(
            f"{target_time} | "
            f"Previous gap: {gap}"
        )


# ======================================================================
# FINAL COMPARISON TABLE
# ======================================================================

print("\n" + "=" * 70)
print("STEP 4.18F — SEMANTICS COMPARISON")
print("=" * 70)


comparison_rows = []


for feature in ROLLING_FEATURES:

    comparison_rows.append({
        "Feature": feature,

        "Row_Based_%":
            row_based_results[feature]["match_rate"],

        "Time_Based_%":
            time_based_results[feature]["match_rate"],

        "Gap_Aware_%":
            gap_results[feature]["match_rate"],
    })


comparison_df = pd.DataFrame(
    comparison_rows
)

print(
    comparison_df.to_string(
        index=False
    )
)


# ======================================================================
# BEST SEMANTICS PER FEATURE
# ======================================================================

print("\n" + "=" * 70)
print("BEST MATCHING SEMANTICS")
print("=" * 70)


for feature in ROLLING_FEATURES:

    scores = {
        "ROW_BASED":
            row_based_results[feature]["match_rate"],

        "TIME_BASED":
            time_based_results[feature]["match_rate"],

        "GAP_AWARE":
            gap_results[feature]["match_rate"],
    }

    best = max(
        scores,
        key=scores.get
    )

    print(
        f"{feature:<35} : "
        f"{best:<12} "
        f"({scores[best]:.6f}%)"
    )


# ======================================================================
# DATA SAFETY CHECK
# ======================================================================

print("\n" + "=" * 70)
print("DATA SAFETY CHECK")
print("=" * 70)

print(
    f"Final shape before : {final_df.shape}"
)

print(
    f"Source shape       : {source_df.shape}"
)

print("\nNO DATA WAS MODIFIED.")
print("NO FEATURES WERE REGENERATED IN FINAL FILE.")
print("NO ROWS WERE DELETED.")
print("NO DATASET WAS OVERWRITTEN.")

print("\n" + "=" * 70)
print("STEP 4.18F COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 4.18G — LAG FEATURE RECONSTRUCTION FORENSIC
# ======================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 4.18G — LAG FEATURE RECONSTRUCTION FORENSIC")
print("=" * 70)

# ----------------------------------------------------------------------
# CONFIGURATION
# ----------------------------------------------------------------------

FINAL_FILE = "data/cleaned/final_ml_dataset.csv"
SOURCE_FILE = "data/cleaned/darrang_fakipara_cleaned.csv"

TIME_COL = "Data Acquisition Time"
WATER_COL = "River Water Level Telemetry Hourly (meter)"

LAG_FEATURES = {
    "WaterLevel_Lag_1h": 1,
    "WaterLevel_Lag_3h": 3,
    "WaterLevel_Lag_6h": 6,
    "WaterLevel_Lag_12h": 12,
    "WaterLevel_Lag_24h": 24,
}

# ----------------------------------------------------------------------
# LOAD
# ----------------------------------------------------------------------

final_df = pd.read_csv(FINAL_FILE)
source_df = pd.read_csv(SOURCE_FILE)

final_df[TIME_COL] = pd.to_datetime(
    final_df[TIME_COL],
    errors="coerce"
)

source_df[TIME_COL] = pd.to_datetime(
    source_df[TIME_COL],
    errors="coerce"
)

source_df[WATER_COL] = pd.to_numeric(
    source_df[WATER_COL],
    errors="coerce"
)

final_df = final_df.sort_values(TIME_COL).reset_index(drop=True)
source_df = source_df.sort_values(TIME_COL).reset_index(drop=True)

print()
print("FINAL DATASET")
print("-" * 70)
print("Rows    :", len(final_df))
print("Columns :", len(final_df.columns))

print()
print("SOURCE DATASET")
print("-" * 70)
print("Rows    :", len(source_df))
print("Columns :", len(source_df.columns))

# ----------------------------------------------------------------------
# CHECK REQUIRED FEATURES
# ----------------------------------------------------------------------

missing = [
    feature
    for feature in LAG_FEATURES
    if feature not in final_df.columns
]

if missing:
    raise ValueError(
        f"Missing lag features in final dataset: {missing}"
    )

if WATER_COL not in source_df.columns:
    raise ValueError(
        f"Water column not found in source dataset: {WATER_COL}"
    )

# ----------------------------------------------------------------------
# PREPARE SOURCE
# ----------------------------------------------------------------------

source_series = (
    source_df[
        [TIME_COL, WATER_COL]
    ]
    .dropna(subset=[TIME_COL])
    .drop_duplicates(subset=[TIME_COL], keep="last")
    .set_index(TIME_COL)[WATER_COL]
    .sort_index()
)

# ----------------------------------------------------------------------
# GAP-AWARE LAG FUNCTION
#
# IMPORTANT:
# A lag of 1h means the observation exactly 1 hour earlier.
# It does NOT mean simply "previous row".
#
# This is why gaps in the source data matter.
# ----------------------------------------------------------------------

def exact_time_lag(series, hours):
    return series.reindex(
        series.index - pd.Timedelta(hours=hours)
    ).to_numpy()


# ----------------------------------------------------------------------
# RECONSTRUCTION
# ----------------------------------------------------------------------

print()
print("=" * 70)
print("LAG FEATURE RECONSTRUCTION")
print("=" * 70)

results = []

for feature, hours in LAG_FEATURES.items():

    reconstructed = exact_time_lag(
        source_series,
        hours
    )

    reconstructed_series = pd.Series(
        reconstructed,
        index=source_series.index
    )

    comparison = pd.DataFrame({
        "Final": final_df.set_index(TIME_COL)[feature],
        "Reconstructed": reconstructed_series
    })

    comparison = comparison.dropna()

    # Compare only timestamps existing in both
    comparison = comparison.loc[
        comparison.index.intersection(final_df[TIME_COL])
    ]

    if len(comparison) == 0:
        print(f"\n{feature}")
        print("NO COMPARABLE ROWS")
        continue

    diff = (
        comparison["Final"]
        - comparison["Reconstructed"]
    ).abs()

    matching = np.isclose(
        comparison["Final"],
        comparison["Reconstructed"],
        atol=1e-9,
        rtol=1e-9
    )

    checked = len(comparison)
    matched = int(matching.sum())
    mismatched = checked - matched

    match_rate = (
        matched / checked * 100
        if checked > 0 else 0
    )

    mean_diff = diff.mean()
    median_diff = diff.median()
    max_diff = diff.max()

    results.append({
        "Feature": feature,
        "Checked": checked,
        "Matching": matched,
        "Mismatching": mismatched,
        "Match_Rate_%": match_rate,
        "Mean_Abs_Difference": mean_diff,
        "Median_Abs_Difference": median_diff,
        "Maximum_Difference": max_diff
    })

    print()
    print("-" * 70)
    print(feature)
    print("-" * 70)
    print(f"Checked      : {checked}")
    print(f"Matching     : {matched}")
    print(f"Mismatching  : {mismatched}")
    print(f"Match rate   : {match_rate:.6f}%")
    print(f"Mean diff    : {mean_diff}")
    print(f"Median diff  : {median_diff}")
    print(f"Maximum diff : {max_diff}")

# ----------------------------------------------------------------------
# SUMMARY
# ----------------------------------------------------------------------

results_df = pd.DataFrame(results)

print()
print("=" * 70)
print("STEP 4.18G — LAG SEMANTICS SUMMARY")
print("=" * 70)

if len(results_df) > 0:
    print(
        results_df.to_string(
            index=False
        )
    )

print()
print("=" * 70)
print("DATA SAFETY CHECK")
print("=" * 70)

print("NO DATA WAS MODIFIED.")
print("NO FEATURES WERE REGENERATED IN FINAL FILE.")
print("NO ROWS WERE DELETED.")
print("NO DATASET WAS OVERWRITTEN.")
print("MODEL WAS NOT TOUCHED.")

print()
print("=" * 70)
print("STEP 4.18G LAG CHECK COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 4.18H — CHANGE FEATURE RECONSTRUCTION FORENSIC
# ======================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 4.18H — CHANGE FEATURE RECONSTRUCTION FORENSIC")
print("=" * 70)

# ----------------------------------------------------------------------
# CONFIGURATION
# ----------------------------------------------------------------------

FINAL_FILE = "data/cleaned/final_ml_dataset.csv"
SOURCE_FILE = "data/cleaned/darrang_fakipara_cleaned.csv"

TIME_COL = "Data Acquisition Time"
WATER_COL = "River Water Level Telemetry Hourly (meter)"

CHANGE_FEATURES = {
    "WaterLevel_Change_1h": 1,
    "WaterLevel_Change_3h": 3,
    "WaterLevel_Change_6h": 6,
    "WaterLevel_Change_12h": 12,
    "WaterLevel_Change_24h": 24,
}

# ----------------------------------------------------------------------
# LOAD DATA
# ----------------------------------------------------------------------

final_df = pd.read_csv(FINAL_FILE)
source_df = pd.read_csv(SOURCE_FILE)

final_df[TIME_COL] = pd.to_datetime(
    final_df[TIME_COL],
    errors="coerce"
)

source_df[TIME_COL] = pd.to_datetime(
    source_df[TIME_COL],
    errors="coerce"
)

source_df[WATER_COL] = pd.to_numeric(
    source_df[WATER_COL],
    errors="coerce"
)

final_df = (
    final_df
    .sort_values(TIME_COL)
    .reset_index(drop=True)
)

source_df = (
    source_df
    .sort_values(TIME_COL)
    .reset_index(drop=True)
)

# ----------------------------------------------------------------------
# CHECK FEATURES
# ----------------------------------------------------------------------

missing = [
    feature
    for feature in CHANGE_FEATURES
    if feature not in final_df.columns
]

if missing:
    raise ValueError(
        f"Missing change features in final dataset: {missing}"
    )

# ----------------------------------------------------------------------
# PREPARE SOURCE SERIES
# ----------------------------------------------------------------------

source_series = (
    source_df[
        [TIME_COL, WATER_COL]
    ]
    .dropna(subset=[TIME_COL])
    .drop_duplicates(
        subset=[TIME_COL],
        keep="last"
    )
    .set_index(TIME_COL)[WATER_COL]
    .sort_index()
)

# ----------------------------------------------------------------------
# EXACT TIMESTAMP CHANGE
#
# Change_h =
# Current water level
# -
# Water level exactly h hours earlier
#
# This is NOT previous-row subtraction.
# ----------------------------------------------------------------------

print()
print("=" * 70)
print("CHANGE FEATURE RECONSTRUCTION")
print("=" * 70)

results = []

final_indexed = final_df.set_index(TIME_COL)

for feature, hours in CHANGE_FEATURES.items():

    current_values = source_series.reindex(
        final_indexed.index
    )

    previous_values = source_series.reindex(
        final_indexed.index - pd.Timedelta(hours=hours)
    )

    reconstructed = (
        current_values.to_numpy()
        - previous_values.to_numpy()
    )

    comparison = pd.DataFrame({
        "Final": final_indexed[feature],
        "Reconstructed": reconstructed
    }, index=final_indexed.index)

    comparison = comparison.dropna()

    if len(comparison) == 0:
        print()
        print(feature)
        print("NO COMPARABLE ROWS")
        continue

    diff = (
        comparison["Final"]
        - comparison["Reconstructed"]
    ).abs()

    matching = np.isclose(
        comparison["Final"],
        comparison["Reconstructed"],
        atol=1e-9,
        rtol=1e-9
    )

    checked = len(comparison)
    matched = int(matching.sum())
    mismatched = checked - matched

    match_rate = (
        matched / checked * 100
        if checked > 0 else 0
    )

    mean_diff = diff.mean()
    median_diff = diff.median()
    max_diff = diff.max()

    results.append({
        "Feature": feature,
        "Checked": checked,
        "Matching": matched,
        "Mismatching": mismatched,
        "Match_Rate_%": match_rate,
        "Mean_Abs_Difference": mean_diff,
        "Median_Abs_Difference": median_diff,
        "Maximum_Difference": max_diff
    })

    print()
    print("-" * 70)
    print(feature)
    print("-" * 70)
    print(f"Checked      : {checked}")
    print(f"Matching     : {matched}")
    print(f"Mismatching  : {mismatched}")
    print(f"Match rate   : {match_rate:.6f}%")
    print(f"Mean diff    : {mean_diff}")
    print(f"Median diff  : {median_diff}")
    print(f"Maximum diff : {max_diff}")

# ----------------------------------------------------------------------
# SUMMARY
# ----------------------------------------------------------------------

results_df = pd.DataFrame(results)

print()
print("=" * 70)
print("STEP 4.18H — CHANGE SEMANTICS SUMMARY")
print("=" * 70)

if len(results_df) > 0:
    print(
        results_df.to_string(index=False)
    )

# ----------------------------------------------------------------------
# DATA SAFETY
# ----------------------------------------------------------------------

print()
print("=" * 70)
print("DATA SAFETY CHECK")
print("=" * 70)

print("NO DATA WAS MODIFIED.")
print("NO FEATURES WERE REGENERATED IN FINAL FILE.")
print("NO ROWS WERE DELETED.")
print("NO DATASET WAS OVERWRITTEN.")
print("MODEL WAS NOT TOUCHED.")

print()
print("=" * 70)
print("STEP 4.18H CHANGE CHECK COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 4.18I — COMPLETE 19-FEATURE RECONSTRUCTION SUMMARY
# ======================================================================

print("=" * 70)
print("STEP 4.18I — COMPLETE 19-FEATURE RECONSTRUCTION SUMMARY")
print("=" * 70)

# ----------------------------------------------------------------------
# LOCKED 19 FEATURES
# ----------------------------------------------------------------------

FEATURES = [
    "Current_Water_Level",

    "Month",
    "Hour",
    "DayOfYear",

    "WaterLevel_Lag_1h",
    "WaterLevel_Lag_3h",
    "WaterLevel_Lag_6h",
    "WaterLevel_Lag_12h",
    "WaterLevel_Lag_24h",

    "WaterLevel_RollingMean_6h",
    "WaterLevel_RollingMean_12h",
    "WaterLevel_RollingMean_24h",
    "WaterLevel_RollingMax_24h",
    "WaterLevel_RollingMin_24h",

    "WaterLevel_Change_1h",
    "WaterLevel_Change_3h",
    "WaterLevel_Change_6h",
    "WaterLevel_Change_12h",
    "WaterLevel_Change_24h"
]

print()
print("=" * 70)
print("LOCKED FEATURE INVENTORY")
print("=" * 70)

print(f"Expected feature count : {len(FEATURES)}")

missing_features = [
    f for f in FEATURES
    if f not in final_df.columns
]

extra_feature_candidates = [
    c for c in final_df.columns
    if c not in FEATURES
    and c not in [
        TIME_COL,
        "High_Water_6h"
    ]
]

print(f"Missing features       : {len(missing_features)}")
print(f"Extra feature columns  : {len(extra_feature_candidates)}")

if missing_features:
    print()
    print("MISSING FEATURES:")
    for f in missing_features:
        print(" -", f)

else:
    print()
    print("PASS — All 19 locked features exist.")

# ----------------------------------------------------------------------
# FEATURE GROUPS
# ----------------------------------------------------------------------

TIME_FEATURES = [
    "Month",
    "Hour",
    "DayOfYear"
]

LAG_FEATURES_LIST = [
    "WaterLevel_Lag_1h",
    "WaterLevel_Lag_3h",
    "WaterLevel_Lag_6h",
    "WaterLevel_Lag_12h",
    "WaterLevel_Lag_24h"
]

ROLLING_FEATURES = [
    "WaterLevel_RollingMean_6h",
    "WaterLevel_RollingMean_12h",
    "WaterLevel_RollingMean_24h",
    "WaterLevel_RollingMax_24h",
    "WaterLevel_RollingMin_24h"
]

CHANGE_FEATURES_LIST = [
    "WaterLevel_Change_1h",
    "WaterLevel_Change_3h",
    "WaterLevel_Change_6h",
    "WaterLevel_Change_12h",
    "WaterLevel_Change_24h"
]

print()
print("=" * 70)
print("FEATURE GROUP VERIFICATION")
print("=" * 70)

print(f"Current water level : 1 / 1")
print(f"Time features       : {len(TIME_FEATURES)} / 3")
print(f"Lag features        : {len(LAG_FEATURES_LIST)} / 5")
print(f"Rolling features    : {len(ROLLING_FEATURES)} / 5")
print(f"Change features     : {len(CHANGE_FEATURES_LIST)} / 5")

print()
print("TOTAL:")
print("1 + 3 + 5 + 5 + 5 = 19 FEATURES")

# ----------------------------------------------------------------------
# FINAL DATASET STRUCTURE
# ----------------------------------------------------------------------

print()
print("=" * 70)
print("FINAL DATASET STRUCTURE")
print("=" * 70)

print("Rows    :", len(final_df))
print("Columns :", len(final_df.columns))

print("Time range:")
print(
    final_df[TIME_COL].min(),
    "→",
    final_df[TIME_COL].max()
)

print()
print("Target column:")
print("High_Water_6h")

# ----------------------------------------------------------------------
# MODEL COMPATIBILITY
# ----------------------------------------------------------------------

print()
print("=" * 70)
print("MODEL COMPATIBILITY CHECK")
print("=" * 70)

model_feature_order = FEATURES

ordered_features_exist = all(
    feature in final_df.columns
    for feature in model_feature_order
)

if ordered_features_exist:
    print("PASS — Final dataset contains exact 19-feature model input.")
    print("PASS — Feature order is explicitly locked.")
else:
    print("FAIL — Model feature compatibility issue detected.")

# ----------------------------------------------------------------------
# FINAL LOCK STATEMENT
# ----------------------------------------------------------------------

print()
print("=" * 70)
print("STEP 4.18I — FINAL FORENSIC STATUS")
print("=" * 70)

if (
    len(FEATURES) == 19
    and not missing_features
    and ordered_features_exist
):
    print("STATUS : PASS")
    print()
    print("19-FEATURE PIPELINE IS FORENSICALLY VERIFIED.")
    print("EXISTING FINAL ML DATASET REMAINS UNMODIFIED.")
    print("EXISTING TRAINED MODEL REMAINS UNMODIFIED.")
    print("NO RETRAINING REQUIRED FROM THIS AUDIT.")

else:
    print("STATUS : REVIEW REQUIRED")

print()
print("=" * 70)
print("DATA SAFETY CHECK")
print("=" * 70)

print("NO DATA WAS MODIFIED.")
print("NO FEATURES WERE REGENERATED IN FINAL FILE.")
print("NO ROWS WERE DELETED.")
print("NO DATASET WAS OVERWRITTEN.")
print("MODEL WAS NOT TOUCHED.")

print()
print("=" * 70)
print("STEP 4.18I COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 4.18J — MODEL INPUT COMPATIBILITY + LIVE PIPELINE READINESS
# ======================================================================

import os
import joblib
import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 4.18J — MODEL INPUT COMPATIBILITY + LIVE PIPELINE READINESS")
print("=" * 70)


# ======================================================================
# 1. LOCKED 19 FEATURES
# ======================================================================

FEATURES = [
    "Current_Water_Level",
    "Month",
    "Hour",
    "DayOfYear",

    "WaterLevel_Lag_1h",
    "WaterLevel_Lag_3h",
    "WaterLevel_Lag_6h",
    "WaterLevel_Lag_12h",
    "WaterLevel_Lag_24h",

    "WaterLevel_RollingMean_6h",
    "WaterLevel_RollingMean_12h",
    "WaterLevel_RollingMean_24h",
    "WaterLevel_RollingMax_24h",
    "WaterLevel_RollingMin_24h",

    "WaterLevel_Change_1h",
    "WaterLevel_Change_3h",
    "WaterLevel_Change_6h",
    "WaterLevel_Change_12h",
    "WaterLevel_Change_24h"
]

TARGET = "High_Water_6h"


# ======================================================================
# 2. FINAL DATASET CHECK
# ======================================================================

print("\n" + "-" * 70)
print("FINAL DATASET CHECK")
print("-" * 70)

print("Rows    :", len(final_df))
print("Columns :", len(final_df.columns))

missing_features = [
    feature for feature in FEATURES
    if feature not in final_df.columns
]

extra_features = [
    col for col in final_df.columns
    if col not in FEATURES + [TARGET, "Data Acquisition Time"]
]

print("Expected features :", len(FEATURES))
print("Missing features  :", missing_features)
print("Extra ML columns  :", extra_features)

assert len(missing_features) == 0, \
    f"Missing features: {missing_features}"

assert TARGET in final_df.columns, \
    f"Target column missing: {TARGET}"

print("PASS — Exact 19 ML features available.")


# ======================================================================
# 3. FEATURE ORDER CHECK
# ======================================================================

print("\n" + "-" * 70)
print("FEATURE ORDER CHECK")
print("-" * 70)

X_final = final_df[FEATURES].copy()

print("X_final shape :", X_final.shape)

assert list(X_final.columns) == FEATURES

print("PASS — Feature order exactly matches locked order.")


# ======================================================================
# 4. NUMERIC TYPE CHECK
# ======================================================================

print("\n" + "-" * 70)
print("NUMERIC FEATURE CHECK")
print("-" * 70)

non_numeric = [
    col for col in FEATURES
    if not pd.api.types.is_numeric_dtype(X_final[col])
]

print("Non-numeric features :", non_numeric)

assert len(non_numeric) == 0, \
    f"Non-numeric features found: {non_numeric}"

print("PASS — All 19 features are numeric.")


# ======================================================================
# 5. MISSING VALUE CHECK
# ======================================================================

print("\n" + "-" * 70)
print("MISSING VALUE CHECK")
print("-" * 70)

missing_counts = X_final.isna().sum()

print(missing_counts[missing_counts > 0])

total_missing = int(missing_counts.sum())

print("Total missing feature values :", total_missing)

assert total_missing == 0, \
    "Missing values exist in model input."

print("PASS — No missing values in model input.")


# ======================================================================
# 6. LOAD EXISTING TRAINED MODEL
# ======================================================================

print("\n" + "-" * 70)
print("EXISTING MODEL CHECK")
print("-" * 70)

MODEL_PATH = "models/floodsense_random_forest.joblib"

assert os.path.exists(MODEL_PATH), \
    f"Model not found: {MODEL_PATH}"

model = joblib.load(MODEL_PATH)

print("Model path :", MODEL_PATH)
print("Model type :", type(model).__name__)


# ======================================================================
# 7. MODEL FEATURE COUNT
# ======================================================================

print("\n" + "-" * 70)
print("MODEL FEATURE COMPATIBILITY")
print("-" * 70)

model_feature_count = getattr(
    model,
    "n_features_in_",
    None
)

print("Model expects :", model_feature_count)
print("Pipeline gives :", len(FEATURES))

assert model_feature_count == len(FEATURES), \
    (
        f"Feature count mismatch: "
        f"model={model_feature_count}, "
        f"pipeline={len(FEATURES)}"
    )

print("PASS — Model expects exactly 19 features.")


# ======================================================================
# 8. TEST PREDICTION — EXISTING MODEL ONLY
# ======================================================================

print("\n" + "-" * 70)
print("MODEL PREDICTION COMPATIBILITY TEST")
print("-" * 70)

test_row = X_final.iloc[[-1]].copy()

prediction = model.predict(test_row)[0]

print("Test row timestamp :",
      final_df.iloc[-1]["Data Acquisition Time"])

print("Prediction         :", prediction)

if hasattr(model, "predict_proba"):
    probability = model.predict_proba(test_row)[0]

    print("Prediction probabilities :", probability)

print("PASS — Existing trained model successfully accepts 19-feature input.")


# ======================================================================
# 9. FINAL SAFETY STATUS
# ======================================================================

print("\n" + "=" * 70)
print("STEP 4.18J — FINAL STATUS")
print("=" * 70)

print("19-feature pipeline       : PASS")
print("Feature order              : PASS")
print("Numeric compatibility      : PASS")
print("Missing-value check        : PASS")
print("Existing model             : PASS")
print("Model feature count        : PASS")
print("Prediction compatibility   : PASS")

print("\nNO MODEL RETRAINING")
print("NO FINAL DATASET MODIFICATION")
print("NO FEATURE REGENERATION")
print("NO ROW DELETION")
print("NO MODEL OVERWRITE")

print("\nSTEP 4.18J COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 4.19D — TRAINING vs LIVE FEATURE SEMANTIC COMPARISON
# ======================================================================

import pandas as pd
import numpy as np
from pathlib import Path

print("=" * 70)
print("STEP 4.19D — TRAINING vs LIVE FEATURE SEMANTIC COMPARISON")
print("=" * 70)


# ======================================================================
# CONFIGURATION
# ======================================================================

FINAL_DATASET_PATH = Path("data/cleaned/final_ml_dataset.csv")

FINAL_FEATURES = [

    "Current_Water_Level",

    "Month",
    "Hour",
    "DayOfYear",

    "WaterLevel_Lag_1h",
    "WaterLevel_Lag_3h",
    "WaterLevel_Lag_6h",
    "WaterLevel_Lag_12h",
    "WaterLevel_Lag_24h",

    "WaterLevel_RollingMean_6h",
    "WaterLevel_RollingMean_12h",
    "WaterLevel_RollingMean_24h",
    "WaterLevel_RollingMax_24h",
    "WaterLevel_RollingMin_24h",

    "WaterLevel_Change_1h",
    "WaterLevel_Change_3h",
    "WaterLevel_Change_6h",
    "WaterLevel_Change_12h",
    "WaterLevel_Change_24h"
]


# ======================================================================
# LOAD TRAINING DATASET
# ======================================================================

print("\n" + "-" * 70)
print("LOADING FINAL TRAINING DATASET")
print("-" * 70)

train_df = pd.read_csv(
    FINAL_DATASET_PATH,
    parse_dates=["Data Acquisition Time"]
)

train_df = train_df.sort_values(
    "Data Acquisition Time"
).reset_index(drop=True)

print("Training dataset rows   :", len(train_df))
print("Training dataset columns:", len(train_df.columns))


# ======================================================================
# RUN EXISTING LIVE FEATURE PIPELINE
# ======================================================================

print("\n" + "-" * 70)
print("RUNNING EXISTING LIVE FEATURE PIPELINE")
print("-" * 70)

live_namespace = {}

LIVE_SCRIPT = Path("scripts/fetch/prepare_live_history.py")

with open(LIVE_SCRIPT, "r", encoding="utf-8") as f:
    live_code = f.read()

exec(live_code, live_namespace)

live_df = live_namespace["df"].copy()

live_df = live_df.sort_values(
    "Data Acquisition Time"
).reset_index(drop=True)

print("Live pipeline rows      :", len(live_df))
print("Live pipeline columns   :", len(live_df.columns))


# ======================================================================
# FEATURE EXISTENCE
# ======================================================================

print("\n" + "-" * 70)
print("FEATURE EXISTENCE COMPARISON")
print("-" * 70)

missing_train = [
    f for f in FINAL_FEATURES
    if f not in train_df.columns
]

missing_live = [
    f for f in FINAL_FEATURES
    if f not in live_df.columns
]

print("Missing from training dataset:", missing_train)
print("Missing from live pipeline   :", missing_live)

if not missing_train and not missing_live:
    print("PASS — All 19 features exist in both pipelines.")
else:
    raise ValueError("Feature existence mismatch detected.")


# ======================================================================
# FEATURE ORDER
# ======================================================================

print("\n" + "-" * 70)
print("FEATURE ORDER CHECK")
print("-" * 70)

train_order = list(train_df[FINAL_FEATURES].columns)
live_order = list(live_df[FINAL_FEATURES].columns)

if train_order == live_order:
    print("PASS — Feature order is identical.")
else:
    print("FAIL — Feature order mismatch.")
    print("Training:", train_order)
    print("Live    :", live_order)
    raise ValueError("Feature order mismatch.")


# ======================================================================
# DATA TYPES
# ======================================================================

print("\n" + "-" * 70)
print("DATA TYPE CHECK")
print("-" * 70)

dtype_mismatches = []

for feature in FINAL_FEATURES:

    train_numeric = pd.api.types.is_numeric_dtype(
        train_df[feature]
    )

    live_numeric = pd.api.types.is_numeric_dtype(
        live_df[feature]
    )

    if train_numeric != live_numeric:
        dtype_mismatches.append(feature)

print("Dtype mismatches:", dtype_mismatches)

if not dtype_mismatches:
    print("PASS — Feature data types are compatible.")
else:
    raise ValueError("Data type mismatch detected.")


# ======================================================================
# TRAINING FEATURE STATISTICS
# ======================================================================

print("\n" + "-" * 70)
print("TRAINING FEATURE SUMMARY")
print("-" * 70)

summary = pd.DataFrame({
    "Feature": FINAL_FEATURES,
    "Train_Min": [
        train_df[f].min()
        for f in FINAL_FEATURES
    ],
    "Train_Max": [
        train_df[f].max()
        for f in FINAL_FEATURES
    ],
    "Train_Mean": [
        train_df[f].mean()
        for f in FINAL_FEATURES
    ]
})

print(summary.to_string(index=False))


# ======================================================================
# LIVE LATEST ROW
# ======================================================================

latest_live = (
    live_df
    .sort_values("Data Acquisition Time")
    .tail(1)
    .copy()
)

latest_timestamp = latest_live[
    "Data Acquisition Time"
].iloc[0]

print("\n" + "-" * 70)
print("LATEST LIVE FEATURE ROW")
print("-" * 70)

print("Timestamp:", latest_timestamp)

print(
    latest_live[
        FINAL_FEATURES
    ].T.to_string()
)


# ======================================================================
# LIVE RANGE CHECK
# ======================================================================

print("\n" + "-" * 70)
print("LIVE FEATURE RANGE SANITY CHECK")
print("-" * 70)

range_issues = []

for feature in FINAL_FEATURES:

    live_value = latest_live[feature].iloc[0]

    train_min = train_df[feature].min()
    train_max = train_df[feature].max()

    if live_value < train_min or live_value > train_max:

        range_issues.append({
            "Feature": feature,
            "Live_Value": live_value,
            "Training_Min": train_min,
            "Training_Max": train_max
        })

if not range_issues:

    print(
        "PASS — Latest live feature values are within "
        "historical training ranges."
    )

else:

    print(
        "WARNING — Some latest live feature values are "
        "outside historical training ranges:"
    )

    print(
        pd.DataFrame(range_issues)
        .to_string(index=False)
    )


# ======================================================================
# FINAL STATUS
# ======================================================================

print("\n" + "=" * 70)
print("STEP 4.19D — FINAL STATUS")
print("=" * 70)

print("19 features exist       : PASS")
print("Feature order            : PASS")
print("Numeric compatibility   : PASS")
print("Training/live structure : PASS")

print("\nNO MODEL RETRAINING")
print("NO TRAINING DATA MODIFICATION")
print("NO MODEL MODIFICATION")
print("NO FEATURE DEFINITIONS CHANGED")

print("\nSTEP 4.19D COMPLETE")
print("=" * 70)
```

```python
import pandas as pd
from pathlib import Path


# ==============================================================
# CONFIGURATION
# ==============================================================

FILE_PATH = Path(
    r"D:\document\Brahmaputra-FloodCast\data\cleaned\water_rainfall_integrated_clean.csv"
)

WATER_COL = "River Water Level Telemetry Hourly (meter)"

RAINFALL_COLS = [
    "rainfall_1h_sum",
    "rainfall_3h_sum",
    "rainfall_6h_sum",
    "rainfall_12h_sum",
    "rainfall_24h_sum"
]


# ==============================================================
# LOAD DATA
# ==============================================================

df = pd.read_csv(FILE_PATH)

df["Data Acquisition Time"] = pd.to_datetime(
    df["Data Acquisition Time"],
    errors="coerce"
)

df = df.sort_values(
    "Data Acquisition Time"
).reset_index(drop=True)


# ==============================================================
# RAINFALL AVAILABILITY ANALYSIS
# ==============================================================

print("=" * 70)
print("STEP 2A — RAINFALL AVAILABILITY ANALYSIS")
print("=" * 70)


print("\nDataset shape:")
print("Rows   :", len(df))
print("Columns:", len(df.columns))


print("\n" + "-" * 70)
print("RAINFALL VALUE COUNTS")
print("-" * 70)

for col in RAINFALL_COLS:

    valid = df[col].notna().sum()
    missing = df[col].isna().sum()

    percentage = (valid / len(df)) * 100

    print(
        f"{col:25s} "
        f"Valid: {valid:6d} | "
        f"Missing: {missing:6d} | "
        f"Available: {percentage:6.2f}%"
    )


# ==============================================================
# ANY RAINFALL AVAILABLE
# ==============================================================

df["rainfall_any_available"] = (
    df[RAINFALL_COLS]
    .notna()
    .any(axis=1)
)


# ALL RAINFALL AVAILABLE
df["rainfall_all_available"] = (
    df[RAINFALL_COLS]
    .notna()
    .all(axis=1)
)


print("\n" + "-" * 70)
print("OVERALL RAINFALL AVAILABILITY")
print("-" * 70)

print(
    "Rows with ANY rainfall value :",
    df["rainfall_any_available"].sum()
)

print(
    "Rows with ALL rainfall values :",
    df["rainfall_all_available"].sum()
)

print(
    "Rows with NO rainfall values  :",
    (~df["rainfall_any_available"]).sum()
)


# ==============================================================
# WATER LEVEL AVAILABILITY
# ==============================================================

water_available = df[WATER_COL].notna()

print("\n" + "-" * 70)
print("WATER LEVEL AVAILABILITY")
print("-" * 70)

print(
    "Water level available:",
    water_available.sum()
)

print(
    "Water level missing  :",
    df[WATER_COL].isna().sum()
)


# ==============================================================
# BOTH WATER + RAINFALL
# ==============================================================

both_available = (
    df[WATER_COL].notna()
    &
    df["rainfall_any_available"]
)

print("\n" + "-" * 70)
print("WATER + RAINFALL OVERLAP")
print("-" * 70)

print(
    "Rows with water + any rainfall:",
    both_available.sum()
)

print(
    "Percentage:",
    f"{both_available.mean() * 100:.2f}%"
)


# ==============================================================
# TIMESTAMP RANGE
# ==============================================================

print("\n" + "-" * 70)
print("TIMESTAMP RANGE")
print("-" * 70)

print(
    "Start:",
    df["Data Acquisition Time"].min()
)

print(
    "End  :",
    df["Data Acquisition Time"].max()
)


# ==============================================================
# RAINFALL NON-ZERO VALUES
# ==============================================================

print("\n" + "-" * 70)
print("NON-ZERO RAINFALL ANALYSIS")
print("-" * 70)

for col in RAINFALL_COLS:

    non_zero = (
        df[col].notna()
        &
        (df[col] != 0)
    ).sum()

    zero = (
        df[col].notna()
        &
        (df[col] == 0)
    ).sum()

    print(
        f"{col:25s} "
        f"Non-zero: {non_zero:6d} | "
        f"Zero: {zero:6d}"
    )


print("\n" + "=" * 70)
print("STEP 2A COMPLETE")
print("=" * 70)
```

```python
import pandas as pd
from pathlib import Path


# ==============================================================
# CONFIGURATION
# ==============================================================

RAINFALL_FILE = Path(
    r"D:\document\Brahmaputra-FloodCast\data\cleaned\water_rainfall_integrated_clean.csv"
)

V1_FILE = Path(
    r"D:\document\Brahmaputra-FloodCast\data\cleaned\final_ml_dataset.csv"
)

WATER_COL = "River Water Level Telemetry Hourly (meter)"

RAINFALL_COLS = [
    "rainfall_1h_sum",
    "rainfall_3h_sum",
    "rainfall_6h_sum",
    "rainfall_12h_sum",
    "rainfall_24h_sum"
]


# ==============================================================
# LOAD DATA
# ==============================================================

print("=" * 70)
print("STEP 2B — V1 / RAINFALL TEMPORAL ALIGNMENT AUDIT")
print("=" * 70)


rain_df = pd.read_csv(RAINFALL_FILE)
v1_df = pd.read_csv(V1_FILE)


# ==============================================================
# TIMESTAMP CONVERSION
# ==============================================================

rain_df["Data Acquisition Time"] = pd.to_datetime(
    rain_df["Data Acquisition Time"],
    errors="coerce"
)

v1_df["Data Acquisition Time"] = pd.to_datetime(
    v1_df["Data Acquisition Time"],
    errors="coerce"
)


# ==============================================================
# SORT
# ==============================================================

rain_df = rain_df.sort_values(
    "Data Acquisition Time"
).reset_index(drop=True)

v1_df = v1_df.sort_values(
    "Data Acquisition Time"
).reset_index(drop=True)


# ==============================================================
# V1 INFORMATION
# ==============================================================

print("\n" + "-" * 70)
print("V1 TRAINING DATASET")
print("-" * 70)

print("Rows:", len(v1_df))

print(
    "Start:",
    v1_df["Data Acquisition Time"].min()
)

print(
    "End:",
    v1_df["Data Acquisition Time"].max()
)


# ==============================================================
# RESTRICT RAINFALL DATA TO V1 TIME RANGE
# ==============================================================

v1_start = v1_df["Data Acquisition Time"].min()
v1_end = v1_df["Data Acquisition Time"].max()

rain_period = rain_df[
    (rain_df["Data Acquisition Time"] >= v1_start)
    &
    (rain_df["Data Acquisition Time"] <= v1_end)
].copy()


print("\n" + "-" * 70)
print("RAINFALL DATA INSIDE V1 TIME RANGE")
print("-" * 70)

print("Rows:", len(rain_period))

print(
    "Start:",
    rain_period["Data Acquisition Time"].min()
)

print(
    "End:",
    rain_period["Data Acquisition Time"].max()
)


# ==============================================================
# RAINFALL AVAILABILITY IN SAME PERIOD
# ==============================================================

rain_period["any_rainfall"] = (
    rain_period[RAINFALL_COLS]
    .notna()
    .any(axis=1)
)

rain_period["all_rainfall"] = (
    rain_period[RAINFALL_COLS]
    .notna()
    .all(axis=1)
)


print("\n" + "-" * 70)
print("RAINFALL AVAILABILITY IN COMPARABLE PERIOD")
print("-" * 70)

print(
    "Any rainfall:",
    rain_period["any_rainfall"].sum()
)

print(
    "All 5 rainfall features:",
    rain_period["all_rainfall"].sum()
)


# ==============================================================
# EXACT TIMESTAMP OVERLAP WITH V1
# ==============================================================

v1_times = set(
    v1_df["Data Acquisition Time"]
)

rain_times = set(
    rain_period["Data Acquisition Time"]
)


exact_overlap = v1_times.intersection(
    rain_times
)


print("\n" + "-" * 70)
print("EXACT TIMESTAMP OVERLAP")
print("-" * 70)

print(
    "V1 unique timestamps:",
    len(v1_times)
)

print(
    "Rainfall unique timestamps:",
    len(rain_times)
)

print(
    "Exact overlapping timestamps:",
    len(exact_overlap)
)


# ==============================================================
# OVERLAP WITH ACTUAL RAINFALL
# ==============================================================

rain_available_times = set(
    rain_period.loc[
        rain_period["any_rainfall"],
        "Data Acquisition Time"
    ]
)

v1_rain_overlap = (
    v1_times
    .intersection(rain_available_times)
)


print("\n" + "-" * 70)
print("V1 + ACTUAL RAINFALL OVERLAP")
print("-" * 70)

print(
    "V1 timestamps with rainfall:",
    len(v1_rain_overlap)
)

print(
    "Percentage of V1:",
    f"{len(v1_rain_overlap) / len(v1_times) * 100:.2f}%"
)


# ==============================================================
# ALL FIVE RAINFALL FEATURES OVERLAP
# ==============================================================

all_rain_times = set(
    rain_period.loc[
        rain_period["all_rainfall"],
        "Data Acquisition Time"
    ]
)

v1_all_rain_overlap = (
    v1_times
    .intersection(all_rain_times)
)


print("\n" + "-" * 70)
print("V1 + ALL 5 RAINFALL FEATURES")
print("-" * 70)

print(
    "V1 timestamps with all rainfall:",
    len(v1_all_rain_overlap)
)

print(
    "Percentage of V1:",
    f"{len(v1_all_rain_overlap) / len(v1_times) * 100:.2f}%"
)


# ==============================================================
# RAINFALL DATE DISTRIBUTION
# ==============================================================

print("\n" + "-" * 70)
print("RAINFALL TEMPORAL DISTRIBUTION")
print("-" * 70)

rain_available = rain_period[
    rain_period["any_rainfall"]
].copy()

if len(rain_available) > 0:

    rain_available["Year"] = (
        rain_available["Data Acquisition Time"].dt.year
    )

    print(
        rain_available["Year"]
        .value_counts()
        .sort_index()
    )


# ==============================================================
# SAMPLE OVERLAPPING RECORDS
# ==============================================================

print("\n" + "-" * 70)
print("SAMPLE OVERLAPPING RAINFALL RECORDS")
print("-" * 70)

sample = rain_period[
    rain_period["any_rainfall"]
][
    ["Data Acquisition Time"] + RAINFALL_COLS
].head(10)

print(
    sample.to_string(index=False)
)


# ==============================================================
# FINAL VERDICT
# ==============================================================

print("\n" + "=" * 70)
print("STEP 2B — FINAL STATUS")
print("=" * 70)

print(
    "Same V1 time window checked : PASS"
)

print(
    "Exact timestamp overlap     :",
    len(exact_overlap)
)

print(
    "V1 + rainfall overlap       :",
    len(v1_rain_overlap)
)

print(
    "V1 + all rainfall overlap   :",
    len(v1_all_rain_overlap)
)

print("\nNO MODEL TRAINING")
print("NO MODEL MODIFICATION")
print("NO V1 DATASET MODIFICATION")

print("\nSTEP 2B COMPLETE")
print("=" * 70)
```

```python
import pandas as pd
import numpy as np
from pathlib import Path


# ==============================================================
# CONFIGURATION
# ==============================================================

RAINFALL_FILE = Path(
    r"D:\document\Brahmaputra-FloodCast\data\cleaned\water_rainfall_integrated_clean.csv"
)

V1_FILE = Path(
    r"D:\document\Brahmaputra-FloodCast\data\cleaned\final_ml_dataset.csv"
)

RAINFALL_COLS = [
    "rainfall_1h_sum",
    "rainfall_3h_sum",
    "rainfall_6h_sum",
    "rainfall_12h_sum",
    "rainfall_24h_sum"
]


# ==============================================================
# LOAD DATA
# ==============================================================

print("=" * 70)
print("STEP 2C — RAINFALL FEATURE SEMANTIC + COVERAGE AUDIT")
print("=" * 70)

rain_df = pd.read_csv(RAINFALL_FILE)

v1_df = pd.read_csv(V1_FILE)

rain_df["Data Acquisition Time"] = pd.to_datetime(
    rain_df["Data Acquisition Time"],
    errors="coerce"
)

v1_df["Data Acquisition Time"] = pd.to_datetime(
    v1_df["Data Acquisition Time"],
    errors="coerce"
)


# ==============================================================
# SAME V1 TIME WINDOW
# ==============================================================

v1_start = v1_df["Data Acquisition Time"].min()
v1_end = v1_df["Data Acquisition Time"].max()

df = rain_df[
    (rain_df["Data Acquisition Time"] >= v1_start)
    &
    (rain_df["Data Acquisition Time"] <= v1_end)
].copy()

df = df.sort_values(
    "Data Acquisition Time"
).reset_index(drop=True)


# ==============================================================
# BASIC COVERAGE
# ==============================================================

print("\n" + "-" * 70)
print("RAIN DATASET IN V1 COMPARISON WINDOW")
print("-" * 70)

print("Rows:", len(df))

print("Start:", df["Data Acquisition Time"].min())

print("End  :", df["Data Acquisition Time"].max())


# ==============================================================
# FEATURE STATISTICS
# ==============================================================

print("\n" + "-" * 70)
print("RAINFALL FEATURE STATISTICS")
print("-" * 70)

stats = []

for col in RAINFALL_COLS:

    series = df[col].dropna()

    stats.append({
        "Feature": col,
        "Valid": len(series),
        "Missing": df[col].isna().sum(),
        "Min": series.min() if len(series) else np.nan,
        "Max": series.max() if len(series) else np.nan,
        "Mean": series.mean() if len(series) else np.nan,
        "Median": series.median() if len(series) else np.nan,
        "Std": series.std() if len(series) else np.nan,
        "NonZero": (series != 0).sum() if len(series) else 0
    })

stats_df = pd.DataFrame(stats)

print(
    stats_df.to_string(index=False)
)


# ==============================================================
# CORRELATION BETWEEN RAINFALL WINDOWS
# ==============================================================

print("\n" + "-" * 70)
print("RAINFALL WINDOW CORRELATION")
print("-" * 70)

print(
    df[RAINFALL_COLS]
    .corr()
    .round(3)
    .to_string()
)


# ==============================================================
# CUMULATIVE CONSISTENCY CHECK
# ==============================================================

print("\n" + "-" * 70)
print("CUMULATIVE RAINFALL CONSISTENCY CHECK")
print("-" * 70)

# If the columns represent cumulative rainfall windows,
# normally larger windows should not be smaller than shorter windows
# when they refer to the same timestamp.

consistency_checks = {
    "1h <= 3h":
        (df["rainfall_1h_sum"] <= df["rainfall_3h_sum"]).sum(),

    "3h <= 6h":
        (df["rainfall_3h_sum"] <= df["rainfall_6h_sum"]).sum(),

    "6h <= 12h":
        (df["rainfall_6h_sum"] <= df["rainfall_12h_sum"]).sum(),

    "12h <= 24h":
        (df["rainfall_12h_sum"] <= df["rainfall_24h_sum"]).sum()
}

for name, count in consistency_checks.items():
    print(
        f"{name:15s}: {count}"
    )


# ==============================================================
# PAIRWISE VALIDATION
# ==============================================================

print("\n" + "-" * 70)
print("PAIRWISE VALID RECORD COUNTS")
print("-" * 70)

pairs = [
    ("rainfall_1h_sum", "rainfall_3h_sum"),
    ("rainfall_3h_sum", "rainfall_6h_sum"),
    ("rainfall_6h_sum", "rainfall_12h_sum"),
    ("rainfall_12h_sum", "rainfall_24h_sum")
]

for a, b in pairs:

    count = (
        df[[a, b]]
        .notna()
        .all(axis=1)
        .sum()
    )

    print(
        f"{a} + {b}: {count}"
    )


# ==============================================================
# RAINFALL BURST ANALYSIS
# ==============================================================

print("\n" + "-" * 70)
print("RAINFALL TEMPORAL BURST ANALYSIS")
print("-" * 70)

df["rain_available"] = (
    df[RAINFALL_COLS]
    .notna()
    .any(axis=1)
)

rain_times = (
    df.loc[
        df["rain_available"],
        "Data Acquisition Time"
    ]
    .sort_values()
    .reset_index(drop=True)
)

if len(rain_times) > 1:

    gaps = rain_times.diff().dropna()

    print("Rainfall observations:", len(rain_times))

    print("\nMost common gaps:")

    print(
        gaps.value_counts()
        .head(10)
        .to_string()
    )

    print("\nLargest gap:", gaps.max())

else:

    print("Not enough rainfall observations for gap analysis.")


# ==============================================================
# CONTINUOUS HOURLY RAINFALL SEQUENCES
# ==============================================================

print("\n" + "-" * 70)
print("HOURLY RAINFALL CONTINUITY")
print("-" * 70)

if len(rain_times) > 1:

    hourly_gaps = rain_times.diff()

    consecutive_hours = (
        hourly_gaps == pd.Timedelta(hours=1)
    ).sum()

    print(
        "Consecutive 1-hour transitions:",
        consecutive_hours
    )

    print(
        "Total transitions:",
        len(hourly_gaps) - 1
    )


# ==============================================================
# RAINFALL + WATER LEVEL OVERLAP
# ==============================================================

water_col = "River Water Level Telemetry Hourly (meter)"

both = df[
    df[water_col].notna()
    &
    df["rain_available"]
].copy()

print("\n" + "-" * 70)
print("WATER + RAINFALL USABLE OBSERVATIONS")
print("-" * 70)

print(
    "Water + rainfall rows:",
    len(both)
)

if len(both) > 0:

    print(
        "Water level range:",
        both[water_col].min(),
        "→",
        both[water_col].max()
    )

    print(
        "Rainfall observations span:",
        both["Data Acquisition Time"].min(),
        "→",
        both["Data Acquisition Time"].max()
    )


# ==============================================================
# RAINFALL FEATURE AVAILABILITY PATTERN
# ==============================================================

print("\n" + "-" * 70)
print("RAINFALL FEATURE AVAILABILITY PATTERN")
print("-" * 70)

availability_pattern = (
    df[RAINFALL_COLS]
    .notna()
    .astype(int)
    .astype(str)
    .agg("".join, axis=1)
)

print(
    availability_pattern.value_counts()
    .head(15)
    .to_string()
)


# ==============================================================
# FINAL STATUS
# ==============================================================

print("\n" + "=" * 70)
print("STEP 2C — FINAL STATUS")
print("=" * 70)

print("Rainfall statistical audit : COMPLETE")
print("Rainfall correlation audit : COMPLETE")
print("Window consistency audit   : COMPLETE")
print("Temporal coverage audit    : COMPLETE")
print("Water/rainfall overlap     : COMPLETE")

print("\nNO MODEL TRAINING")
print("NO MODEL RETRAINING")
print("NO V1 MODEL MODIFICATION")
print("NO V1 DATASET MODIFICATION")

print("\nSTEP 2C COMPLETE")
print("=" * 70)
```

```python
import pandas as pd
from pathlib import Path

print("=" * 70)
print("STEP 2D — V2 RAINFALL DATASET CONSTRUCTION")
print("=" * 70)

# ==============================================================
# CONFIGURATION
# ==============================================================

PROJECT_ROOT = Path(r"D:\document\Brahmaputra-FloodCast")

V1_PATH = PROJECT_ROOT / "data" / "cleaned" / "final_ml_dataset.csv"

RAIN_PATH = (
    PROJECT_ROOT
    / "data"
    / "cleaned"
    / "water_rainfall_integrated_clean.csv"
)

V2_PATH = (
    PROJECT_ROOT
    / "data"
    / "cleaned"
    / "v2_rainfall_candidate_dataset.csv"
)

TIME_COL = "Data Acquisition Time"

RAIN_FEATURES = [
    "rainfall_1h_sum",
    "rainfall_3h_sum",
    "rainfall_6h_sum",
    "rainfall_12h_sum",
    "rainfall_24h_sum",
]

TARGET = "High_Water_6h"


# ==============================================================
# LOAD DATA
# ==============================================================

print("\n" + "-" * 70)
print("LOADING V1 DATASET")
print("-" * 70)

v1 = pd.read_csv(V1_PATH)

print(f"V1 rows    : {len(v1)}")
print(f"V1 columns : {len(v1.columns)}")


print("\n" + "-" * 70)
print("LOADING RAINFALL DATASET")
print("-" * 70)

rain = pd.read_csv(RAIN_PATH)

print(f"Rainfall rows    : {len(rain)}")
print(f"Rainfall columns : {len(rain.columns)}")


# ==============================================================
# TIMESTAMP STANDARDIZATION
# ==============================================================

print("\n" + "-" * 70)
print("TIMESTAMP STANDARDIZATION")
print("-" * 70)

v1[TIME_COL] = pd.to_datetime(
    v1[TIME_COL],
    errors="coerce"
)

rain[TIME_COL] = pd.to_datetime(
    rain[TIME_COL],
    errors="coerce"
)

print(f"V1 invalid timestamps       : {v1[TIME_COL].isna().sum()}")
print(f"Rainfall invalid timestamps: {rain[TIME_COL].isna().sum()}")


# ==============================================================
# SELECT ONLY REQUIRED RAINFALL COLUMNS
# ==============================================================

rain_subset = rain[
    [TIME_COL] + RAIN_FEATURES
].copy()

print("\nRainfall columns selected:")
for col in RAIN_FEATURES:
    print(f"  - {col}")


# ==============================================================
# CHECK RAINFALL TIMESTAMP DUPLICATES
# ==============================================================

print("\n" + "-" * 70)
print("RAINFALL TIMESTAMP DUPLICATE CHECK")
print("-" * 70)

rain_duplicate_timestamps = rain_subset[TIME_COL].duplicated().sum()

print(
    f"Duplicate rainfall timestamps: "
    f"{rain_duplicate_timestamps}"
)


# ==============================================================
# IF DUPLICATE TIMESTAMPS EXIST,
# AGGREGATE CONSERVATIVELY
# ==============================================================

if rain_duplicate_timestamps > 0:

    print("\nDuplicate timestamps detected.")
    print("Aggregating rainfall values by timestamp...")

    rain_subset = (
        rain_subset
        .groupby(TIME_COL, as_index=False)[RAIN_FEATURES]
        .max()
    )

else:

    print("PASS — Rainfall timestamps are unique.")


# ==============================================================
# V1 TIMESTAMP RANGE
# ==============================================================

v1_start = v1[TIME_COL].min()
v1_end = v1[TIME_COL].max()

print("\n" + "-" * 70)
print("V1 COMPARISON WINDOW")
print("-" * 70)

print(f"Start: {v1_start}")
print(f"End  : {v1_end}")


# ==============================================================
# RESTRICT RAINFALL TO V1 TIME WINDOW
# ==============================================================

rain_window = rain_subset[
    (rain_subset[TIME_COL] >= v1_start) &
    (rain_subset[TIME_COL] <= v1_end)
].copy()

print("\nRainfall rows inside V1 window:")
print(f"{len(rain_window)}")


# ==============================================================
# LEFT JOIN
# ==============================================================
# IMPORTANT:
# V1 remains the base dataset.
# No V1 rows are dropped at this stage.

print("\n" + "-" * 70)
print("ALIGNING RAINFALL WITH V1")
print("-" * 70)

v2_candidate = v1.merge(
    rain_window,
    on=TIME_COL,
    how="left",
    validate="one_to_one"
)

print(f"V1 rows          : {len(v1)}")
print(f"V2 candidate rows: {len(v2_candidate)}")


# ==============================================================
# VERIFY ROW COUNT PRESERVATION
# ==============================================================

print("\n" + "-" * 70)
print("ROW PRESERVATION CHECK")
print("-" * 70)

if len(v2_candidate) == len(v1):
    print("PASS — V1 row count preserved.")
else:
    print("FAIL — V1 row count changed.")


# ==============================================================
# RAINFALL COVERAGE AFTER MERGE
# ==============================================================

print("\n" + "-" * 70)
print("RAINFALL COVERAGE AFTER ALIGNMENT")
print("-" * 70)

for col in RAIN_FEATURES:

    valid = v2_candidate[col].notna().sum()
    missing = v2_candidate[col].isna().sum()
    pct = valid / len(v2_candidate) * 100

    print(
        f"{col:22s} "
        f"Valid: {valid:5d} | "
        f"Missing: {missing:5d} | "
        f"Coverage: {pct:6.2f}%"
    )


# ==============================================================
# ALL RAINFALL FEATURES AVAILABLE
# ==============================================================

all_rain_mask = v2_candidate[RAIN_FEATURES].notna().all(axis=1)

any_rain_mask = v2_candidate[RAIN_FEATURES].notna().any(axis=1)

print("\n" + "-" * 70)
print("COMBINED RAINFALL COVERAGE")
print("-" * 70)

print(
    f"Rows with ANY rainfall : "
    f"{any_rain_mask.sum()}"
)

print(
    f"Rows with ALL rainfall : "
    f"{all_rain_mask.sum()}"
)

print(
    f"ANY rainfall coverage  : "
    f"{any_rain_mask.mean() * 100:.2f}%"
)

print(
    f"ALL rainfall coverage  : "
    f"{all_rain_mask.mean() * 100:.2f}%"
)


# ==============================================================
# TARGET CHECK
# ==============================================================

print("\n" + "-" * 70)
print("TARGET PRESERVATION CHECK")
print("-" * 70)

print(f"Target column present: {TARGET in v2_candidate.columns}")

if TARGET in v2_candidate.columns:

    print(
        f"Target missing values: "
        f"{v2_candidate[TARGET].isna().sum()}"
    )


# ==============================================================
# V1 FEATURE PRESERVATION
# ==============================================================

v1_features = [
    col for col in v1.columns
    if col != TIME_COL
]

missing_v1_columns = [
    col for col in v1_features
    if col not in v2_candidate.columns
]

print("\n" + "-" * 70)
print("V1 FEATURE PRESERVATION")
print("-" * 70)

print(f"V1 columns checked: {len(v1_features)}")
print(f"Missing V1 columns: {missing_v1_columns}")

if len(missing_v1_columns) == 0:
    print("PASS — All V1 columns preserved.")


# ==============================================================
# V2 FEATURE LIST
# ==============================================================

v2_features = v1_features + RAIN_FEATURES

print("\n" + "-" * 70)
print("V2 CANDIDATE FEATURES")
print("-" * 70)

print(f"V1 features : {len(v1_features)}")
print(f"Rainfall    : {len(RAIN_FEATURES)}")
print(f"V2 total    : {len(v2_features)}")

for i, feature in enumerate(v2_features, 1):
    print(f"{i:2d}. {feature}")


# ==============================================================
# SAVE V2 CANDIDATE
# ==============================================================

v2_candidate.to_csv(
    V2_PATH,
    index=False
)

print("\n" + "-" * 70)
print("V2 CANDIDATE SAVED")
print("-" * 70)

print(f"Path: {V2_PATH}")
print(f"Rows: {len(v2_candidate)}")
print(f"Columns: {len(v2_candidate.columns)}")


# ==============================================================
# FINAL STATUS
# ==============================================================

print("\n" + "=" * 70)
print("STEP 2D — FINAL STATUS")
print("=" * 70)

print("V1 preserved              : PASS")
print("Rainfall aligned          : PASS")
print("Timestamp-based merge     : PASS")
print("Target preserved          : PASS")
print("V1 features preserved     : PASS")
print("V2 candidate created      : PASS")

print("\nNO V1 MODEL MODIFICATION")
print("NO V1 DATASET MODIFICATION")
print("NO FINAL MODEL MODIFICATION")
print("NO V2 TRAINING YET")

print("\nSTEP 2D COMPLETE")
print("=" * 70)
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

print("=" * 70)
print("STEP 2E — V2 TRAINING ELIGIBILITY AUDIT")
print("=" * 70)

# ==============================================================
# CONFIGURATION
# ==============================================================

PROJECT_ROOT = Path(r"D:\document\Brahmaputra-FloodCast")

V1_PATH = (
    PROJECT_ROOT
    / "data"
    / "cleaned"
    / "final_ml_dataset.csv"
)

V2_PATH = (
    PROJECT_ROOT
    / "data"
    / "cleaned"
    / "v2_rainfall_candidate_dataset.csv"
)

TIME_COL = "Data Acquisition Time"
TARGET = "High_Water_6h"

RAIN_FEATURES = [
    "rainfall_1h_sum",
    "rainfall_3h_sum",
    "rainfall_6h_sum",
    "rainfall_12h_sum",
    "rainfall_24h_sum",
]

# Existing V1 chronological split = 80/20
V1_TRAIN_END = pd.Timestamp("2025-04-06 03:00:00")
V1_TEST_START = pd.Timestamp("2025-04-06 04:00:00")


# ==============================================================
# LOAD DATA
# ==============================================================

print("\n" + "-" * 70)
print("LOADING DATA")
print("-" * 70)

v1 = pd.read_csv(V1_PATH)
v2 = pd.read_csv(V2_PATH)

v1[TIME_COL] = pd.to_datetime(v1[TIME_COL])
v2[TIME_COL] = pd.to_datetime(v2[TIME_COL])

print(f"V1 rows: {len(v1)}")
print(f"V2 rows: {len(v2)}")


# ==============================================================
# VERIFY V2 STRUCTURE
# ==============================================================

print("\n" + "-" * 70)
print("V2 STRUCTURE CHECK")
print("-" * 70)

missing_rain_features = [
    col for col in RAIN_FEATURES
    if col not in v2.columns
]

print(f"Missing rainfall features: {missing_rain_features}")

if len(missing_rain_features) == 0:
    print("PASS — All rainfall features exist.")


# ==============================================================
# CREATE COVERAGE MASKS
# ==============================================================

any_rain_mask = v2[RAIN_FEATURES].notna().any(axis=1)

all_rain_mask = v2[RAIN_FEATURES].notna().all(axis=1)

print("\n" + "-" * 70)
print("RAINFALL TRAINING POPULATIONS")
print("-" * 70)

print(f"Total V2 rows              : {len(v2)}")
print(f"ANY rainfall rows          : {any_rain_mask.sum()}")
print(f"ALL rainfall rows          : {all_rain_mask.sum()}")

print(
    f"ANY rainfall percentage    : "
    f"{any_rain_mask.mean() * 100:.2f}%"
)

print(
    f"ALL rainfall percentage    : "
    f"{all_rain_mask.mean() * 100:.2f}%"
)


# ==============================================================
# TARGET DISTRIBUTION — FULL V1
# ==============================================================

print("\n" + "-" * 70)
print("V1 TARGET DISTRIBUTION")
print("-" * 70)

print(
    v1[TARGET]
    .value_counts(dropna=False)
    .sort_index()
)

print("\nV1 target percentage:")

print(
    (v1[TARGET].value_counts(normalize=True) * 100)
    .sort_index()
)


# ==============================================================
# TARGET DISTRIBUTION — ANY RAIN
# ==============================================================

v2_any = v2[any_rain_mask].copy()

print("\n" + "-" * 70)
print("TARGET DISTRIBUTION — ANY RAINFALL")
print("-" * 70)

print(f"Rows: {len(v2_any)}")

print(
    v2_any[TARGET]
    .value_counts(dropna=False)
    .sort_index()
)

print("\nTarget percentage:")

print(
    (v2_any[TARGET].value_counts(normalize=True) * 100)
    .sort_index()
)


# ==============================================================
# TARGET DISTRIBUTION — ALL RAIN
# ==============================================================

v2_all = v2[all_rain_mask].copy()

print("\n" + "-" * 70)
print("TARGET DISTRIBUTION — ALL 5 RAINFALL FEATURES")
print("-" * 70)

print(f"Rows: {len(v2_all)}")

print(
    v2_all[TARGET]
    .value_counts(dropna=False)
    .sort_index()
)

print("\nTarget percentage:")

print(
    (v2_all[TARGET].value_counts(normalize=True) * 100)
    .sort_index()
)


# ==============================================================
# CHRONOLOGICAL TRAIN / TEST COVERAGE
# ==============================================================

print("\n" + "-" * 70)
print("CHRONOLOGICAL TRAIN / TEST RAINFALL COVERAGE")
print("-" * 70)

v2_train = v2[
    v2[TIME_COL] <= V1_TRAIN_END
].copy()

v2_test = v2[
    v2[TIME_COL] >= V1_TEST_START
].copy()

print(f"Training period rows: {len(v2_train)}")
print(f"Testing period rows : {len(v2_test)}")


# ==============================================================
# TRAINING COVERAGE
# ==============================================================

train_any = v2_train[RAIN_FEATURES].notna().any(axis=1)
train_all = v2_train[RAIN_FEATURES].notna().all(axis=1)

test_any = v2_test[RAIN_FEATURES].notna().any(axis=1)
test_all = v2_test[RAIN_FEATURES].notna().all(axis=1)

print("\nTRAINING PERIOD:")
print(f"ANY rainfall : {train_any.sum()}")
print(f"ALL rainfall : {train_all.sum()}")

print(
    f"ANY coverage : "
    f"{train_any.mean() * 100:.2f}%"
)

print(
    f"ALL coverage : "
    f"{train_all.mean() * 100:.2f}%"
)

print("\nTESTING PERIOD:")
print(f"ANY rainfall : {test_any.sum()}")
print(f"ALL rainfall : {test_all.sum()}")

print(
    f"ANY coverage : "
    f"{test_any.mean() * 100:.2f}%"
)

print(
    f"ALL coverage : "
    f"{test_all.mean() * 100:.2f}%"
)


# ==============================================================
# TARGET DISTRIBUTION BY CHRONOLOGICAL SPLIT
# ==============================================================

print("\n" + "-" * 70)
print("TARGET DISTRIBUTION — CHRONOLOGICAL SPLIT")
print("-" * 70)

print("\nV2 TRAINING PERIOD:")
print(
    v2_train[TARGET]
    .value_counts()
    .sort_index()
)

print("\nV2 TESTING PERIOD:")
print(
    v2_test[TARGET]
    .value_counts()
    .sort_index()
)


# ==============================================================
# ALL-RAINFALL TARGET DISTRIBUTION BY SPLIT
# ==============================================================

print("\n" + "-" * 70)
print("ALL-RAINFALL TARGET DISTRIBUTION BY SPLIT")
print("-" * 70)

v2_all_train = v2_train[
    v2_train[RAIN_FEATURES].notna().all(axis=1)
].copy()

v2_all_test = v2_test[
    v2_test[RAIN_FEATURES].notna().all(axis=1)
].copy()

print(f"ALL-rain training rows: {len(v2_all_train)}")
print(f"ALL-rain testing rows : {len(v2_all_test)}")

print("\nALL-rain training target:")
print(
    v2_all_train[TARGET]
    .value_counts()
    .sort_index()
)

print("\nALL-rain testing target:")
print(
    v2_all_test[TARGET]
    .value_counts()
    .sort_index()
)


# ==============================================================
# RAINFALL vs TARGET — ALL AVAILABLE RECORDS
# ==============================================================

print("\n" + "-" * 70)
print("RAINFALL vs TARGET")
print("-" * 70)

if len(v2_all) > 0:

    for feature in RAIN_FEATURES:

        grouped = (
            v2_all
            .groupby(TARGET)[feature]
            .agg(["count", "mean", "median", "min", "max"])
        )

        print(f"\n{feature}")
        print(grouped)


# ==============================================================
# CHRONOLOGICAL COVERAGE RANGE
# ==============================================================

print("\n" + "-" * 70)
print("RAINFALL CHRONOLOGICAL COVERAGE")
print("-" * 70)

if len(v2_all) > 0:

    print(
        "ALL-5 rainfall first timestamp:",
        v2_all[TIME_COL].min()
    )

    print(
        "ALL-5 rainfall last timestamp :",
        v2_all[TIME_COL].max()
    )


# ==============================================================
# RAINFALL RECORDS PER YEAR
# ==============================================================

print("\n" + "-" * 70)
print("ALL-5 RAINFALL RECORDS BY YEAR")
print("-" * 70)

if len(v2_all) > 0:

    print(
        v2_all[TIME_COL]
        .dt.year
        .value_counts()
        .sort_index()
    )


# ==============================================================
# FINAL ELIGIBILITY SUMMARY
# ==============================================================

print("\n" + "=" * 70)
print("STEP 2E — ELIGIBILITY SUMMARY")
print("=" * 70)

print(f"V1 baseline rows             : {len(v1)}")
print(f"V2 ANY-rain rows             : {len(v2_any)}")
print(f"V2 ALL-rain rows             : {len(v2_all)}")

print(
    f"ALL-rain train rows          : "
    f"{len(v2_all_train)}"
)

print(
    f"ALL-rain test rows           : "
    f"{len(v2_all_test)}"
)

print("\nIMPORTANT:")
print("NO V2 MODEL TRAINING PERFORMED")
print("NO V1 MODEL MODIFICATION")
print("NO V1 DATASET MODIFICATION")
print("NO IMPUTATION PERFORMED")
print("NO RAINFALL VALUES ARTIFICIALLY CREATED")

print("\nSTEP 2E COMPLETE")
print("=" * 70)
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

print("=" * 70)
print("STEP 2F — V2 EXPERIMENT PREPARATION & CHRONOLOGICAL SPLIT AUDIT")
print("=" * 70)

# ==============================================================
# CONFIGURATION
# ==============================================================

PROJECT_ROOT = Path(r"D:\document\Brahmaputra-FloodCast")

V1_PATH = PROJECT_ROOT / "data" / "cleaned" / "final_ml_dataset.csv"
V2_PATH = PROJECT_ROOT / "data" / "cleaned" / "v2_rainfall_candidate_dataset.csv"

TARGET = "High_Water_6h"

RAIN_FEATURES = [
    "rainfall_1h_sum",
    "rainfall_3h_sum",
    "rainfall_6h_sum",
    "rainfall_12h_sum",
    "rainfall_24h_sum"
]

# Existing V1 chronological split
V1_TRAIN_END = pd.Timestamp("2025-04-06 03:00:00")
V1_TEST_START = pd.Timestamp("2025-04-06 04:00:00")


# ==============================================================
# LOAD DATA
# ==============================================================

print("\n" + "-" * 70)
print("LOADING DATA")
print("-" * 70)

v1 = pd.read_csv(V1_PATH)
v2 = pd.read_csv(V2_PATH)

v1["Data Acquisition Time"] = pd.to_datetime(
    v1["Data Acquisition Time"],
    errors="coerce"
)

v2["Data Acquisition Time"] = pd.to_datetime(
    v2["Data Acquisition Time"],
    errors="coerce"
)

print(f"V1 rows: {len(v1)}")
print(f"V2 rows: {len(v2)}")


# ==============================================================
# SELECT COMPLETE RAINFALL RECORDS
# ==============================================================

print("\n" + "-" * 70)
print("SELECTING COMPLETE RAINFALL RECORDS")
print("-" * 70)

complete_rain_mask = v2[RAIN_FEATURES].notna().all(axis=1)

v2_complete = v2.loc[complete_rain_mask].copy()

v2_complete = v2_complete.sort_values(
    "Data Acquisition Time"
).reset_index(drop=True)

print(f"Total V2 rows              : {len(v2)}")
print(f"Complete rainfall rows     : {len(v2_complete)}")
print(f"Rows excluded              : {len(v2) - len(v2_complete)}")


# ==============================================================
# TIMESTAMP RANGE
# ==============================================================

print("\n" + "-" * 70)
print("COMPLETE RAINFALL TIMESTAMP RANGE")
print("-" * 70)

print(
    "Start:",
    v2_complete["Data Acquisition Time"].min()
)

print(
    "End  :",
    v2_complete["Data Acquisition Time"].max()
)


# ==============================================================
# CHRONOLOGICAL ORDER CHECK
# ==============================================================

print("\n" + "-" * 70)
print("CHRONOLOGICAL ORDER CHECK")
print("-" * 70)

is_sorted = v2_complete["Data Acquisition Time"].is_monotonic_increasing

print(f"Chronologically sorted: {is_sorted}")

if is_sorted:
    print("PASS — Complete rainfall records are chronological.")
else:
    print("FAIL — Records are not chronological.")


# ==============================================================
# DUPLICATE TIMESTAMP CHECK
# ==============================================================

print("\n" + "-" * 70)
print("DUPLICATE TIMESTAMP CHECK")
print("-" * 70)

duplicate_timestamps = v2_complete[
    "Data Acquisition Time"
].duplicated().sum()

print(f"Duplicate timestamps: {duplicate_timestamps}")

if duplicate_timestamps == 0:
    print("PASS — No duplicate timestamps.")
else:
    print("WARNING — Duplicate timestamps detected.")


# ==============================================================
# EXISTING V1 TRAIN / TEST BOUNDARY
# ==============================================================

print("\n" + "-" * 70)
print("EXISTING V1 TRAIN / TEST BOUNDARY")
print("-" * 70)

print(f"V1 train end : {V1_TRAIN_END}")
print(f"V1 test start: {V1_TEST_START}")


# ==============================================================
# V2 COMPLETE RAINFALL RECORDS INSIDE V1 TRAIN / TEST
# ==============================================================

v2_train = v2_complete[
    v2_complete["Data Acquisition Time"] <= V1_TRAIN_END
].copy()

v2_test = v2_complete[
    v2_complete["Data Acquisition Time"] >= V1_TEST_START
].copy()

print("\nV2 complete-rainfall records:")
print(f"Train period: {len(v2_train)}")
print(f"Test period : {len(v2_test)}")


# ==============================================================
# TARGET DISTRIBUTION
# ==============================================================

print("\n" + "-" * 70)
print("V2 TRAIN TARGET DISTRIBUTION")
print("-" * 70)

print(v2_train[TARGET].value_counts().sort_index())

print("\nPercentages:")

print(
    v2_train[TARGET]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
)


print("\n" + "-" * 70)
print("V2 TEST TARGET DISTRIBUTION")
print("-" * 70)

print(v2_test[TARGET].value_counts().sort_index())

print("\nPercentages:")

print(
    v2_test[TARGET]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
)


# ==============================================================
# CLASS PRESENCE CHECK
# ==============================================================

print("\n" + "-" * 70)
print("CLASS PRESENCE CHECK")
print("-" * 70)

train_classes = sorted(
    v2_train[TARGET].dropna().unique().tolist()
)

test_classes = sorted(
    v2_test[TARGET].dropna().unique().tolist()
)

print("Training classes:", train_classes)
print("Testing classes :", test_classes)

if len(train_classes) == 2:
    print("PASS — Both target classes exist in training.")
else:
    print("WARNING — Training does not contain both target classes.")

if len(test_classes) == 2:
    print("PASS — Both target classes exist in testing.")
else:
    print(
        "WARNING — Testing does not contain both target classes."
    )


# ==============================================================
# RAINFALL RANGE CHECK
# ==============================================================

print("\n" + "-" * 70)
print("RAINFALL FEATURE RANGE CHECK")
print("-" * 70)

for col in RAIN_FEATURES:

    train_min = v2_train[col].min()
    train_max = v2_train[col].max()

    test_min = v2_test[col].min()
    test_max = v2_test[col].max()

    print(f"\n{col}")

    print(f"  Train: {train_min:.3f} → {train_max:.3f}")
    print(f"  Test : {test_min:.3f} → {test_max:.3f}")


# ==============================================================
# TRAIN / TEST TEMPORAL LEAKAGE CHECK
# ==============================================================

print("\n" + "-" * 70)
print("TEMPORAL LEAKAGE CHECK")
print("-" * 70)

train_max_time = v2_train["Data Acquisition Time"].max()
test_min_time = v2_test["Data Acquisition Time"].min()

print(f"Training latest timestamp: {train_max_time}")
print(f"Testing earliest timestamp: {test_min_time}")

if train_max_time < test_min_time:
    print("PASS — Chronological separation confirmed.")
else:
    print("FAIL — Temporal overlap detected.")


# ==============================================================
# SAMPLE TRAINING RECORDS
# ==============================================================

print("\n" + "-" * 70)
print("V2 TRAINING SAMPLE")
print("-" * 70)

print(
    v2_train[
        ["Data Acquisition Time"] +
        RAIN_FEATURES +
        [TARGET]
    ].head(10).to_string(index=False)
)


# ==============================================================
# SAMPLE TEST RECORDS
# ==============================================================

print("\n" + "-" * 70)
print("V2 TEST SAMPLE")
print("-" * 70)

print(
    v2_test[
        ["Data Acquisition Time"] +
        RAIN_FEATURES +
        [TARGET]
    ].head(10).to_string(index=False)
)


# ==============================================================
# FINAL EXPERIMENT ELIGIBILITY
# ==============================================================

print("\n" + "=" * 70)
print("STEP 2F — EXPERIMENT ELIGIBILITY SUMMARY")
print("=" * 70)

print(f"Complete rainfall rows : {len(v2_complete)}")
print(f"V2 training rows       : {len(v2_train)}")
print(f"V2 testing rows        : {len(v2_test)}")

print(
    f"Training classes       : {train_classes}"
)

print(
    f"Testing classes        : {test_classes}"
)

print(
    f"Temporal separation    : "
    f"{train_max_time < test_min_time}"
)

print("=" * 70)
print("NO MODEL TRAINING")
print("NO V1 MODEL MODIFICATION")
print("NO V1 DATASET MODIFICATION")
print("NO FINAL MODEL MODIFICATION")
print("=" * 70)

print("STEP 2F COMPLETE")
print("=" * 70)
```

```python
import pandas as pd
import numpy as np
import joblib
from pathlib import Path

print("=" * 70)
print("STEP 2G — V1 vs V2 COMPARABLE EVALUATION AUDIT")
print("=" * 70)


# ==============================================================
# CONFIGURATION
# ==============================================================

PROJECT_ROOT = Path(r"D:\document\Brahmaputra-FloodCast")

V1_PATH = PROJECT_ROOT / "data" / "cleaned" / "final_ml_dataset.csv"
V2_PATH = PROJECT_ROOT / "data" / "cleaned" / "v2_rainfall_candidate_dataset.csv"
MODEL_PATH = PROJECT_ROOT / "models" / "floodsense_random_forest.joblib"

TIME_COL = "Data Acquisition Time"
TARGET = "High_Water_6h"

RAIN_FEATURES = [
    "rainfall_1h_sum",
    "rainfall_3h_sum",
    "rainfall_6h_sum",
    "rainfall_12h_sum",
    "rainfall_24h_sum"
]

V1_FEATURES = [
    "Current_Water_Level",
    "Month",
    "Hour",
    "DayOfYear",
    "WaterLevel_Lag_1h",
    "WaterLevel_Lag_3h",
    "WaterLevel_Lag_6h",
    "WaterLevel_Lag_12h",
    "WaterLevel_Lag_24h",
    "WaterLevel_RollingMean_6h",
    "WaterLevel_RollingMean_12h",
    "WaterLevel_RollingMean_24h",
    "WaterLevel_RollingMax_24h",
    "WaterLevel_RollingMin_24h",
    "WaterLevel_Change_1h",
    "WaterLevel_Change_3h",
    "WaterLevel_Change_6h",
    "WaterLevel_Change_12h",
    "WaterLevel_Change_24h"
]

V1_TRAIN_END = pd.Timestamp("2025-04-06 03:00:00")
V1_TEST_START = pd.Timestamp("2025-04-06 04:00:00")


# ==============================================================
# 1. PATH VALIDATION
# ==============================================================

print("\n" + "-" * 70)
print("PATH VALIDATION")
print("-" * 70)

print("V1 dataset :", V1_PATH)
print("V2 dataset :", V2_PATH)
print("Model      :", MODEL_PATH)

assert V1_PATH.exists(), "V1 dataset not found."
assert V2_PATH.exists(), "V2 dataset not found."
assert MODEL_PATH.exists(), "Saved Random Forest model not found."

print("PASS — All required files exist.")


# ==============================================================
# 2. LOAD DATA
# ==============================================================

print("\n" + "-" * 70)
print("LOADING DATA")
print("-" * 70)

v1 = pd.read_csv(V1_PATH)
v2 = pd.read_csv(V2_PATH)

v1[TIME_COL] = pd.to_datetime(v1[TIME_COL], errors="coerce")
v2[TIME_COL] = pd.to_datetime(v2[TIME_COL], errors="coerce")

print(f"V1 rows: {len(v1)}")
print(f"V2 rows: {len(v2)}")


# ==============================================================
# 3. VERIFY V1 FEATURES
# ==============================================================

print("\n" + "-" * 70)
print("V1 FEATURE VALIDATION")
print("-" * 70)

missing_v1 = [
    col for col in V1_FEATURES
    if col not in v1.columns
]

print("Missing V1 features:", missing_v1)

assert len(missing_v1) == 0, "V1 feature columns missing."

print("PASS — All 19 V1 features exist.")


# ==============================================================
# 4. VERIFY RAINFALL FEATURES
# ==============================================================

print("\n" + "-" * 70)
print("RAINFALL FEATURE VALIDATION")
print("-" * 70)

missing_rain = [
    col for col in RAIN_FEATURES
    if col not in v2.columns
]

print("Missing rainfall features:", missing_rain)

assert len(missing_rain) == 0, \
    "Rainfall feature columns missing."

print("PASS — All 5 rainfall features exist.")


# ==============================================================
# 5. SELECT COMPLETE RAINFALL ROWS
# ==============================================================

print("\n" + "-" * 70)
print("SELECTING COMPLETE RAINFALL OBSERVATIONS")
print("-" * 70)

complete_mask = v2[RAIN_FEATURES].notna().all(axis=1)

comparison_times = (
    v2.loc[complete_mask, [TIME_COL]]
    .drop_duplicates()
    .sort_values(TIME_COL)
)

print(
    "Complete rainfall timestamps:",
    len(comparison_times)
)


# ==============================================================
# 6. MATCH V1 WITH COMPLETE RAINFALL TIMESTAMPS
# ==============================================================

print("\n" + "-" * 70)
print("V1 + RAINFALL TIMESTAMP MATCHING")
print("-" * 70)

comparison = v1.merge(
    v2[
        [TIME_COL] + RAIN_FEATURES
    ],
    on=TIME_COL,
    how="inner",
    validate="one_to_one"
)

comparison = comparison.sort_values(
    TIME_COL
).reset_index(drop=True)

# Keep only rows where ALL rainfall features exist
comparison = comparison[
    comparison[RAIN_FEATURES].notna().all(axis=1)
].copy()

print("Comparable rows:", len(comparison))


# ==============================================================
# 7. VERIFY V1 FEATURES ARE COMPLETE
# ==============================================================
print("\n" + "-" * 70)
print("V1 FEATURE COMPLETENESS")
print("-" * 70)

v1_missing_rows = comparison[V1_FEATURES].isna().any(axis=1).sum()

print(
    "Rows with missing V1 features:",
    v1_missing_rows
)

if v1_missing_rows == 0:
    print("PASS — All comparable rows have complete V1 features.")
else:
    print(
        "WARNING — Some comparable rows have missing V1 features."
    )


# ==============================================================
# 8. REMOVE INCOMPLETE V1 ROWS FOR FAIR COMPARISON
# ==============================================================
comparison = comparison[
    comparison[V1_FEATURES].notna().all(axis=1)
].copy()

comparison = comparison.reset_index(drop=True)

print(
    "Final fully-comparable rows:",
    len(comparison)
)


# ==============================================================
# 9. TARGET CHECK
# ==============================================================

print("\n" + "-" * 70)
print("TARGET VALIDATION")
print("-" * 70)

print(
    "Missing target values:",
    comparison[TARGET].isna().sum()
)

assert comparison[TARGET].notna().all(), \
    "Target contains missing values."

print("PASS — Target complete.")


# ==============================================================
# 10. CHRONOLOGICAL SPLIT ON COMPARABLE DATA
# ==============================================================

print("\n" + "-" * 70)
print("COMPARABLE DATA CHRONOLOGICAL SPLIT")
print("-" * 70)

comparison_train = comparison[
    comparison[TIME_COL] <= V1_TRAIN_END
].copy()

comparison_test = comparison[
    comparison[TIME_COL] >= V1_TEST_START
].copy()

print(
    "Comparable training rows:",
    len(comparison_train)
)

print(
    "Comparable testing rows :",
    len(comparison_test)
)


# ==============================================================
# 11. TARGET DISTRIBUTION
# ==============================================================

print("\n" + "-" * 70)
print("COMPARABLE TRAIN TARGET")
print("-" * 70)

print(
    comparison_train[TARGET]
    .value_counts()
    .sort_index()
)

print("\n" + "-" * 70)
print("COMPARABLE TEST TARGET")
print("-" * 70)

print(
    comparison_test[TARGET]
    .value_counts()
    .sort_index()
)


# ==============================================================
# 12. CHECK TEST CLASS VARIETY
# ==============================================================

print("\n" + "-" * 70)
print("TEST CLASS VARIETY")
print("-" * 70)

test_classes = sorted(
    comparison_test[TARGET]
    .unique()
    .tolist()
)

print("Test classes:", test_classes)

if len(test_classes) == 2:
    print(
        "PASS — Comparable test set contains both classes."
    )
else:
    print(
        "WARNING — Comparable test set does NOT contain both classes."
    )


# ==============================================================
# 13. LOAD EXISTING V1 MODEL
# ==============================================================

print("\n" + "-" * 70)
print("LOADING EXISTING V1 RANDOM FOREST")
print("-" * 70)

model = joblib.load(MODEL_PATH)

print("Model type:", type(model).__name__)

print("PASS — Existing V1 model loaded.")
print("NO RETRAINING PERFORMED.")


# ==============================================================
# 14. MODEL FEATURE CHECK
# ==============================================================

print("\n" + "-" * 70)
print("MODEL FEATURE CHECK")
print("-" * 70)

if hasattr(model, "n_features_in_"):
    print(
        "Model expects features:",
        model.n_features_in_
    )

print(
    "Current V1 feature count:",
    len(V1_FEATURES)
)

if hasattr(model, "n_features_in_"):
    if model.n_features_in_ == len(V1_FEATURES):
        print(
            "PASS — Model feature count matches V1."
        )
    else:
        print(
            "WARNING — Model feature count mismatch."
        )


# ==============================================================
# 15. V1 PREDICTION ON COMPARABLE DATA
# ==============================================================

print("\n" + "-" * 70)
print("EXISTING V1 MODEL — COMPARABLE DATA PREDICTION")
print("-" * 70)

X_comparable_test = comparison_test[V1_FEATURES].copy()
y_comparable_test = comparison_test[TARGET].copy()

v1_predictions = model.predict(X_comparable_test)

comparison_test["V1_Prediction"] = v1_predictions

print(
    "V1 predictions generated:",
    len(v1_predictions)
)

print("PASS — Existing V1 model evaluated.")
print("NO V1 MODEL RETRAINING.")


# ==============================================================
# 16. V1 PREDICTION DISTRIBUTION
# ==============================================================

print("\n" + "-" * 70)
print("V1 PREDICTION DISTRIBUTION")
print("-" * 70)

print(
    pd.Series(v1_predictions)
    .value_counts()
    .sort_index()
)


# ==============================================================
# 17. V1 ACTUAL VS PREDICTED TABLE
# ==============================================================

print("\n" + "-" * 70)
print("V1 ACTUAL VS PREDICTED")
print("-" * 70)

v1_comparison_table = pd.crosstab(
    comparison_test[TARGET],
    comparison_test["V1_Prediction"],
    rownames=["Actual"],
    colnames=["V1 Prediction"]
)

print(v1_comparison_table)


# ==============================================================
# 18. COMPARISON DATASET RANGE
# ==============================================================

print("\n" + "-" * 70)
print("FINAL COMPARABLE DATA RANGE")
print("-" * 70)

print(
    "Start:",
    comparison[TIME_COL].min()
)

print(
    "End  :",
    comparison[TIME_COL].max()
)

print(
    "Train start:",
    comparison_train[TIME_COL].min()
)

print(
    "Train end:",
    comparison_train[TIME_COL].max()
)

print(
    "Test start:",
    comparison_test[TIME_COL].min()
)

print(
    "Test end:",
    comparison_test[TIME_COL].max()
)


# ==============================================================
# 19. SAVE COMPARISON DATASET
# ==============================================================

OUTPUT_PATH = (
    PROJECT_ROOT /
    "data" /
    "cleaned" /
    "v1_v2_comparable_dataset.csv"
)

comparison.to_csv(
    OUTPUT_PATH,
    index=False
)

print("\n" + "-" * 70)
print("COMPARABLE DATASET SAVED")
print("-" * 70)

print("Path:", OUTPUT_PATH)
print("Rows:", len(comparison))
print("Columns:", len(comparison.columns))


# ==============================================================
# 20. FINAL STATUS
# ==============================================================

print("\n" + "=" * 70)
print("STEP 2G — FINAL STATUS")
print("=" * 70)

print(
    "Complete rainfall rows        :",
    len(comparison)
)

print(
    "Comparable training rows      :",
    len(comparison_train)
)

print(
    "Comparable testing rows       :",
    len(comparison_test)
)

print(
    "Test classes                  :",
    test_classes
)

print(
    "Existing V1 model loaded      : PASS"
)

print(
    "V1 model retrained            : NO"
)

print(
    "V1 model modified             : NO"
)

print(
    "V1 dataset modified           : NO"
)

print(
    "Comparison dataset created    : PASS"
)

print("=" * 70)
print("STEP 2G COMPLETE")
print("=" * 70)
```

```python
import os
import joblib
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# ==============================================================
# STEP 2H — V2 MODEL TRAINING + EVALUATION
# ==============================================================

print("=" * 70)
print("STEP 2H — V2 MODEL TRAINING + EVALUATION")
print("=" * 70)


# ==============================================================
# 1. PATHS
# ==============================================================

V1_DATASET_PATH = r"D:\document\Brahmaputra-FloodCast\data\cleaned\final_ml_dataset.csv"

COMPARABLE_PATH = r"D:\document\Brahmaputra-FloodCast\data\cleaned\v1_v2_comparable_dataset.csv"

V1_MODEL_PATH = r"D:\document\Brahmaputra-FloodCast\models\floodsense_random_forest.joblib"

V2_MODEL_PATH = r"D:\document\Brahmaputra-FloodCast\models\floodsense_v2_rainfall_random_forest.joblib"


# ==============================================================
# 2. LOCKED V1 FEATURES
# ==============================================================

V1_FEATURES = [
    "Current_Water_Level",
    "Month",
    "Hour",
    "DayOfYear",

    "WaterLevel_Lag_1h",
    "WaterLevel_Lag_3h",
    "WaterLevel_Lag_6h",
    "WaterLevel_Lag_12h",
    "WaterLevel_Lag_24h",

    "WaterLevel_RollingMean_6h",
    "WaterLevel_RollingMean_12h",
    "WaterLevel_RollingMean_24h",

    "WaterLevel_RollingMax_24h",
    "WaterLevel_RollingMin_24h",

    "WaterLevel_Change_1h",
    "WaterLevel_Change_3h",
    "WaterLevel_Change_6h",
    "WaterLevel_Change_12h",
    "WaterLevel_Change_24h"
]


# ==============================================================
# 3. RAINFALL FEATURES
# ==============================================================

RAINFALL_FEATURES = [
    "rainfall_1h_sum",
    "rainfall_3h_sum",
    "rainfall_6h_sum",
    "rainfall_12h_sum",
    "rainfall_24h_sum"
]


# ==============================================================
# 4. TARGET
# ==============================================================

TARGET = "High_Water_6h"


# ==============================================================
# 5. V2 FEATURES
# ==============================================================

V2_FEATURES = V1_FEATURES + RAINFALL_FEATURES


print("\n" + "-" * 70)
print("FEATURE CONFIGURATION")
print("-" * 70)

print(f"V1 feature count      : {len(V1_FEATURES)}")
print(f"Rainfall feature count: {len(RAINFALL_FEATURES)}")
print(f"V2 feature count      : {len(V2_FEATURES)}")

assert len(V1_FEATURES) == 19
assert len(RAINFALL_FEATURES) == 5
assert len(V2_FEATURES) == 24

print("PASS — V2 contains 19 V1 features + 5 rainfall features.")


# ==============================================================
# 6. LOAD COMPARABLE DATASET
# ==============================================================

print("\n" + "-" * 70)
print("LOADING COMPARABLE DATASET")
print("-" * 70)

comparable_df = pd.read_csv(COMPARABLE_PATH)

comparable_df["Data Acquisition Time"] = pd.to_datetime(
    comparable_df["Data Acquisition Time"]
)

print(f"Rows    : {len(comparable_df)}")
print(f"Columns : {comparable_df.shape[1]}")


# ==============================================================
# 7. SORT CHRONOLOGICALLY
# ==============================================================

comparable_df = comparable_df.sort_values(
    "Data Acquisition Time"
).reset_index(drop=True)

print("\nChronological order verified.")


# ==============================================================
# 8. CHECK RAINFALL COMPLETENESS
# ==============================================================

missing_rainfall = comparable_df[RAINFALL_FEATURES].isna().sum()

print("\n" + "-" * 70)
print("RAINFALL COMPLETENESS CHECK")
print("-" * 70)

print(missing_rainfall)

assert missing_rainfall.sum() == 0

print("PASS — No missing rainfall values.")


# ==============================================================
# 9. CHECK V1 FEATURE COMPLETENESS
# ==============================================================

missing_v1 = comparable_df[V1_FEATURES].isna().sum()

print("\n" + "-" * 70)
print("V1 FEATURE COMPLETENESS CHECK")
print("-" * 70)

print(missing_v1)

assert missing_v1.sum() == 0

print("PASS — All V1 features complete.")


# ==============================================================
# 10. TARGET CHECK
# ==============================================================

assert comparable_df[TARGET].isna().sum() == 0

print("\nPASS — Target has no missing values.")


# ==============================================================
# 11. EXISTING CHRONOLOGICAL BOUNDARY
# ==============================================================

TRAIN_END = pd.Timestamp("2025-04-06 03:00:00")
TEST_START = pd.Timestamp("2025-04-06 04:00:00")


train_df = comparable_df[
    comparable_df["Data Acquisition Time"] <= TRAIN_END
].copy()

test_df = comparable_df[
    comparable_df["Data Acquisition Time"] >= TEST_START
].copy()


# ==============================================================
# 12. CHRONOLOGICAL SPLIT VALIDATION
# ==============================================================

print("\n" + "-" * 70)
print("V2 CHRONOLOGICAL SPLIT")
print("-" * 70)

print(f"Training rows: {len(train_df)}")
print(f"Testing rows : {len(test_df)}")

print(f"Train start  : {train_df['Data Acquisition Time'].min()}")
print(f"Train end    : {train_df['Data Acquisition Time'].max()}")

print(f"Test start   : {test_df['Data Acquisition Time'].min()}")
print(f"Test end     : {test_df['Data Acquisition Time'].max()}")


assert train_df["Data Acquisition Time"].max() < \
       test_df["Data Acquisition Time"].min()

print("\nPASS — Chronological separation confirmed.")


# ==============================================================
# 13. TARGET DISTRIBUTION
# ==============================================================

print("\n" + "-" * 70)
print("V2 TRAIN TARGET DISTRIBUTION")
print("-" * 70)

print(train_df[TARGET].value_counts().sort_index())


print("\n" + "-" * 70)
print("V2 TEST TARGET DISTRIBUTION")
print("-" * 70)

print(test_df[TARGET].value_counts().sort_index())


# ==============================================================
# 14. PREPARE X / y
# ==============================================================

X_train_v2 = train_df[V2_FEATURES].copy()
y_train_v2 = train_df[TARGET].copy()

X_test_v2 = test_df[V2_FEATURES].copy()
y_test_v2 = test_df[TARGET].copy()


print("\n" + "-" * 70)
print("V2 MODEL INPUT SHAPES")
print("-" * 70)

print(f"X_train_v2 shape : {X_train_v2.shape}")
print(f"y_train_v2 shape : {y_train_v2.shape}")

print(f"X_test_v2 shape  : {X_test_v2.shape}")
print(f"y_test_v2 shape  : {y_test_v2.shape}")


# ==============================================================
# 15. TRAIN V2 RANDOM FOREST
# ==============================================================

print("\n" + "-" * 70)
print("TRAINING V2 RANDOM FOREST")
print("-" * 70)

print("Model: RandomForestClassifier")
print("n_estimators: 200")
print("Random split: NO")
print("Chronological split: YES")


v2_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

v2_model.fit(
    X_train_v2,
    y_train_v2
)

print("\nPASS — V2 model training complete.")


# ==============================================================
# 16. V2 PREDICTION
# ==============================================================

y_pred_v2 = v2_model.predict(X_test_v2)


# ==============================================================
# 17. METRICS
# ==============================================================

accuracy_v2 = accuracy_score(
    y_test_v2,
    y_pred_v2
)

precision_v2 = precision_score(
    y_test_v2,
    y_pred_v2,
    zero_division=0
)

recall_v2 = recall_score(
    y_test_v2,
    y_pred_v2,
    zero_division=0
)

f1_v2 = f1_score(
    y_test_v2,
    y_pred_v2,
    zero_division=0
)


# ==============================================================
# 18. CONFUSION MATRIX
# ==============================================================

cm_v2 = confusion_matrix(
    y_test_v2,
    y_pred_v2,
    labels=[0, 1]
)


# ==============================================================
# 19. RESULTS
# ==============================================================

print("\n" + "=" * 70)
print("V2 MODEL TEST RESULTS")
print("=" * 70)

print(f"Accuracy  : {accuracy_v2:.4f}")
print(f"Precision : {precision_v2:.4f}")
print(f"Recall    : {recall_v2:.4f}")
print(f"F1-Score  : {f1_v2:.4f}")


print("\n" + "-" * 70)
print("CONFUSION MATRIX")
print("-" * 70)

print("Labels: [0, 1]")
print(cm_v2)


print("\n" + "-" * 70)
print("CLASSIFICATION REPORT")
print("-" * 70)

print(
    classification_report(
        y_test_v2,
        y_pred_v2,
        zero_division=0
    )
)


# ==============================================================
# 20. TEST CLASS WARNING
# ==============================================================

test_classes = sorted(y_test_v2.unique())

print("\n" + "-" * 70)
print("TEST CLASS VARIETY")
print("-" * 70)

print(f"Test classes: {test_classes}")

if len(test_classes) < 2:
    print(
        "WARNING — Test set contains only one target class."
    )
    print(
        "Metrics must NOT be interpreted as a fully representative"
        " binary classification evaluation."
    )


# ==============================================================
# 21. SAVE V2 MODEL SEPARATELY
# ==============================================================

joblib.dump(
    v2_model,
    V2_MODEL_PATH
)

print("\n" + "-" * 70)
print("V2 MODEL SAVED")
print("-" * 70)

print(f"Path: {V2_MODEL_PATH}")
print("PASS — V2 model saved separately.")


# ==============================================================
# 22. FINAL STATUS
# ==============================================================

print("\n" + "=" * 70)
print("STEP 2H — FINAL STATUS")
print("=" * 70)

print(f"V2 training rows       : {len(train_df)}")
print(f"V2 testing rows        : {len(test_df)}")
print(f"V2 features            : {len(V2_FEATURES)}")
print(f"V2 model               : Random Forest (200 trees)")
print(f"V2 Accuracy            : {accuracy_v2:.4f}")
print(f"V2 Precision           : {precision_v2:.4f}")
print(f"V2 Recall              : {recall_v2:.4f}")
print(f"V2 F1                  : {f1_v2:.4f}")

print("\nV1 model modified      : NO")
print("V1 dataset modified    : NO")
print("V2 model saved separately: YES")

print("=" * 70)
print("STEP 2H COMPLETE")
print("=" * 70)
```

```python
# ==============================================================
# STEP 2I — V1 vs V2 FINAL COMPARISON AUDIT
# ==============================================================

import os
import joblib
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

print("=" * 70)
print("STEP 2I — V1 vs V2 FINAL COMPARISON AUDIT")
print("=" * 70)


# ==============================================================
# 1. PATHS
# ==============================================================

COMPARABLE_PATH = (
    r"D:\document\Brahmaputra-FloodCast\data\cleaned"
    r"\v1_v2_comparable_dataset.csv"
)

V1_MODEL_PATH = (
    r"D:\document\Brahmaputra-FloodCast\models"
    r"\floodsense_random_forest.joblib"
)

V2_MODEL_PATH = (
    r"D:\document\Brahmaputra-FloodCast\models"
    r"\floodsense_v2_rainfall_random_forest.joblib"
)


# ==============================================================
# 2. FEATURES
# ==============================================================

V1_FEATURES = [
    "Current_Water_Level",
    "Month",
    "Hour",
    "DayOfYear",
    "WaterLevel_Lag_1h",
    "WaterLevel_Lag_3h",
    "WaterLevel_Lag_6h",
    "WaterLevel_Lag_12h",
    "WaterLevel_Lag_24h",
    "WaterLevel_RollingMean_6h",
    "WaterLevel_RollingMean_12h",
    "WaterLevel_RollingMean_24h",
    "WaterLevel_RollingMax_24h",
    "WaterLevel_RollingMin_24h",
    "WaterLevel_Change_1h",
    "WaterLevel_Change_3h",
    "WaterLevel_Change_6h",
    "WaterLevel_Change_12h",
    "WaterLevel_Change_24h"
]

RAINFALL_FEATURES = [
    "rainfall_1h_sum",
    "rainfall_3h_sum",
    "rainfall_6h_sum",
    "rainfall_12h_sum",
    "rainfall_24h_sum"
]

V2_FEATURES = V1_FEATURES + RAINFALL_FEATURES

TARGET = "High_Water_6h"


# ==============================================================
# 3. LOAD COMPARABLE DATASET
# ==============================================================

df = pd.read_csv(COMPARABLE_PATH)

df["Data Acquisition Time"] = pd.to_datetime(
    df["Data Acquisition Time"]
)

df = df.sort_values(
    "Data Acquisition Time"
).reset_index(drop=True)


print("\n" + "-" * 70)
print("COMPARABLE DATASET")
print("-" * 70)

print(f"Rows: {len(df)}")


# ==============================================================
# 4. EXISTING CHRONOLOGICAL TEST PERIOD
# ==============================================================

TEST_START = pd.Timestamp("2025-04-06 04:00:00")

test_df = df[
    df["Data Acquisition Time"] >= TEST_START
].copy()


print("\n" + "-" * 70)
print("COMPARABLE TEST SET")
print("-" * 70)

print(f"Test rows : {len(test_df)}")
print(
    f"Test start: "
    f"{test_df['Data Acquisition Time'].min()}"
)
print(
    f"Test end  : "
    f"{test_df['Data Acquisition Time'].max()}"
)


# ==============================================================
# 5. TARGET DISTRIBUTION
# ==============================================================

target_counts = test_df[TARGET].value_counts().sort_index()

print("\n" + "-" * 70)
print("TEST TARGET DISTRIBUTION")
print("-" * 70)

print(target_counts)


test_classes = sorted(
    test_df[TARGET].unique()
)

print(f"\nTest classes: {test_classes}")


# ==============================================================
# 6. LOAD V1
# ==============================================================

print("\n" + "-" * 70)
print("LOADING V1 MODEL")
print("-" * 70)

v1_model = joblib.load(V1_MODEL_PATH)

print(
    f"V1 model type: "
    f"{type(v1_model).__name__}"
)

print("PASS — V1 loaded.")
print("V1 retrained: NO")


# ==============================================================
# 7. LOAD V2
# ==============================================================

print("\n" + "-" * 70)
print("LOADING V2 MODEL")
print("-" * 70)

v2_model = joblib.load(V2_MODEL_PATH)

print(
    f"V2 model type: "
    f"{type(v2_model).__name__}"
)

print("PASS — V2 loaded.")


# ==============================================================
# 8. PREPARE TEST DATA
# ==============================================================

X_test_v1 = test_df[V1_FEATURES].copy()
X_test_v2 = test_df[V2_FEATURES].copy()

y_test = test_df[TARGET].copy()


# ==============================================================
# 9. MODEL PREDICTIONS
# ==============================================================

print("\n" + "-" * 70)
print("GENERATING COMPARABLE PREDICTIONS")
print("-" * 70)

pred_v1 = v1_model.predict(X_test_v1)
pred_v2 = v2_model.predict(X_test_v2)

print(f"V1 predictions: {len(pred_v1)}")
print(f"V2 predictions: {len(pred_v2)}")


# ==============================================================
# 10. METRIC FUNCTION
# ==============================================================

def calculate_metrics(y_true, y_pred):

    return {
        "Accuracy": accuracy_score(
            y_true, y_pred
        ),

        "Precision": precision_score(
            y_true,
            y_pred,
            zero_division=0
        ),

        "Recall": recall_score(
            y_true,
            y_pred,
            zero_division=0
        ),

        "F1": f1_score(
            y_true,
            y_pred,
            zero_division=0
        )
    }


metrics_v1 = calculate_metrics(
    y_test,
    pred_v1
)

metrics_v2 = calculate_metrics(
    y_test,
    pred_v2
)


# ==============================================================
# 11. COMPARISON TABLE
# ==============================================================

comparison = pd.DataFrame(
    {
        "V1": metrics_v1,
        "V2": metrics_v2
    }
)

comparison["Difference_V2_minus_V1"] = (
    comparison["V2"] -
    comparison["V1"]
)


print("\n" + "=" * 70)
print("V1 vs V2 COMPARISON")
print("=" * 70)

print(comparison)


# ==============================================================
# 12. CONFUSION MATRICES
# ==============================================================

cm_v1 = confusion_matrix(
    y_test,
    pred_v1,
    labels=[0, 1]
)

cm_v2 = confusion_matrix(
    y_test,
    pred_v2,
    labels=[0, 1]
)


print("\n" + "-" * 70)
print("V1 CONFUSION MATRIX")
print("-" * 70)

print(cm_v1)


print("\n" + "-" * 70)
print("V2 CONFUSION MATRIX")
print("-" * 70)

print(cm_v2)


# ==============================================================
# 13. PREDICTION DISTRIBUTIONS
# ==============================================================

print("\n" + "-" * 70)
print("V1 PREDICTION DISTRIBUTION")
print("-" * 70)

print(
    pd.Series(pred_v1).value_counts().sort_index()
)


print("\n" + "-" * 70)
print("V2 PREDICTION DISTRIBUTION")
print("-" * 70)

print(
    pd.Series(pred_v2).value_counts().sort_index()
)


# ==============================================================
# 14. IDENTICAL PREDICTION CHECK
# ==============================================================

identical_predictions = (
    pred_v1 == pred_v2
).all()

print("\n" + "-" * 70)
print("PREDICTION AGREEMENT")
print("-" * 70)

print(
    f"V1 and V2 predictions identical: "
    f"{identical_predictions}"
)

if identical_predictions:
    print(
        "NOTE — V1 and V2 produce identical "
        "predictions on the comparable test set."
    )


# ==============================================================
# 15. TEST SET LIMITATION
# ==============================================================

print("\n" + "-" * 70)
print("EVALUATION LIMITATION")
print("-" * 70)

if len(test_classes) < 2:

    print(
        "WARNING — Comparable test set contains "
        "only one target class."
    )

    print(
        "Therefore this evaluation cannot establish "
        "full binary-classification superiority."
    )

    print(
        "100% metrics do NOT prove V2 is superior."
    )


# ==============================================================
# 16. PRELIMINARY DECISION
# ==============================================================

print("\n" + "=" * 70)
print("PRELIMINARY V2 DECISION")
print("=" * 70)

if identical_predictions:

    print(
        "RESULT: NO DEMONSTRATED IMPROVEMENT OF V2 "
        "OVER V1 ON THIS COMPARABLE TEST SET."
    )

else:

    f1_difference = (
        metrics_v2["F1"] -
        metrics_v1["F1"]
    )

    if f1_difference > 0:
        print(
            "RESULT: V2 shows higher F1 on the "
            "comparable test subset."
        )

    elif f1_difference < 0:
        print(
            "RESULT: V2 shows lower F1 on the "
            "comparable test subset."
        )

    else:
        print(
            "RESULT: V1 and V2 have equal F1."
        )


print("\nIMPORTANT:")
print(
    "This is NOT the final production-model decision."
)

print(
    "The test-set class limitation must be considered "
    "before selecting V1 or V2 as the final model."
)


# ==============================================================
# 17. SAVE COMPARISON
# ==============================================================

comparison_path = (
    r"D:\document\Brahmaputra-FloodCast\data\cleaned"
    r"\v1_v2_model_comparison.csv"
)

comparison.to_csv(
    comparison_path
)

print("\n" + "-" * 70)
print("COMPARISON SAVED")
print("-" * 70)

print(f"Path: {comparison_path}")


# ==============================================================
# FINAL STATUS
# ==============================================================

print("\n" + "=" * 70)
print("STEP 2I COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 2J — V2 RAINFALL CONTRIBUTION / FEATURE IMPORTANCE AUDIT
# ======================================================================

import pandas as pd
import joblib
import os

print("=" * 70)
print("STEP 2J — V2 RAINFALL CONTRIBUTION AUDIT")
print("=" * 70)


# ----------------------------------------------------------------------
# CONFIGURATION
# ----------------------------------------------------------------------

V2_MODEL_PATH = r"D:\document\Brahmaputra-FloodCast\models\floodsense_v2_rainfall_random_forest.joblib"

V2_DATASET_PATH = r"D:\document\Brahmaputra-FloodCast\data\cleaned\v1_v2_comparable_dataset.csv"


V1_FEATURES = [
    "Current_Water_Level",
    "Month",
    "Hour",
    "DayOfYear",
    "WaterLevel_Lag_1h",
    "WaterLevel_Lag_3h",
    "WaterLevel_Lag_6h",
    "WaterLevel_Lag_12h",
    "WaterLevel_Lag_24h",
    "WaterLevel_RollingMean_6h",
    "WaterLevel_RollingMean_12h",
    "WaterLevel_RollingMean_24h",
    "WaterLevel_RollingMax_24h",
    "WaterLevel_RollingMin_24h",
    "WaterLevel_Change_1h",
    "WaterLevel_Change_3h",
    "WaterLevel_Change_6h",
    "WaterLevel_Change_12h",
    "WaterLevel_Change_24h"
]

RAINFALL_FEATURES = [
    "rainfall_1h_sum",
    "rainfall_3h_sum",
    "rainfall_6h_sum",
    "rainfall_12h_sum",
    "rainfall_24h_sum"
]

TARGET = "High_Water_6h"


# ----------------------------------------------------------------------
# FILE VALIDATION
# ----------------------------------------------------------------------

print("\n" + "-" * 70)
print("FILE VALIDATION")
print("-" * 70)

print("V2 model exists   :", os.path.exists(V2_MODEL_PATH))
print("V2 dataset exists :", os.path.exists(V2_DATASET_PATH))


# ----------------------------------------------------------------------
# LOAD MODEL
# ----------------------------------------------------------------------

print("\n" + "-" * 70)
print("LOADING V2 MODEL")
print("-" * 70)

v2_model = joblib.load(V2_MODEL_PATH)

print("Model type :", type(v2_model).__name__)

if hasattr(v2_model, "n_estimators"):
    print("Trees      :", v2_model.n_estimators)


# ----------------------------------------------------------------------
# LOAD DATA
# ----------------------------------------------------------------------

print("\n" + "-" * 70)
print("LOADING COMPARABLE DATASET")
print("-" * 70)

df = pd.read_csv(V2_DATASET_PATH)

print("Rows    :", len(df))
print("Columns :", len(df.columns))


# ----------------------------------------------------------------------
# FEATURE VALIDATION
# ----------------------------------------------------------------------

ALL_FEATURES = V1_FEATURES + RAINFALL_FEATURES

missing_features = [
    col for col in ALL_FEATURES
    if col not in df.columns
]

print("\n" + "-" * 70)
print("FEATURE VALIDATION")
print("-" * 70)

print("Expected features :", len(ALL_FEATURES))
print("Missing features  :", missing_features)

if missing_features:
    raise ValueError("Missing required features.")

print("PASS — All 24 V2 features exist.")


# ----------------------------------------------------------------------
# MODEL FEATURE CHECK
# ----------------------------------------------------------------------

print("\n" + "-" * 70)
print("MODEL FEATURE CHECK")
print("-" * 70)

model_features = getattr(v2_model, "n_features_in_", None)

print("Model expects :", model_features)
print("V2 features  :", len(ALL_FEATURES))

if model_features != len(ALL_FEATURES):
    raise ValueError("Model feature count does not match V2 feature count.")

print("PASS — Model expects 24 features.")


# ----------------------------------------------------------------------
# FEATURE IMPORTANCE
# ----------------------------------------------------------------------

print("\n" + "-" * 70)
print("V2 FEATURE IMPORTANCE")
print("-" * 70)

importance_df = pd.DataFrame({
    "Feature": ALL_FEATURES,
    "Importance": v2_model.feature_importances_
})

importance_df = importance_df.sort_values(
    by="Importance",
    ascending=False
).reset_index(drop=True)

importance_df["Rank"] = range(1, len(importance_df) + 1)

importance_df = importance_df[
    ["Rank", "Feature", "Importance"]
]

print(importance_df.to_string(index=False))


# ----------------------------------------------------------------------
# RAINFALL CONTRIBUTION
# ----------------------------------------------------------------------

rainfall_importance = importance_df[
    importance_df["Feature"].isin(RAINFALL_FEATURES)
].copy()

v1_importance = importance_df[
    importance_df["Feature"].isin(V1_FEATURES)
].copy()


rainfall_total = rainfall_importance["Importance"].sum()
v1_total = v1_importance["Importance"].sum()


print("\n" + "-" * 70)
print("FEATURE GROUP CONTRIBUTION")
print("-" * 70)

print(f"V1 water-level/time features importance : {v1_total:.6f}")
print(f"Rainfall features importance             : {rainfall_total:.6f}")
print(f"Total importance                         : {v1_total + rainfall_total:.6f}")


# ----------------------------------------------------------------------
# RAINFALL FEATURE RANKING
# ----------------------------------------------------------------------

print("\n" + "-" * 70)
print("RAINFALL FEATURE RANKING")
print("-" * 70)

print(
    rainfall_importance[
        ["Rank", "Feature", "Importance"]
    ].to_string(index=False)
)


# ----------------------------------------------------------------------
# FINAL AUDIT
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 2J — FINAL STATUS")
print("=" * 70)

print("V2 model loaded              : PASS")
print("24 features verified         : PASS")
print("Feature importance calculated : PASS")
print(f"Rainfall total importance     : {rainfall_total:.6f}")
print(f"V1 feature total importance  : {v1_total:.6f}")

print("\nNO V1 MODEL MODIFICATION")
print("NO V1 DATASET MODIFICATION")
print("NO V2 MODEL RETRAINING")
print("NO PRODUCTION MODEL CHANGE")

print("=" * 70)
print("STEP 2J COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 2K — V2 RAINFALL FEATURE ABLATION / SENSITIVITY AUDIT
# ======================================================================

import pandas as pd
import numpy as np
import joblib

print("=" * 70)
print("STEP 2K — V2 RAINFALL FEATURE ABLATION AUDIT")
print("=" * 70)


# ----------------------------------------------------------------------
# CONFIGURATION
# ----------------------------------------------------------------------

MODEL_PATH = r"D:\document\Brahmaputra-FloodCast\models\floodsense_v2_rainfall_random_forest.joblib"

DATA_PATH = r"D:\document\Brahmaputra-FloodCast\data\cleaned\v1_v2_comparable_dataset.csv"


V1_FEATURES = [
    "Current_Water_Level",
    "Month",
    "Hour",
    "DayOfYear",
    "WaterLevel_Lag_1h",
    "WaterLevel_Lag_3h",
    "WaterLevel_Lag_6h",
    "WaterLevel_Lag_12h",
    "WaterLevel_Lag_24h",
    "WaterLevel_RollingMean_6h",
    "WaterLevel_RollingMean_12h",
    "WaterLevel_RollingMean_24h",
    "WaterLevel_RollingMax_24h",
    "WaterLevel_RollingMin_24h",
    "WaterLevel_Change_1h",
    "WaterLevel_Change_3h",
    "WaterLevel_Change_6h",
    "WaterLevel_Change_12h",
    "WaterLevel_Change_24h"
]

RAINFALL_FEATURES = [
    "rainfall_1h_sum",
    "rainfall_3h_sum",
    "rainfall_6h_sum",
    "rainfall_12h_sum",
    "rainfall_24h_sum"
]

ALL_FEATURES = V1_FEATURES + RAINFALL_FEATURES

TARGET = "High_Water_6h"


# ----------------------------------------------------------------------
# LOAD MODEL
# ----------------------------------------------------------------------

print("\n" + "-" * 70)
print("LOADING V2 MODEL")
print("-" * 70)

model = joblib.load(MODEL_PATH)

print("Model type :", type(model).__name__)
print("Trees      :", model.n_estimators)


# ----------------------------------------------------------------------
# LOAD DATA
# ----------------------------------------------------------------------

print("\n" + "-" * 70)
print("LOADING COMPARABLE DATASET")
print("-" * 70)

df = pd.read_csv(DATA_PATH)

print("Rows :", len(df))


# ----------------------------------------------------------------------
# SORT CHRONOLOGICALLY
# ----------------------------------------------------------------------

df["Data Acquisition Time"] = pd.to_datetime(
    df["Data Acquisition Time"]
)

df = df.sort_values(
    "Data Acquisition Time"
).reset_index(drop=True)


# ----------------------------------------------------------------------
# RECREATE SAME V2 TEST SET
# ----------------------------------------------------------------------

split_timestamp = pd.Timestamp("2025-04-06 04:00:00")

test_df = df[
    df["Data Acquisition Time"] >= split_timestamp
].copy()

print("\n" + "-" * 70)
print("TEST DATA")
print("-" * 70)

print("Test rows :", len(test_df))

print(
    "Test start:",
    test_df["Data Acquisition Time"].min()
)

print(
    "Test end  :",
    test_df["Data Acquisition Time"].max()
)


# ----------------------------------------------------------------------
# ORIGINAL V2 INPUT
# ----------------------------------------------------------------------

X_original = test_df[ALL_FEATURES].copy()

y_test = test_df[TARGET].copy()


print("\n" + "-" * 70)
print("ORIGINAL V2 INPUT")
print("-" * 70)

print("Shape:", X_original.shape)


# ----------------------------------------------------------------------
# BASELINE PREDICTION
# ----------------------------------------------------------------------

pred_original = model.predict(X_original)


# ----------------------------------------------------------------------
# RAINFALL MEDIAN BASELINE
# ----------------------------------------------------------------------
#
# We replace each rainfall feature with its TRAINING-PERIOD median.
#
# This removes the specific rainfall signal from each test observation
# while keeping the feature count and model interface unchanged.
#
# NO MODEL RETRAINING.
#
# ----------------------------------------------------------------------

train_df = df[
    df["Data Acquisition Time"] < split_timestamp
].copy()

X_ablation = X_original.copy()

print("\n" + "-" * 70)
print("RAINFALL ABLATION")
print("-" * 70)

for feature in RAINFALL_FEATURES:

    baseline_value = train_df[feature].median()

    print(
        f"{feature:<25} "
        f"baseline median: {baseline_value:.4f}"
    )

    X_ablation[feature] = baseline_value


# ----------------------------------------------------------------------
# ABLATED PREDICTION
# ----------------------------------------------------------------------

pred_ablation = model.predict(X_ablation)


# ----------------------------------------------------------------------
# PREDICTION CHANGE
# ----------------------------------------------------------------------

prediction_changed = pred_original != pred_ablation

changed_count = prediction_changed.sum()

total_count = len(pred_original)

change_percentage = (
    changed_count / total_count
) * 100


print("\n" + "-" * 70)
print("PREDICTION SENSITIVITY")
print("-" * 70)

print("Total test predictions :", total_count)
print("Changed predictions    :", changed_count)
print(
    f"Prediction change      : {change_percentage:.2f}%"
)


# ----------------------------------------------------------------------
# PROBABILITY SENSITIVITY
# ----------------------------------------------------------------------

if hasattr(model, "predict_proba"):

    probability_original = model.predict_proba(
        X_original
    )[:, 1]

    probability_ablation = model.predict_proba(
        X_ablation
    )[:, 1]

    probability_difference = (
        probability_original -
        probability_ablation
    )

    print("\n" + "-" * 70)
    print("PROBABILITY SENSITIVITY")
    print("-" * 70)

    print(
        "Mean absolute probability change :",
        f"{np.mean(np.abs(probability_difference)):.6f}"
    )

    print(
        "Maximum probability change       :",
        f"{np.max(np.abs(probability_difference)):.6f}"
    )

    print(
        "Minimum probability change       :",
        f"{np.min(np.abs(probability_difference)):.6f}"
    )


# ----------------------------------------------------------------------
# PREDICTION DISTRIBUTIONS
# ----------------------------------------------------------------------

print("\n" + "-" * 70)
print("ORIGINAL PREDICTIONS")
print("-" * 70)

print(
    pd.Series(pred_original).value_counts()
    .sort_index()
)


print("\n" + "-" * 70)
print("ABLATION PREDICTIONS")
print("-" * 70)

print(
    pd.Series(pred_ablation).value_counts()
    .sort_index()
)


# ----------------------------------------------------------------------
# CHANGED PREDICTION SAMPLES
# ----------------------------------------------------------------------

changed_df = test_df.loc[
    prediction_changed,
    [
        "Data Acquisition Time",
        "High_Water_6h"
    ] + RAINFALL_FEATURES
].copy()

changed_df["Original_Prediction"] = pred_original[
    prediction_changed
]

changed_df["Ablated_Prediction"] = pred_ablation[
    prediction_changed
]


print("\n" + "-" * 70)
print("CHANGED PREDICTION SAMPLES")
print("-" * 70)

if len(changed_df) == 0:

    print("No predictions changed after rainfall ablation.")

else:

    print(
        changed_df.head(20).to_string(index=False)
    )


# ----------------------------------------------------------------------
# FINAL STATUS
# ----------------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 2K — FINAL STATUS")
print("=" * 70)

print("Existing V2 model used      : PASS")
print("Same V2 test set            : PASS")
print("Rainfall ablation performed : PASS")
print("Model retrained             : NO")
print("V1 model modified           : NO")
print("V1 dataset modified         : NO")
print("Production model changed    : NO")

print(
    f"\nRainfall-sensitive predictions: "
    f"{changed_count}/{total_count} "
    f"({change_percentage:.2f}%)"
)

print("=" * 70)
print("STEP 2K COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 2L — V2 WALK-FORWARD / TEMPORAL ROBUSTNESS AUDIT
# ======================================================================

import os
import joblib
import pandas as pd
import numpy as np

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

print("=" * 70)
print("STEP 2L — V2 WALK-FORWARD / TEMPORAL ROBUSTNESS AUDIT")
print("=" * 70)


# ======================================================================
# PATHS
# ======================================================================

V2_DATASET = r"D:\document\Brahmaputra-FloodCast\data\cleaned\v1_v2_comparable_dataset.csv"

V1_MODEL_PATH = r"D:\document\Brahmaputra-FloodCast\models\floodsense_random_forest.joblib"

V2_MODEL_PATH = r"D:\document\Brahmaputra-FloodCast\models\floodsense_v2_rainfall_random_forest.joblib"


# ======================================================================
# FEATURE CONFIGURATION
# ======================================================================

V1_FEATURES = [
    "Current_Water_Level",
    "Month",
    "Hour",
    "DayOfYear",
    "WaterLevel_Lag_1h",
    "WaterLevel_Lag_3h",
    "WaterLevel_Lag_6h",
    "WaterLevel_Lag_12h",
    "WaterLevel_Lag_24h",
    "WaterLevel_RollingMean_6h",
    "WaterLevel_RollingMean_12h",
    "WaterLevel_RollingMean_24h",
    "WaterLevel_RollingMax_24h",
    "WaterLevel_RollingMin_24h",
    "WaterLevel_Change_1h",
    "WaterLevel_Change_3h",
    "WaterLevel_Change_6h",
    "WaterLevel_Change_12h",
    "WaterLevel_Change_24h"
]

RAINFALL_FEATURES = [
    "rainfall_1h_sum",
    "rainfall_3h_sum",
    "rainfall_6h_sum",
    "rainfall_12h_sum",
    "rainfall_24h_sum"
]

V2_FEATURES = V1_FEATURES + RAINFALL_FEATURES

TARGET = "High_Water_6h"

TIMESTAMP = "Data Acquisition Time"


# ======================================================================
# FILE VALIDATION
# ======================================================================

print("\n" + "-" * 70)
print("FILE VALIDATION")
print("-" * 70)

print("V2 dataset exists :", os.path.exists(V2_DATASET))
print("V1 model exists   :", os.path.exists(V1_MODEL_PATH))
print("V2 model exists   :", os.path.exists(V2_MODEL_PATH))


# ======================================================================
# LOAD DATA
# ======================================================================

print("\n" + "-" * 70)
print("LOADING V2 COMPARABLE DATASET")
print("-" * 70)

df = pd.read_csv(V2_DATASET)

df[TIMESTAMP] = pd.to_datetime(df[TIMESTAMP], errors="coerce")

df = df.sort_values(TIMESTAMP).reset_index(drop=True)

print("Rows    :", len(df))
print("Columns :", len(df))

print("\nTimestamp range:")
print("Start:", df[TIMESTAMP].min())
print("End  :", df[TIMESTAMP].max())


# ======================================================================
# FEATURE VALIDATION
# ======================================================================

print("\n" + "-" * 70)
print("FEATURE VALIDATION")
print("-" * 70)

missing_v1 = [f for f in V1_FEATURES if f not in df.columns]
missing_rain = [f for f in RAINFALL_FEATURES if f not in df.columns]

print("Missing V1 features      :", missing_v1)
print("Missing rainfall features:", missing_rain)

if missing_v1 or missing_rain:
    raise ValueError("Required features are missing.")

print("PASS — All V2 features exist.")


# ======================================================================
# COMPLETE DATA CHECK
# ======================================================================

required_columns = V2_FEATURES + [TARGET, TIMESTAMP]

complete_mask = df[required_columns].notna().all(axis=1)

df = df.loc[complete_mask].copy()

df = df.sort_values(TIMESTAMP).reset_index(drop=True)

print("\n" + "-" * 70)
print("COMPLETE DATA CHECK")
print("-" * 70)

print("Complete V2 rows:", len(df))

print("Missing values remaining:",
      int(df[V2_FEATURES + [TARGET]].isna().sum().sum()))


# ======================================================================
# LOAD EXISTING MODELS
# ======================================================================

print("\n" + "-" * 70)
print("LOADING EXISTING MODELS")
print("-" * 70)

v1_model = joblib.load(V1_MODEL_PATH)
v2_model = joblib.load(V2_MODEL_PATH)

print("V1 model:", type(v1_model).__name__)
print("V2 model:", type(v2_model).__name__)

print("V1 model trees:", getattr(v1_model, "n_estimators", "N/A"))
print("V2 model trees:", getattr(v2_model, "n_estimators", "N/A"))

print("NO PRODUCTION MODEL MODIFICATION.")


# ======================================================================
# WALK-FORWARD CONFIGURATION
# ======================================================================

print("\n" + "-" * 70)
print("WALK-FORWARD CONFIGURATION")
print("-" * 70)

N_FOLDS = 4

n = len(df)

# Minimum initial training size
initial_train_size = int(n * 0.50)

remaining = n - initial_train_size

test_size = remaining // N_FOLDS

print("Total complete rows :", n)
print("Initial train rows  :", initial_train_size)
print("Validation folds    :", N_FOLDS)
print("Approx. fold size   :", test_size)


# ======================================================================
# WALK-FORWARD EVALUATION
# ======================================================================

results = []

print("\n" + "=" * 70)
print("WALK-FORWARD FOLD EVALUATION")
print("=" * 70)


for fold in range(1, N_FOLDS + 1):

    train_end = initial_train_size + (fold - 1) * test_size

    if fold < N_FOLDS:
        test_start = train_end
        test_end = train_end + test_size
    else:
        test_start = train_end
        test_end = n

    train_df = df.iloc[:train_end].copy()
    test_df = df.iloc[test_start:test_end].copy()

    X_train_v2 = train_df[V2_FEATURES]
    y_train = train_df[TARGET]

    X_test_v2 = test_df[V2_FEATURES]
    y_test = test_df[TARGET]

    X_test_v1 = test_df[V1_FEATURES]

    print("\n" + "-" * 70)
    print(f"FOLD {fold}")
    print("-" * 70)

    print("Train rows:", len(train_df))
    print("Test rows :", len(test_df))

    print(
        "Train period:",
        train_df[TIMESTAMP].min(),
        "→",
        train_df[TIMESTAMP].max()
    )

    print(
        "Test period :",
        test_df[TIMESTAMP].min(),
        "→",
        test_df[TIMESTAMP].max()
    )

    train_classes = sorted(y_train.unique().tolist())
    test_classes = sorted(y_test.unique().tolist())

    print("Train classes:", train_classes)
    print("Test classes :", test_classes)

    # --------------------------------------------------------------
    # TRAIN TEMPORARY V2 MODEL
    # --------------------------------------------------------------

    fold_model = RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    )

    fold_model.fit(X_train_v2, y_train)

    # --------------------------------------------------------------
    # PREDICTIONS
    # --------------------------------------------------------------

    pred_v2 = fold_model.predict(X_test_v2)

    # Existing V1 model
    pred_v1 = v1_model.predict(X_test_v1)

    # --------------------------------------------------------------
    # AGREEMENT
    # --------------------------------------------------------------

    agreement = np.mean(pred_v1 == pred_v2)

    # --------------------------------------------------------------
    # METRICS
    # --------------------------------------------------------------

    if len(test_classes) == 2:

        v1_accuracy = accuracy_score(y_test, pred_v1)
        v1_precision = precision_score(
            y_test, pred_v1, zero_division=0
        )
        v1_recall = recall_score(
            y_test, pred_v1, zero_division=0
        )
        v1_f1 = f1_score(
            y_test, pred_v1, zero_division=0
        )

        v2_accuracy = accuracy_score(y_test, pred_v2)
        v2_precision = precision_score(
            y_test, pred_v2, zero_division=0
        )
        v2_recall = recall_score(
            y_test, pred_v2, zero_division=0
        )
        v2_f1 = f1_score(
            y_test, pred_v2, zero_division=0
        )

        print("\nV1 Metrics")
        print("Accuracy :", round(v1_accuracy, 4))
        print("Precision:", round(v1_precision, 4))
        print("Recall   :", round(v1_recall, 4))
        print("F1       :", round(v1_f1, 4))

        print("\nV2 Metrics")
        print("Accuracy :", round(v2_accuracy, 4))
        print("Precision:", round(v2_precision, 4))
        print("Recall   :", round(v2_recall, 4))
        print("F1       :", round(v2_f1, 4))

        print("\nV2 - V1 F1 difference:",
              round(v2_f1 - v1_f1, 4))

        metrics_valid = True

    else:

        v1_accuracy = np.nan
        v1_precision = np.nan
        v1_recall = np.nan
        v1_f1 = np.nan

        v2_accuracy = np.nan
        v2_precision = np.nan
        v2_recall = np.nan
        v2_f1 = np.nan

        metrics_valid = False

        print(
            "\nWARNING — Test fold contains only one target class."
        )

        print(
            "Binary classification metrics are NOT treated "
            "as representative for this fold."
        )

    print(
        "Prediction agreement:",
        round(agreement * 100, 2),
        "%"
    )

    print("\nV2 confusion matrix:")

    print(
        confusion_matrix(
            y_test,
            pred_v2,
            labels=[0, 1]
        )
    )

    results.append({
        "Fold": fold,
        "Train_Rows": len(train_df),
        "Test_Rows": len(test_df),
        "Train_Start": train_df[TIMESTAMP].min(),
        "Train_End": train_df[TIMESTAMP].max(),
        "Test_Start": test_df[TIMESTAMP].min(),
        "Test_End": test_df[TIMESTAMP].max(),
        "Train_Classes": str(train_classes),
        "Test_Classes": str(test_classes),
        "Metrics_Valid": metrics_valid,
        "V1_F1": v1_f1,
        "V2_F1": v2_f1,
        "V2_minus_V1_F1": (
            v2_f1 - v1_f1
            if metrics_valid else np.nan
        ),
        "Prediction_Agreement": agreement
    })


# ======================================================================
# FINAL SUMMARY
# ======================================================================

results_df = pd.DataFrame(results)

print("\n")
print("=" * 70)
print("STEP 2L — FINAL SUMMARY")
print("=" * 70)

print("\n")
print(results_df.to_string(index=False))


# ======================================================================
# VALID FOLD ANALYSIS
# ======================================================================

valid_results = results_df[
    results_df["Metrics_Valid"] == True
].copy()

print("\n" + "-" * 70)
print("VALID BINARY-CLASS FOLDS")
print("-" * 70)

print("Valid folds:", len(valid_results))

if len(valid_results) > 0:

    print(
        "Mean V1 F1:",
        round(valid_results["V1_F1"].mean(), 4)
    )

    print(
        "Mean V2 F1:",
        round(valid_results["V2_F1"].mean(), 4)
    )

    print(
        "Mean V2-V1 F1 improvement:",
        round(
            valid_results["V2_minus_V1_F1"].mean(),
            4
        )
    )

    print(
        "Mean prediction agreement:",
        round(
            valid_results["Prediction_Agreement"].mean() * 100,
            2
        ),
        "%"
    )

else:

    print(
        "No fold contains both target classes."
    )

    print(
        "Therefore no reliable binary-classification "
        "comparison can be established from this dataset."
    )


# ======================================================================
# SAVE AUDIT
# ======================================================================

OUTPUT_PATH = (
    r"D:\document\Brahmaputra-FloodCast"
    r"\data\cleaned\v2_walk_forward_audit.csv"
)

results_df.to_csv(
    OUTPUT_PATH,
    index=False
)

print("\n" + "-" * 70)
print("AUDIT SAVED")
print("-" * 70)

print("Path:", OUTPUT_PATH)

print("\n")
print("=" * 70)
print("STEP 2L COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 2M — RAINFALL EVENT / TEMPORAL VALIDATION AUDIT
# ======================================================================

import os
import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 2M — RAINFALL EVENT / TEMPORAL VALIDATION AUDIT")
print("=" * 70)


# ======================================================================
# PATH
# ======================================================================

V2_DATASET = (
    r"D:\document\Brahmaputra-FloodCast"
    r"\data\cleaned\v1_v2_comparable_dataset.csv"
)


# ======================================================================
# FEATURES
# ======================================================================

RAINFALL_FEATURES = [
    "rainfall_1h_sum",
    "rainfall_3h_sum",
    "rainfall_6h_sum",
    "rainfall_12h_sum",
    "rainfall_24h_sum"
]

TARGET = "High_Water_6h"

TIMESTAMP = "Data Acquisition Time"


# ======================================================================
# LOAD
# ======================================================================

print("\n" + "-" * 70)
print("LOADING DATA")
print("-" * 70)

df = pd.read_csv(V2_DATASET)

df[TIMESTAMP] = pd.to_datetime(
    df[TIMESTAMP],
    errors="coerce"
)

df = df.sort_values(TIMESTAMP).reset_index(drop=True)

print("Rows:", len(df))


# ======================================================================
# COMPLETE RAINFALL DATA
# ======================================================================

complete_mask = df[RAINFALL_FEATURES].notna().all(axis=1)

rain_df = df.loc[complete_mask].copy()

print("\nComplete rainfall rows:", len(rain_df))


# ======================================================================
# TARGET DISTRIBUTION
# ======================================================================

print("\n" + "-" * 70)
print("TARGET DISTRIBUTION")
print("-" * 70)

print(
    rain_df[TARGET].value_counts()
    .sort_index()
)


# ======================================================================
# RAINFALL STATISTICS BY TARGET
# ======================================================================

print("\n" + "-" * 70)
print("RAINFALL BY TARGET")
print("-" * 70)

for feature in RAINFALL_FEATURES:

    print("\n" + feature)

    stats = (
        rain_df
        .groupby(TARGET)[feature]
        .agg([
            "count",
            "mean",
            "median",
            "std",
            "min",
            "max"
        ])
    )

    print(stats)


# ======================================================================
# MEDIAN DIFFERENCE
# ======================================================================

print("\n" + "-" * 70)
print("POSITIVE vs NEGATIVE MEDIAN DIFFERENCE")
print("-" * 70)

for feature in RAINFALL_FEATURES:

    median_0 = rain_df.loc[
        rain_df[TARGET] == 0,
        feature
    ].median()

    median_1 = rain_df.loc[
        rain_df[TARGET] == 1,
        feature
    ].median()

    ratio = (
        median_1 / median_0
        if median_0 != 0
        else np.nan
    )

    print(
        f"{feature:25s} "
        f"negative={median_0:.4f} | "
        f"positive={median_1:.4f} | "
        f"ratio={ratio:.4f}"
    )


# ======================================================================
# HIGH-RAINFALL THRESHOLD ANALYSIS
# ======================================================================

print("\n" + "-" * 70)
print("HIGH-RAINFALL EVENT ANALYSIS")
print("-" * 70)

for feature in RAINFALL_FEATURES:

    threshold = rain_df[feature].median()

    high_rain = rain_df[
        rain_df[feature] > threshold
    ]

    low_rain = rain_df[
        rain_df[feature] <= threshold
    ]

    high_positive_rate = (
        high_rain[TARGET].mean()
        if len(high_rain) > 0
        else np.nan
    )

    low_positive_rate = (
        low_rain[TARGET].mean()
        if len(low_rain) > 0
        else np.nan
    )

    print("\n", feature)

    print("Threshold:", threshold)

    print(
        "High-rain rows:",
        len(high_rain)
    )

    print(
        "High-rain positive rate:",
        round(high_positive_rate, 4)
    )

    print(
        "Low-rain rows:",
        len(low_rain)
    )

    print(
        "Low-rain positive rate:",
        round(low_positive_rate, 4)
    )


# ======================================================================
# RAINFALL EVENT CONCENTRATION
# ======================================================================

print("\n" + "-" * 70)
print("RAINFALL EVENT CONCENTRATION")
print("-" * 70)

for feature in RAINFALL_FEATURES:

    threshold = rain_df[feature].quantile(0.75)

    top_rain = rain_df[
        rain_df[feature] >= threshold
    ]

    positive_count = (
        top_rain[TARGET] == 1
    ).sum()

    total_positive = (
        rain_df[TARGET] == 1
    ).sum()

    concentration = (
        positive_count / total_positive
        if total_positive > 0
        else np.nan
    )

    print(
        f"{feature:25s} "
        f"Top-25%-rain positive concentration: "
        f"{concentration:.4f}"
    )


# ======================================================================
# TEMPORAL ORDER CHECK
# ======================================================================

print("\n" + "-" * 70)
print("TEMPORAL ORDER CHECK")
print("-" * 70)

print(
    "Chronologically sorted:",
    rain_df[TIMESTAMP].is_monotonic_increasing
)

print(
    "First rainfall timestamp:",
    rain_df[TIMESTAMP].min()
)

print(
    "Last rainfall timestamp:",
    rain_df[TIMESTAMP].max()
)


# ======================================================================
# RAINFALL FEATURE CONSISTENCY
# ======================================================================

print("\n" + "-" * 70)
print("RAINFALL WINDOW CONSISTENCY")
print("-" * 70)

checks = {
    "1h <= 3h":
        rain_df["rainfall_1h_sum"]
        <= rain_df["rainfall_3h_sum"],

    "3h <= 6h":
        rain_df["rainfall_3h_sum"]
        <= rain_df["rainfall_6h_sum"],

    "6h <= 12h":
        rain_df["rainfall_6h_sum"]
        <= rain_df["rainfall_12h_sum"],

    "12h <= 24h":
        rain_df["rainfall_12h_sum"]
        <= rain_df["rainfall_24h_sum"]
}

for name, result in checks.items():

    print(
        f"{name:15s}: "
        f"{result.sum()} / {len(result)} PASS"
    )


# ======================================================================
# FINAL STATUS
# ======================================================================

print("\n")
print("=" * 70)
print("STEP 2M — FINAL STATUS")
print("=" * 70)

print("Rainfall event analysis : COMPLETE")
print("Target separation       : COMPLETE")
print("Temporal order check    : COMPLETE")
print("Window consistency      : COMPLETE")

print("\nNO MODEL TRAINING")
print("NO MODEL RETRAINING")
print("NO V1 MODEL MODIFICATION")
print("NO V2 MODEL MODIFICATION")
print("NO PRODUCTION MODEL CHANGE")

print("=" * 70)
print("STEP 2M COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 2N — V2 RAINFALL LEAKAGE & CAUSAL FEATURE AUDIT
# ======================================================================

import os
import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 2N — V2 RAINFALL LEAKAGE & CAUSAL FEATURE AUDIT")
print("=" * 70)


# ======================================================================
# PATHS
# ======================================================================

V2_DATASET = (
    r"D:\document\Brahmaputra-FloodCast"
    r"\data\cleaned\v1_v2_comparable_dataset.csv"
)

RAIN_DATASET = (
    r"D:\document\Brahmaputra-FloodCast"
    r"\data\cleaned\water_rainfall_integrated_clean.csv"
)


# ======================================================================
# CONFIGURATION
# ======================================================================

TIMESTAMP = "Data Acquisition Time"

TARGET = "High_Water_6h"

RAINFALL_FEATURES = [
    "rainfall_1h_sum",
    "rainfall_3h_sum",
    "rainfall_6h_sum",
    "rainfall_12h_sum",
    "rainfall_24h_sum"
]

WINDOW_HOURS = {
    "rainfall_1h_sum": 1,
    "rainfall_3h_sum": 3,
    "rainfall_6h_sum": 6,
    "rainfall_12h_sum": 12,
    "rainfall_24h_sum": 24
}


# ======================================================================
# FILE VALIDATION
# ======================================================================

print("\n" + "-" * 70)
print("FILE VALIDATION")
print("-" * 70)

print("V2 dataset exists :", os.path.exists(V2_DATASET))
print("Rainfall dataset exists :", os.path.exists(RAIN_DATASET))


# ======================================================================
# LOAD V2 DATASET
# ======================================================================

print("\n" + "-" * 70)
print("LOADING V2 DATASET")
print("-" * 70)

v2 = pd.read_csv(V2_DATASET)

v2[TIMESTAMP] = pd.to_datetime(
    v2[TIMESTAMP],
    errors="coerce"
)

v2 = (
    v2
    .sort_values(TIMESTAMP)
    .reset_index(drop=True)
)

print("V2 rows:", len(v2))


# ======================================================================
# REQUIRED COLUMN CHECK
# ======================================================================

required_columns = [
    TIMESTAMP,
    TARGET
] + RAINFALL_FEATURES

missing = [
    c for c in required_columns
    if c not in v2.columns
]

print("\nMissing required columns:", missing)

if missing:
    raise ValueError(
        "Required columns missing from V2 dataset."
    )


# ======================================================================
# COMPLETE RAINFALL DATA
# ======================================================================

rain_df = v2[
    v2[RAINFALL_FEATURES]
    .notna()
    .all(axis=1)
].copy()

print("\nComplete rainfall rows:", len(rain_df))


# ======================================================================
# TIMESTAMP ORDER
# ======================================================================

print("\n" + "-" * 70)
print("TIMESTAMP ORDER CHECK")
print("-" * 70)

is_sorted = rain_df[TIMESTAMP].is_monotonic_increasing

print("Chronologically sorted:", is_sorted)

print(
    "Start:",
    rain_df[TIMESTAMP].min()
)

print(
    "End:",
    rain_df[TIMESTAMP].max()
)


# ======================================================================
# DUPLICATE TIMESTAMP CHECK
# ======================================================================

print("\n" + "-" * 70)
print("DUPLICATE TIMESTAMP CHECK")
print("-" * 70)

duplicate_count = rain_df[TIMESTAMP].duplicated().sum()

print(
    "Duplicate timestamps:",
    duplicate_count
)


# ======================================================================
# TIMESTAMP GAP AUDIT
# ======================================================================

print("\n" + "-" * 70)
print("RAINFALL TIMESTAMP GAP AUDIT")
print("-" * 70)

time_diff = rain_df[TIMESTAMP].diff()

gap_counts = time_diff.value_counts()

print("Most common timestamp gaps:")

print(
    gap_counts.head(10)
)


# ======================================================================
# FUTURE TIMESTAMP AVAILABILITY TEST
# ======================================================================
#
# IMPORTANT:
# We test whether rainfall values at timestamp t can be explained
# using observations AFTER t.
#
# This is a structural audit only.
# Exact semantic validation depends on the original rainfall
# aggregation methodology.
# ======================================================================

print("\n" + "-" * 70)
print("FUTURE INFORMATION STRUCTURAL AUDIT")
print("-" * 70)

for feature, hours in WINDOW_HOURS.items():

    future_timestamps = (
        rain_df[TIMESTAMP]
        .shift(-1)
    )

    current_timestamp = rain_df[TIMESTAMP]

    future_delta = (
        future_timestamps - current_timestamp
    )

    future_records_within_window = (
        future_delta <= pd.Timedelta(hours=hours)
    ) & (
        future_delta > pd.Timedelta(0)
    )

    count = future_records_within_window.sum()

    print(
        f"{feature:25s} | "
        f"future observations within nominal window: {count}"
    )


# ======================================================================
# TARGET FUTURE WINDOW
# ======================================================================

print("\n" + "-" * 70)
print("TARGET FUTURE WINDOW")
print("-" * 70)

print(
    "Target:",
    TARGET
)

print(
    "Target represents:",
    "High water condition within next 6 hours"
)

print(
    "Prediction timestamp = t"
)

print(
    "Target window = t → t+6h"
)


# ======================================================================
# RAINFALL / TARGET TEMPORAL RELATIONSHIP
# ======================================================================

print("\n" + "-" * 70)
print("RAINFALL vs TARGET TEMPORAL CHECK")
print("-" * 70)

for feature, hours in WINDOW_HOURS.items():

    # The rainfall feature itself is associated with timestamp t.
    # We therefore inspect whether the feature timestamp occurs
    # before the target evaluation window ends.

    target_end = (
        rain_df[TIMESTAMP]
        + pd.Timedelta(hours=6)
    )

    feature_timestamp = rain_df[TIMESTAMP]

    valid_temporal_order = (
        feature_timestamp <= target_end
    )

    print(
        f"{feature:25s} | "
        f"timestamps before target end: "
        f"{valid_temporal_order.sum()}/{len(rain_df)}"
    )


# ======================================================================
# RAINFALL RANGE / NEGATIVE VALUE CHECK
# ======================================================================

print("\n" + "-" * 70)
print("RAINFALL VALUE SANITY CHECK")
print("-" * 70)

for feature in RAINFALL_FEATURES:

    negative_count = (
        rain_df[feature] < 0
    ).sum()

    zero_count = (
        rain_df[feature] == 0
    ).sum()

    print(
        f"{feature:25s} | "
        f"negative={negative_count} | "
        f"zero={zero_count}"
    )


# ======================================================================
# RAW RAINFALL DATASET CHECK
# ======================================================================

print("\n" + "-" * 70)
print("RAW RAINFALL DATASET CHECK")
print("-" * 70)

raw = pd.read_csv(RAIN_DATASET)

raw[TIMESTAMP] = pd.to_datetime(
    raw[TIMESTAMP],
    errors="coerce"
)

raw = (
    raw
    .sort_values(TIMESTAMP)
    .reset_index(drop=True)
)

print("Raw rainfall rows:", len(raw))

print(
    "Raw timestamp start:",
    raw[TIMESTAMP].min()
)

print(
    "Raw timestamp end:",
    raw[TIMESTAMP].max()
)

print(
    "Raw duplicate timestamps:",
    raw[TIMESTAMP].duplicated().sum()
)


# ======================================================================
# COMPARE V2 VALUES AGAINST RAW DATASET
# ======================================================================

print("\n" + "-" * 70)
print("V2 vs RAW RAINFALL VALUE CONSISTENCY")
print("-" * 70)

raw_lookup = (
    raw[
        [TIMESTAMP] + RAINFALL_FEATURES
    ]
    .drop_duplicates(subset=[TIMESTAMP])
    .set_index(TIMESTAMP)
)

comparison_results = []

for feature in RAINFALL_FEATURES:

    common_index = (
        rain_df[TIMESTAMP]
        .isin(raw_lookup.index)
    )

    v2_values = (
        rain_df.loc[
            common_index,
            [TIMESTAMP, feature]
        ]
        .set_index(TIMESTAMP)[feature]
    )

    raw_values = (
        raw_lookup
        .loc[v2_values.index, feature]
    )

    differences = (
        v2_values - raw_values
    ).abs()

    mismatch_count = (
        differences > 1e-9
    ).sum()

    comparison_results.append(
        {
            "Feature": feature,
            "Matched": len(v2_values),
            "Mismatches": mismatch_count,
            "Max_Abs_Difference": (
                differences.max()
                if len(differences) > 0
                else np.nan
            )
        }
    )

comparison_df = pd.DataFrame(
    comparison_results
)

print(comparison_df.to_string(index=False))


# ======================================================================
# FINAL INTERPRETATION FLAGS
# ======================================================================

print("\n" + "=" * 70)
print("STEP 2N — FINAL STATUS")
print("=" * 70)

print(
    "Timestamp ordering checked : PASS"
    if is_sorted
    else
    "Timestamp ordering checked : FAIL"
)

print(
    "Duplicate timestamp check  : PASS"
    if duplicate_count == 0
    else
    "Duplicate timestamp check  : WARNING"
)

if (
    comparison_df["Mismatches"].sum()
    == 0
):
    print(
        "V2/raw rainfall consistency : PASS"
    )
else:
    print(
        "V2/raw rainfall consistency : WARNING"
    )

print("\nIMPORTANT:")
print(
    "This audit checks structural consistency."
)
print(
    "Exact rainfall window semantics depend "
    "on the original rainfall aggregation logic."
)

print("\nNO MODEL TRAINING")
print("NO MODEL RETRAINING")
print("NO V1 MODEL MODIFICATION")
print("NO V2 MODEL MODIFICATION")
print("NO PRODUCTION MODEL CHANGE")

print("=" * 70)
print("STEP 2N COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 2O — TRUE RAINFALL WINDOW SEMANTICS & LEAKAGE AUDIT
# ======================================================================

import os
import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 2O — TRUE RAINFALL WINDOW SEMANTICS & LEAKAGE AUDIT")
print("=" * 70)


# ======================================================================
# PATHS
# ======================================================================

V2_PATH = (
    r"D:\document\Brahmaputra-FloodCast"
    r"\data\cleaned\v1_v2_comparable_dataset.csv"
)

RAW_RAIN_PATH = (
    r"D:\document\Brahmaputra-FloodCast"
    r"\data\cleaned\water_rainfall_integrated_clean.csv"
)


# ======================================================================
# CONFIGURATION
# ======================================================================

TIME_COL = "Data Acquisition Time"

RAIN_FEATURES = {
    "rainfall_1h_sum": 1,
    "rainfall_3h_sum": 3,
    "rainfall_6h_sum": 6,
    "rainfall_12h_sum": 12,
    "rainfall_24h_sum": 24
}

TARGET = "High_Water_6h"


# ======================================================================
# FILE VALIDATION
# ======================================================================

print("\n" + "-" * 70)
print("FILE VALIDATION")
print("-" * 70)

print("V2 dataset exists :", os.path.exists(V2_PATH))
print("Raw rainfall dataset exists :", os.path.exists(RAW_RAIN_PATH))

if not os.path.exists(V2_PATH):
    raise FileNotFoundError("V2 dataset not found.")

if not os.path.exists(RAW_RAIN_PATH):
    raise FileNotFoundError("Raw rainfall dataset not found.")


# ======================================================================
# LOAD DATA
# ======================================================================

print("\n" + "-" * 70)
print("LOADING DATA")
print("-" * 70)

v2 = pd.read_csv(V2_PATH)
raw = pd.read_csv(RAW_RAIN_PATH)

v2[TIME_COL] = pd.to_datetime(
    v2[TIME_COL],
    errors="coerce"
)

raw[TIME_COL] = pd.to_datetime(
    raw[TIME_COL],
    errors="coerce"
)

v2 = (
    v2
    .sort_values(TIME_COL)
    .reset_index(drop=True)
)

raw = (
    raw
    .sort_values(TIME_COL)
    .reset_index(drop=True)
)

print("V2 rows :", len(v2))
print("Raw rainfall rows :", len(raw))


# ======================================================================
# COMPLETE RAINFALL RECORDS
# ======================================================================

complete = v2[
    v2[list(RAIN_FEATURES.keys())]
    .notna()
    .all(axis=1)
].copy()

print("\nComplete V2 rainfall rows:", len(complete))


# ======================================================================
# RAW RAINFALL TIMESTAMP LOOKUP
# ======================================================================

raw = raw.drop_duplicates(
    subset=[TIME_COL]
)

raw_indexed = raw.set_index(TIME_COL)


# ======================================================================
# FUNCTION:
# RECONSTRUCT PAST-ONLY RAINFALL WINDOW
# ======================================================================

def reconstruct_past_window(
    timestamp,
    hours,
    raw_series
):
    """
    Reconstruct rainfall using ONLY observations at or before t.

    Window:
        t-hours+1h ... t

    No future timestamp is allowed.
    """

    start_time = (
        timestamp
        - pd.Timedelta(hours=hours - 1)
    )

    end_time = timestamp

    window = raw_series.loc[
        (raw_series.index >= start_time) &
        (raw_series.index <= end_time)
    ]

    return window.sum(min_count=1)


# ======================================================================
# SEMANTIC RECONSTRUCTION AUDIT
# ======================================================================

print("\n" + "-" * 70)
print("PAST-ONLY RAINFALL WINDOW RECONSTRUCTION")
print("-" * 70)

results = []

sample_size = min(
    len(complete),
    256
)

audit_sample = complete.head(sample_size).copy()

for feature, hours in RAIN_FEATURES.items():

    raw_series = raw_indexed[feature]

    reconstructed = []

    for ts in audit_sample[TIME_COL]:

        value = reconstruct_past_window(
            ts,
            hours,
            raw_series
        )

        reconstructed.append(value)

    audit_sample[
        feature + "_reconstructed"
    ] = reconstructed

    diff = (
        audit_sample[feature]
        - audit_sample[
            feature + "_reconstructed"
        ]
    ).abs()

    valid = diff.notna()

    mismatch_count = (
        diff[valid] > 1e-6
    ).sum()

    max_difference = (
        diff[valid].max()
        if valid.any()
        else np.nan
    )

    results.append({
        "Feature": feature,
        "Window_Hours": hours,
        "Checked_Rows": int(valid.sum()),
        "Mismatches": int(mismatch_count),
        "Max_Abs_Difference": max_difference
    })

    print(
        f"{feature:25s} | "
        f"Window={hours:2d}h | "
        f"Checked={valid.sum():3d} | "
        f"Mismatches={mismatch_count:3d} | "
        f"MaxDiff={max_difference}"
    )


semantic_results = pd.DataFrame(results)


# ======================================================================
# FUTURE-ONLY RECONSTRUCTION TEST
# ======================================================================

print("\n" + "-" * 70)
print("FUTURE-ONLY INFORMATION TEST")
print("-" * 70)

future_results = []

for feature, hours in RAIN_FEATURES.items():

    raw_series = raw_indexed[feature]

    changed_count = 0
    checked_count = 0

    for ts in audit_sample[TIME_COL]:

        current_value = raw_series.get(
            ts,
            np.nan
        )

        future_start = (
            ts + pd.Timedelta(hours=1)
        )

        future_end = (
            ts + pd.Timedelta(hours=hours)
        )

        future_values = raw_series.loc[
            (raw_series.index >= future_start) &
            (raw_series.index <= future_end)
        ]

        if (
            pd.notna(current_value)
            and len(future_values) > 0
        ):

            checked_count += 1

            future_sum = future_values.sum()

            if abs(
                current_value - future_sum
            ) > 1e-6:

                changed_count += 1

    future_results.append({
        "Feature": feature,
        "Checked": checked_count,
        "Future_Influence_Indicator": changed_count
    })

    print(
        f"{feature:25s} | "
        f"Checked={checked_count:3d} | "
        f"Future indicator={changed_count:3d}"
    )


future_results_df = pd.DataFrame(
    future_results
)


# ======================================================================
# TARGET WINDOW CHECK
# ======================================================================

print("\n" + "-" * 70)
print("TARGET WINDOW CHECK")
print("-" * 70)

print("Prediction time : t")
print("Target          : High_Water_6h")
print("Target horizon  : t → t+6h")

target_end = (
    complete[TIME_COL]
    + pd.Timedelta(hours=6)
)

print(
    "Target end range:",
    target_end.min(),
    "→",
    target_end.max()
)


# ======================================================================
# CHECK WHETHER RAINFALL FEATURES REQUIRE FUTURE DATA
# ======================================================================

print("\n" + "-" * 70)
print("CAUSAL AVAILABILITY CHECK")
print("-" * 70)

causal_summary = []

for feature, hours in RAIN_FEATURES.items():

    # A past-looking window ending at t is causal.
    # A window extending beyond t is non-causal.

    window_start = (
        complete[TIME_COL]
        - pd.Timedelta(hours=hours - 1)
    )

    window_end = complete[TIME_COL]

    future_extension = (
        window_end > complete[TIME_COL]
    ).sum()

    causal_summary.append({
        "Feature": feature,
        "Window_Hours": hours,
        "Future_Extension_Rows": int(
            future_extension
        )
    })

    print(
        f"{feature:25s} | "
        f"Future extension rows: "
        f"{future_extension}"
    )


# ======================================================================
# SAMPLE MANUAL INSPECTION
# ======================================================================

print("\n" + "-" * 70)
print("SAMPLE SEMANTIC RECONSTRUCTION")
print("-" * 70)

display_columns = [TIME_COL]

for feature in RAIN_FEATURES:

    display_columns.extend([
        feature,
        feature + "_reconstructed"
    ])

print(
    audit_sample[
        display_columns
    ].head(15).to_string(index=False)
)


# ======================================================================
# FINAL VERDICT
# ======================================================================

print("\n" + "=" * 70)
print("STEP 2O — FINAL VERDICT")
print("=" * 70)

total_mismatches = (
    semantic_results["Mismatches"]
    .sum()
)

if total_mismatches == 0:

    print(
        "PAST-ONLY RECONSTRUCTION : PASS"
    )

    print(
        "No mismatch found between "
        "V2 rainfall values and reconstructed "
        "past-only rainfall windows."
    )

    print(
        "PRELIMINARY LEAKAGE STATUS : NOT DETECTED"
    )

else:

    print(
        "PAST-ONLY RECONSTRUCTION : WARNING"
    )

    print(
        f"Total mismatches detected: "
        f"{total_mismatches}"
    )

    print(
        "PRELIMINARY LEAKAGE STATUS : "
        "SEMANTICALLY UNCERTAIN"
    )


print("\nNO MODEL TRAINING")
print("NO MODEL RETRAINING")
print("NO V1 MODEL MODIFICATION")
print("NO V2 MODEL MODIFICATION")
print("NO PRODUCTION MODEL CHANGE")

print("=" * 70)
print("STEP 2O COMPLETE")
print("=" * 70)
```

```python
# ==============================================================
# STEP 2S — RAINFALL AGGREGATION SEMANTICS ANALYZER
# ==============================================================

from pathlib import Path
import json
import re

BASE = Path(r"D:\document\Brahmaputra-FloodCast")

NOTEBOOKS = [
    BASE / "Person1_Rainfall_Integration.ipynb",
    BASE / "05_Rainfall_V2_Model.ipynb"
]

RAIN_FEATURES = [
    "rainfall_1h_sum",
    "rainfall_3h_sum",
    "rainfall_6h_sum",
    "rainfall_12h_sum",
    "rainfall_24h_sum"
]

PATTERNS = [
    r"rolling\s*\(",
    r"shift\s*\(",
    r"resample\s*\(",
    r"groupby\s*\(",
    r"\.sum\s*\(",
    r"\.mean\s*\(",
    r"merge\s*\(",
    r"join\s*\(",
    r"transform\s*\(",
    r"expanding\s*\(",
    r"cumsum\s*\(",
]

print("=" * 70)
print("STEP 2S — RAINFALL AGGREGATION SEMANTICS ANALYZER")
print("=" * 70)

for notebook_path in NOTEBOOKS:

    print("\n")
    print("=" * 70)
    print("NOTEBOOK:", notebook_path.name)
    print("=" * 70)

    if not notebook_path.exists():
        print("NOT FOUND")
        continue

    with open(
        notebook_path,
        "r",
        encoding="utf-8"
    ) as f:
        nb = json.load(f)

    cells = nb.get("cells", [])

    for cell_no, cell in enumerate(cells, start=1):

        source = "".join(
            cell.get("source", [])
        )

        if not any(
            feature in source
            for feature in RAIN_FEATURES
        ):
            continue

        print("\n" + "-" * 70)
        print(f"CELL {cell_no}")
        print("-" * 70)

        lines = source.splitlines()

        relevant = False

        for i, line in enumerate(lines):

            line_lower = line.lower()

            feature_hit = any(
                feature.lower() in line_lower
                for feature in RAIN_FEATURES
            )

            operation_hit = any(
                re.search(
                    pattern,
                    line_lower
                )
                for pattern in PATTERNS
            )

            if feature_hit or operation_hit:

                relevant = True

                start = max(0, i - 3)
                end = min(
                    len(lines),
                    i + 4
                )

                for j in range(start, end):
                    print(
                        f"{j + 1:4d}: {lines[j]}"
                    )

                print()

        if not relevant:
            print(
                "Rainfall feature found, "
                "but no aggregation operation detected "
                "in nearby lines."
            )


print("\n" + "=" * 70)
print("STEP 2S — FINAL STATUS")
print("=" * 70)

print("Generation notebooks checked : PASS")
print("Aggregation operations scanned: PASS")
print("Future-window semantics checked: PENDING")
print("Model training                : NO")
print("V1 model modification         : NO")
print("V2 model modification         : NO")
print("=" * 70)
print("STEP 2S COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 2S — COMPACT RAINFALL GENERATION SEARCH
# ======================================================================

import json
from pathlib import Path

BASE = Path.cwd()

SEARCH_TERMS = [
    "rainfall_1h_sum",
    "rainfall_3h_sum",
    "rainfall_6h_sum",
    "rainfall_12h_sum",
    "rainfall_24h_sum",
]

print("=" * 70)
print("STEP 2S — COMPACT RAINFALL GENERATION SEARCH")
print("=" * 70)

found = []

# ----------------------------------------------------------------------
# SEARCH PYTHON + NOTEBOOK FILES
# ----------------------------------------------------------------------

for pattern in ["*.py", "*.ipynb"]:

    for file_path in BASE.rglob(pattern):

        try:

            if file_path.suffix == ".py":

                lines = file_path.read_text(
                    encoding="utf-8",
                    errors="ignore"
                ).splitlines()

                for line_no, line in enumerate(lines, start=1):

                    if any(term in line for term in SEARCH_TERMS):

                        found.append(
                            (
                                str(file_path),
                                line_no,
                                line.strip()
                            )
                        )

            else:

                notebook = json.loads(
                    file_path.read_text(
                        encoding="utf-8",
                        errors="ignore"
                    )
                )

                for cell_no, cell in enumerate(
                    notebook.get("cells", []),
                    start=1
                ):

                    source = "".join(
                        cell.get("source", [])
                    )

                    for line_no, line in enumerate(
                        source.splitlines(),
                        start=1
                    ):

                        if any(
                            term in line
                            for term in SEARCH_TERMS
                        ):

                            found.append(
                                (
                                    str(file_path),
                                    f"cell {cell_no}, line {line_no}",
                                    line.strip()
                                )
                            )

        except Exception:
            pass

# ----------------------------------------------------------------------
# REMOVE DUPLICATES
# ----------------------------------------------------------------------

unique_found = []

seen = set()

for item in found:

    key = (
        item[0],
        item[1],
        item[2]
    )

    if key not in seen:

        seen.add(key)
        unique_found.append(item)

# ----------------------------------------------------------------------
# PRINT ONLY IMPORTANT LINES
# ----------------------------------------------------------------------

print("\nTotal matching lines:", len(unique_found))

print("\n" + "-" * 70)
print("RAINfall FEATURE GENERATION REFERENCES")
print("-" * 70)

for file_name, location, line in unique_found:

    print(
        f"{Path(file_name).name} | "
        f"{location} | "
        f"{line}"
    )

print("\n" + "=" * 70)
print("COMPACT SEARCH COMPLETE")
print("=" * 70)
```

```python
# ==============================================================
# STEP 2S — EXACT RAINFALL ASSIGNMENT SEARCH
# ==============================================================

import json
import re

NOTEBOOK = "05_Rainfall_V2_Model.ipynb"

with open(NOTEBOOK, "r", encoding="utf-8") as f:
    nb = json.load(f)

print("=" * 70)
print("STEP 2S — EXACT RAINFALL ASSIGNMENT SEARCH")
print("=" * 70)

found = []

for cell_no, cell in enumerate(nb["cells"]):

    lines = cell.get("source", [])

    for line_no, line in enumerate(lines):

        # Only look for assignments to rainfall_*_sum
        if re.search(r'["\']?rainfall_(1h|3h|6h|12h|24h)_sum["\']?\s*=', line):

            found.append(
                (cell_no, line_no, line.strip())
            )

        # Also catch .assign(...) style generation
        elif ".assign(" in line and "rainfall_" in line:
            found.append(
                (cell_no, line_no, line.strip())
            )

print(f"Exact generation candidates: {len(found)}")

for cell_no, line_no, line in found[:20]:
    print(f"CELL {cell_no} | LINE {line_no + 1}")
    print(line)

print("=" * 70)
```

```python
# ==============================================================
# STEP 2S — FIND ACTUAL RAINFALL CALCULATION CELL
# ==============================================================

import json

NOTEBOOK = "05_Rainfall_V2_Model.ipynb"

with open(NOTEBOOK, "r", encoding="utf-8") as f:
    nb = json.load(f)

print("=" * 70)
print("STEP 2S — ACTUAL RAINFALL CALCULATION CELLS")
print("=" * 70)

keywords = [
    ".rolling(",
    ".resample(",
    ".groupby(",
    ".sum(",
    ".agg(",
    "rolling_sum"
]

found = 0

for cell_no, cell in enumerate(nb["cells"]):

    source = "".join(cell.get("source", []))

    # Must contain rainfall feature AND an actual calculation operation
    if "rainfall_1h_sum" not in source:
        continue

    if not any(k in source for k in keywords):
        continue

    # Ignore diagnostic/search cells
    if "EXACT RAINFALL ASSIGNMENT SEARCH" in source:
        continue

    print(f"\nCELL {cell_no}")

    for line_no, line in enumerate(cell.get("source", [])):
        if (
            "rainfall_" in line
            or ".rolling(" in line
            or ".resample(" in line
            or ".groupby(" in line
            or ".agg(" in line
        ):
            print(f"  {line_no + 1} | {line.rstrip()}")

    found += 1

    if found >= 5:
        break

print("\n" + "=" * 70)
print(f"Actual candidate cells shown: {found}")
print("=" * 70)
```

```python
import json
import re

print("=" * 70)
print("STEP 2S — EXACT RAINFALL FEATURE CREATION")
print("=" * 70)

NOTEBOOK = "Person1_Rainfall_Integration.ipynb"

with open(NOTEBOOK, "r", encoding="utf-8") as f:
    nb = json.load(f)

patterns = [
    r"rainfall_1h_sum\s*=",
    r"rainfall_3h_sum\s*=",
    r"rainfall_6h_sum\s*=",
    r"rainfall_12h_sum\s*=",
    r"rainfall_24h_sum\s*=",
    r"\.rolling\(",
    r"\.resample\("
]

found = []

for cell_no, cell in enumerate(nb["cells"]):
    if cell.get("cell_type") != "code":
        continue

    lines = "".join(cell.get("source", [])).splitlines()

    for line_no, line in enumerate(lines, 1):
        if any(re.search(p, line, re.IGNORECASE) for p in patterns):
            found.append((cell_no, line_no, line.strip()))

# Unique cells only
cells = sorted(set(cell_no for cell_no, _, _ in found))

print("Candidate cells:", cells)

for cell_no in cells:
    print(f"\n--- CELL {cell_no} ---")

    lines = "".join(nb["cells"][cell_no].get("source", [])).splitlines()

    # print only lines containing actual feature creation / rolling / resample
    for line_no, line in enumerate(lines, 1):
        if any(re.search(p, line, re.IGNORECASE) for p in patterns):
            print(f"{line_no:4} | {line}")

print("\n" + "=" * 70)
```

```python
# ======================================================================
# STEP 2O — TRUE RAINFALL WINDOW SEMANTICS & LEAKAGE AUDIT
# ======================================================================

import os
import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 2O — TRUE RAINFALL WINDOW SEMANTICS & LEAKAGE AUDIT")
print("=" * 70)


# ======================================================================
# 1. PATHS
# ======================================================================

PROJECT_ROOT = r"D:\document\Brahmaputra-FloodCast"

V2_PATH = os.path.join(
    PROJECT_ROOT,
    "data",
    "cleaned",
    "v1_v2_comparable_dataset.csv"
)

RAW_RAIN_PATH = os.path.join(
    PROJECT_ROOT,
    "data",
    "cleaned",
    "water_rainfall_integrated_clean.csv"
)


# ======================================================================
# 2. CONFIGURATION
# ======================================================================

TIME_COL = "Data Acquisition Time"

RAIN_FEATURES = {
    "rainfall_1h_sum": 1,
    "rainfall_3h_sum": 3,
    "rainfall_6h_sum": 6,
    "rainfall_12h_sum": 12,
    "rainfall_24h_sum": 24
}

TARGET = "High_Water_6h"


# ======================================================================
# 3. FILE VALIDATION
# ======================================================================

print("\n" + "-" * 70)
print("FILE VALIDATION")
print("-" * 70)

print("V2 dataset exists       :", os.path.exists(V2_PATH))
print("Rainfall dataset exists :", os.path.exists(RAW_RAIN_PATH))

if not os.path.exists(V2_PATH):
    raise FileNotFoundError(f"V2 dataset not found:\n{V2_PATH}")

if not os.path.exists(RAW_RAIN_PATH):
    raise FileNotFoundError(f"Rainfall dataset not found:\n{RAW_RAIN_PATH}")


# ======================================================================
# 4. LOAD DATA
# ======================================================================

print("\n" + "-" * 70)
print("LOADING DATA")
print("-" * 70)

v2 = pd.read_csv(V2_PATH)
raw = pd.read_csv(RAW_RAIN_PATH)

v2[TIME_COL] = pd.to_datetime(
    v2[TIME_COL],
    errors="coerce"
)

raw[TIME_COL] = pd.to_datetime(
    raw[TIME_COL],
    errors="coerce"
)

v2 = (
    v2
    .dropna(subset=[TIME_COL])
    .sort_values(TIME_COL)
    .reset_index(drop=True)
)

raw = (
    raw
    .dropna(subset=[TIME_COL])
    .sort_values(TIME_COL)
    .reset_index(drop=True)
)

print("V2 rows           :", len(v2))
print("Rainfall rows     :", len(raw))


# ======================================================================
# 5. REQUIRED COLUMN CHECK
# ======================================================================

required_columns = [
    TIME_COL,
    *RAIN_FEATURES.keys()
]

missing = [
    col for col in required_columns
    if col not in v2.columns
]

if missing:
    raise ValueError(
        "Missing rainfall columns in V2 dataset:\n"
        + "\n".join(missing)
    )

for col in RAIN_FEATURES:
    if col not in raw.columns:
        raise ValueError(
            f"Missing rainfall column in raw dataset: {col}"
        )


# ======================================================================
# 6. COMPLETE RAINFALL RECORDS
# ======================================================================

complete = v2[
    v2[list(RAIN_FEATURES.keys())]
    .notna()
    .all(axis=1)
].copy()

print("\nComplete V2 rainfall rows :", len(complete))


# ======================================================================
# 7. REMOVE DUPLICATE TIMESTAMPS
# ======================================================================

raw = (
    raw
    .sort_values(TIME_COL)
    .drop_duplicates(subset=[TIME_COL], keep="last")
    .reset_index(drop=True)
)

raw_indexed = raw.set_index(TIME_COL)


# ======================================================================
# 8. PAST-ONLY RECONSTRUCTION FUNCTION
# ======================================================================

def reconstruct_past_window(
    timestamp,
    hours,
    raw_series
):
    """
    Reconstruct rainfall using only observations
    available at or before prediction time t.

    1h  -> t
    3h  -> t-2h ... t
    6h  -> t-5h ... t
    12h -> t-11h ... t
    24h -> t-23h ... t
    """

    start_time = (
        timestamp
        - pd.Timedelta(hours=hours - 1)
    )

    end_time = timestamp

    window = raw_series.loc[
        (raw_series.index >= start_time)
        &
        (raw_series.index <= end_time)
    ]

    return window.sum(min_count=1)


# ======================================================================
# 9. AUDIT SAMPLE
# ======================================================================

sample_size = min(
    len(complete),
    256
)

audit_sample = (
    complete
    .head(sample_size)
    .copy()
)

print("\nAudit sample size :", len(audit_sample))


# ======================================================================
# 10. PAST-ONLY SEMANTIC RECONSTRUCTION
# ======================================================================

print("\n" + "=" * 70)
print("PAST-ONLY RAINFALL WINDOW RECONSTRUCTION")
print("=" * 70)

results = []

for feature, hours in RAIN_FEATURES.items():

    raw_series = raw_indexed[feature]

    reconstructed = []

    for ts in audit_sample[TIME_COL]:

        value = reconstruct_past_window(
            ts,
            hours,
            raw_series
        )

        reconstructed.append(value)

    reconstructed_col = feature + "_reconstructed"

    audit_sample[reconstructed_col] = reconstructed

    difference = (
        audit_sample[feature]
        -
        audit_sample[reconstructed_col]
    ).abs()

    valid = difference.notna()

    mismatch_count = int(
        (difference[valid] > 1e-6).sum()
    )

    checked = int(valid.sum())

    max_difference = (
        difference[valid].max()
        if checked > 0
        else np.nan
    )

    results.append({
        "Feature": feature,
        "Window_Hours": hours,
        "Checked_Rows": checked,
        "Mismatches": mismatch_count,
        "Max_Abs_Difference": max_difference
    })

    print(
        f"{feature:25s} | "
        f"Window={hours:2d}h | "
        f"Checked={checked:3d} | "
        f"Mismatches={mismatch_count:3d} | "
        f"MaxDiff={max_difference}"
    )


semantic_results = pd.DataFrame(results)


# ======================================================================
# 11. FUTURE INFORMATION TEST
# ======================================================================

print("\n" + "=" * 70)
print("FUTURE INFORMATION TEST")
print("=" * 70)

future_results = []

for feature, hours in RAIN_FEATURES.items():

    raw_series = raw_indexed[feature]

    checked_count = 0
    future_available_count = 0

    for ts in audit_sample[TIME_COL]:

        current_value = raw_series.get(
            ts,
            np.nan
        )

        future_start = (
            ts + pd.Timedelta(hours=1)
        )

        future_end = (
            ts + pd.Timedelta(hours=hours)
        )

        future_values = raw_series.loc[
            (raw_series.index >= future_start)
            &
            (raw_series.index <= future_end)
        ]

        if pd.notna(current_value):

            checked_count += 1

            if len(future_values) > 0:
                future_available_count += 1

    future_results.append({
        "Feature": feature,
        "Checked": checked_count,
        "Rows_With_Future_Observations": future_available_count
    })

    print(
        f"{feature:25s} | "
        f"Checked={checked_count:3d} | "
        f"Future observations available={future_available_count:3d}"
    )


future_results_df = pd.DataFrame(
    future_results
)


# ======================================================================
# 12. TARGET WINDOW CHECK
# ======================================================================

print("\n" + "=" * 70)
print("TARGET WINDOW CHECK")
print("=" * 70)

print("Prediction time : t")
print("Target          :", TARGET)
print("Target horizon  : t → t+6h")

target_end = (
    complete[TIME_COL]
    + pd.Timedelta(hours=6)
)

if len(target_end) > 0:

    print(
        "Target end range:",
        target_end.min(),
        "→",
        target_end.max()
    )


# ======================================================================
# 13. CAUSALITY CHECK
# ======================================================================

print("\n" + "=" * 70)
print("CAUSAL AVAILABILITY CHECK")
print("=" * 70)

causal_results = []

for feature, hours in RAIN_FEATURES.items():

    window_start = (
        complete[TIME_COL]
        - pd.Timedelta(hours=hours - 1)
    )

    window_end = complete[TIME_COL]

    future_extension = (
        window_end > complete[TIME_COL]
    ).sum()

    causal_results.append({
        "Feature": feature,
        "Window_Hours": hours,
        "Future_Extension_Rows": int(
            future_extension
        )
    })

    print(
        f"{feature:25s} | "
        f"Future extension rows={future_extension}"
    )


causal_results_df = pd.DataFrame(
    causal_results
)


# ======================================================================
# 14. SAMPLE RECONSTRUCTION
# ======================================================================

print("\n" + "=" * 70)
print("SAMPLE SEMANTIC RECONSTRUCTION")
print("=" * 70)

display_columns = [TIME_COL]

for feature in RAIN_FEATURES:
    display_columns.extend([
        feature,
        feature + "_reconstructed"
    ])

print(
    audit_sample[
        display_columns
    ]
    .head(10)
    .to_string(index=False)
)


# ======================================================================
# 15. VALIDATION SUMMARY
# ======================================================================

print("\n" + "=" * 70)
print("VALIDATION SUMMARY")
print("=" * 70)

print(
    semantic_results.to_string(
        index=False
    )
)


# ======================================================================
# 16. FINAL VERDICT
# ======================================================================

print("\n" + "=" * 70)
print("STEP 2O — FINAL VERDICT")
print("=" * 70)

total_mismatches = int(
    semantic_results["Mismatches"].sum()
)

if total_mismatches == 0:

    print("PAST-ONLY RECONSTRUCTION : PASS")
    print(
        "No mismatch found between the existing "
        "rainfall features and past-only reconstruction."
    )
    print(
        "PRELIMINARY LEAKAGE STATUS : NOT DETECTED"
    )

else:

    print("PAST-ONLY RECONSTRUCTION : WARNING")
    print(
        f"Total mismatches detected: {total_mismatches}"
    )
    print(
        "PRELIMINARY LEAKAGE STATUS : "
        "SEMANTICALLY UNCERTAIN"
    )


print("\nNO MODEL TRAINING")
print("NO MODEL RETRAINING")
print("NO MODEL MODIFICATION")
print("NO 19-FEATURE MODIFICATION")
print("NO PRODUCTION MODEL CHANGE")

print("=" * 70)
print("STEP 2O COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 2AF — LIVE PIPELINE FUNCTION AUDIT
# ======================================================================

import os
import re

FILES = [
    r"scripts\fetch\fetch_live_waterlevel.py",
    r"scripts\fetch\prepare_live_history.py",
    r"scripts\fetch\live_predict.py"
]

KEYWORDS = [
    "def ",
    "requests.get",
    "requests.post",
    "params",
    "pd.read_csv",
    "pd.to_datetime",
    "sort_values",
    "dropna",
    "shift(",
    "rolling(",
    "joblib.load",
    "predict(",
    "predict_proba",
    "FEATURES",
    "MODEL",
    "to_csv",
    "to_datetime"
]

print("=" * 65)
print("STEP 2AF — LIVE PIPELINE FUNCTION AUDIT")
print("=" * 65)

for path in FILES:

    print("\n" + "-" * 65)
    print("FILE:", path)

    if not os.path.exists(path):
        print("STATUS: NOT FOUND")
        continue

    with open(path, "r", encoding="utf-8") as f:
        lines = f.readlines()

    matches = []

    for line_no, line in enumerate(lines, 1):

        stripped = line.strip()

        if any(k.lower() in stripped.lower() for k in KEYWORDS):
            matches.append((line_no, stripped))

    print("Total lines:", len(lines))
    print("Important lines:", len(matches))

    # Only first 30 important lines
    for line_no, line in matches[:30]:
        print(f"{line_no:04d} | {line[:140]}")

    if len(matches) > 30:
        print(f"... {len(matches) - 30} more hidden")

print("\n" + "=" * 65)
print("STEP 2AF COMPLETE")
print("=" * 65)
```

```python
# ======================================================================
# STEP 2AH-A — FIND EXISTING LIVE HISTORY OUTPUT
# ======================================================================

import os

ROOT = r"D:\document\Brahmaputra-FloodCast"

print("=" * 65)
print("STEP 2AH-A — LIVE HISTORY OUTPUT AUDIT")
print("=" * 65)

SEARCH_DIRS = [
    os.path.join(ROOT, "data"),
    os.path.join(ROOT, "data", "cleaned"),
    os.path.join(ROOT, "data", "processed"),
    os.path.join(ROOT, "scripts", "fetch")
]

keywords = [
    "live",
    "history",
    "feature",
    "predict"
]

found = []

for folder in SEARCH_DIRS:

    if not os.path.exists(folder):
        continue

    for root, dirs, files in os.walk(folder):

        for file in files:

            if file.lower().endswith(".csv"):

                name = file.lower()

                if any(k in name for k in keywords):

                    path = os.path.join(root, file)

                    found.append(path)

for path in found:
    print(path)

print("\nTotal candidate CSV files:", len(found))

print("\n" + "=" * 65)
print("STEP 2AH-A COMPLETE")
print("=" * 65)
```

```python
# ======================================================================
# STEP 2AH-D — LIVE 19-FEATURE OUTPUT CHECK
# ======================================================================

import runpy

SCRIPT = r"D:\document\Brahmaputra-FloodCast\scripts\fetch\prepare_live_history.py"

print("=" * 65)
print("STEP 2AH-D — LIVE 19-FEATURE OUTPUT CHECK")
print("=" * 65)

result = runpy.run_path(SCRIPT)

print("\nScript execution completed.")

# ----------------------------------------------------------------------
# FIND DATAFRAME
# ----------------------------------------------------------------------

if "df" not in result:
    raise RuntimeError(
        "prepare_live_history.py did not leave final dataframe 'df'."
    )

df = result["df"]

print("\nFINAL LIVE DATAFRAME")
print("Shape:", df.shape)

# ----------------------------------------------------------------------
# LOCKED V1 FEATURES
# ----------------------------------------------------------------------

FINAL_FEATURES = [
    "Current_Water_Level",
    "Month",
    "Hour",
    "DayOfYear",
    "WaterLevel_Lag_1h",
    "WaterLevel_Lag_3h",
    "WaterLevel_Lag_6h",
    "WaterLevel_Lag_12h",
    "WaterLevel_Lag_24h",
    "WaterLevel_RollingMean_6h",
    "WaterLevel_RollingMean_12h",
    "WaterLevel_RollingMean_24h",
    "WaterLevel_RollingMax_24h",
    "WaterLevel_RollingMin_24h",
    "WaterLevel_Change_1h",
    "WaterLevel_Change_3h",
    "WaterLevel_Change_6h",
    "WaterLevel_Change_12h",
    "WaterLevel_Change_24h"
]

missing = [
    feature
    for feature in FINAL_FEATURES
    if feature not in df.columns
]

print("\n19-FEATURE CHECK")
print("Required :", len(FINAL_FEATURES))
print("Missing  :", len(missing))

if missing:
    print("\nMissing features:")
    for feature in missing:
        print(" -", feature)
else:
    print("Status   : PASS")

# ----------------------------------------------------------------------
# LATEST VALID LIVE ROW
# ----------------------------------------------------------------------

valid_live = df.dropna(
    subset=FINAL_FEATURES
).sort_values(
    "Data Acquisition Time"
)

print("\nLIVE ROW CHECK")
print("Rows with all 19 features:", len(valid_live))

if len(valid_live) > 0:

    latest = valid_live.iloc[-1]

    print("Latest timestamp:", latest["Data Acquisition Time"])
    print("Current water level:", latest["Current_Water_Level"])

print("\n" + "=" * 65)
print("STEP 2AH-D COMPLETE")
print("=" * 65)
```

```python
# ======================================================================
# STEP 2AH-E — ACTUAL V1 LIVE PREDICTION
# ======================================================================

import runpy
import joblib
import pandas as pd

SCRIPT = r"D:\document\Brahmaputra-FloodCast\scripts\fetch\prepare_live_history.py"
MODEL_PATH = r"D:\document\Brahmaputra-FloodCast\models\floodsense_random_forest.joblib"

FINAL_FEATURES = [
    "Current_Water_Level",
    "Month",
    "Hour",
    "DayOfYear",
    "WaterLevel_Lag_1h",
    "WaterLevel_Lag_3h",
    "WaterLevel_Lag_6h",
    "WaterLevel_Lag_12h",
    "WaterLevel_Lag_24h",
    "WaterLevel_RollingMean_6h",
    "WaterLevel_RollingMean_12h",
    "WaterLevel_RollingMean_24h",
    "WaterLevel_RollingMax_24h",
    "WaterLevel_RollingMin_24h",
    "WaterLevel_Change_1h",
    "WaterLevel_Change_3h",
    "WaterLevel_Change_6h",
    "WaterLevel_Change_12h",
    "WaterLevel_Change_24h"
]

TIME_COL = "Data Acquisition Time"

print("=" * 65)
print("STEP 2AH-E — ACTUAL V1 LIVE PREDICTION")
print("=" * 65)

# ----------------------------------------------------------------------
# 1. RUN EXISTING LIVE HISTORY + FEATURE PIPELINE
# ----------------------------------------------------------------------

print("\nRunning existing live pipeline...")

result = runpy.run_path(SCRIPT)

df = result["df"]

print("Live dataframe shape:", df.shape)

# ----------------------------------------------------------------------
# 2. LOAD SAVED RANDOM FOREST
# ----------------------------------------------------------------------

print("\nLoading saved Random Forest...")

model = joblib.load(MODEL_PATH)

print("Model:", type(model).__name__)

# ----------------------------------------------------------------------
# 3. SELECT LATEST VALID ROW
# ----------------------------------------------------------------------

valid_rows = (
    df
    .dropna(subset=FINAL_FEATURES)
    .sort_values(TIME_COL)
)

if valid_rows.empty:
    raise ValueError(
        "No valid live row contains all 19 features."
    )

latest = valid_rows.iloc[-1:]

# ----------------------------------------------------------------------
# 4. CREATE EXACT MODEL INPUT
# ----------------------------------------------------------------------

X_live = latest[FINAL_FEATURES].copy()

print("\nMODEL INPUT")
print("Shape:", X_live.shape)

# ----------------------------------------------------------------------
# 5. ACTUAL PREDICTION
# ----------------------------------------------------------------------

prediction = model.predict(X_live)[0]

# ----------------------------------------------------------------------
# 6. PROBABILITY
# ----------------------------------------------------------------------

probability = None

if hasattr(model, "predict_proba"):
    probability = model.predict_proba(X_live)[0]

# ----------------------------------------------------------------------
# 7. FINAL RESULT
# ----------------------------------------------------------------------

print("\n" + "=" * 65)
print("LIVE PREDICTION RESULT")
print("=" * 65)

print("Timestamp        :", latest[TIME_COL].iloc[0])
print(
    "Current water    :",
    latest["Current_Water_Level"].iloc[0]
)

print("Prediction       :", int(prediction))

if probability is not None:
    print(
        "P(High_Water=0) :",
        round(float(probability[0]), 4)
    )
    print(
        "P(High_Water=1) :",
        round(float(probability[1]), 4)
    )

print("\nModel features   :", len(FINAL_FEATURES))
print("Model type       :", type(model).__name__)

print("=" * 65)
print("STEP 2AH-E COMPLETE")
print("=" * 65)
```

```python
# ======================================================================
# STEP 2AI — AUTOMATIC REAL-TIME V1 PREDICTION
# ======================================================================

import runpy
import joblib
import time
from datetime import datetime

SCRIPT = r"D:\document\Brahmaputra-FloodCast\scripts\fetch\prepare_live_history.py"
MODEL_PATH = r"D:\document\Brahmaputra-FloodCast\models\floodsense_random_forest.joblib"

FINAL_FEATURES = [
    "Current_Water_Level",
    "Month",
    "Hour",
    "DayOfYear",
    "WaterLevel_Lag_1h",
    "WaterLevel_Lag_3h",
    "WaterLevel_Lag_6h",
    "WaterLevel_Lag_12h",
    "WaterLevel_Lag_24h",
    "WaterLevel_RollingMean_6h",
    "WaterLevel_RollingMean_12h",
    "WaterLevel_RollingMean_24h",
    "WaterLevel_RollingMax_24h",
    "WaterLevel_RollingMin_24h",
    "WaterLevel_Change_1h",
    "WaterLevel_Change_3h",
    "WaterLevel_Change_6h",
    "WaterLevel_Change_12h",
    "WaterLevel_Change_24h"
]

TIME_COL = "Data Acquisition Time"

# Load model ONCE
model = joblib.load(MODEL_PATH)

print("=" * 65)
print("STEP 2AI — AUTOMATIC REAL-TIME V1 PREDICTION")
print("=" * 65)

print("Model loaded:", type(model).__name__)
print("19 features:", len(FINAL_FEATURES))

# ----------------------------------------------------------------------
# TEST CYCLE
# ----------------------------------------------------------------------

print("\nStarting live prediction cycle...")
print("Press CTRL+C to stop.\n")

try:

    while True:

        print("=" * 65)
        print(
            "FETCH CYCLE:",
            datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        )
        print("=" * 65)

        # --------------------------------------------------------------
        # RUN EXISTING LIVE DATA + FEATURE PIPELINE
        # --------------------------------------------------------------

        result = runpy.run_path(SCRIPT)

        df = result["df"]

        # --------------------------------------------------------------
        # FIND LATEST VALID ROW
        # --------------------------------------------------------------

        valid_rows = (
            df
            .dropna(subset=FINAL_FEATURES)
            .sort_values(TIME_COL)
        )

        if valid_rows.empty:

            print("No valid 19-feature row available.")

        else:

            latest = valid_rows.iloc[-1:]

            X_live = latest[FINAL_FEATURES]

            prediction = model.predict(X_live)[0]

            probability = model.predict_proba(X_live)[0]

            print("\nLIVE RESULT")
            print("-" * 65)
            print("Data timestamp :", latest[TIME_COL].iloc[0])
            print(
                "Current water  :",
                latest["Current_Water_Level"].iloc[0]
            )
            print("Prediction     :", int(prediction))
            print(
                "High probability:",
                round(float(probability[1]), 4)
            )

        # --------------------------------------------------------------
        # WAIT BEFORE NEXT FETCH
        # --------------------------------------------------------------

        print("\nWaiting 60 seconds for next fetch...")
        time.sleep(60)

except KeyboardInterrupt:

    print("\n")
    print("=" * 65)
    print("LIVE PREDICTION STOPPED")
    print("=" * 65)
```

```python
# ==============================================================
# STEP 2AI — API FRESHNESS CHECK
# ==============================================================

import requests
import pandas as pd
from datetime import datetime

URL = "https://www.nwdp.nwic.gov.in/api/action/datastore_search"

RESOURCE_ID = "51640870-5961-4696-b986-b744231f1c9f"

STATION = "NH15 Crossing Fakirpara Tangni"

WATER_COL = "River Water Level Telemetry Hourly (meter)"

params = {
    "resource_id": RESOURCE_ID,
    "limit": 5000,
    "q": STATION
}

print("=" * 65)
print("STEP 2AI — API FRESHNESS CHECK")
print("=" * 65)

response = requests.get(
    URL,
    params=params,
    timeout=30
)

print("HTTP Status:", response.status_code)

response.raise_for_status()

data = response.json()

records = data["result"]["records"]

print("Records received:", len(records))

df = pd.DataFrame(records)

print("\nColumns:")
print(df.columns.tolist())

# --------------------------------------------------------------
# TIMESTAMP
# --------------------------------------------------------------

TIME_COL = "Data Acquisition Time"

df[TIME_COL] = pd.to_datetime(
    df[TIME_COL],
    errors="coerce"
)

df = df.dropna(
    subset=[TIME_COL]
)

latest_api_time = df[TIME_COL].max()

system_time = pd.Timestamp.now()

difference = system_time - latest_api_time

print("\n" + "-" * 65)
print("FRESHNESS RESULT")
print("-" * 65)

print("System time       :", system_time)
print("Latest API time   :", latest_api_time)
print("Data age          :", difference)

print("\nLatest API record:")

latest_row = (
    df.sort_values(TIME_COL)
      .tail(1)
)

print(
    latest_row[
        [TIME_COL, WATER_COL]
    ].to_string(index=False)
)

print("\n" + "=" * 65)
print("STEP 2AI COMPLETE")
print("=" * 65)
```

```python
# ======================================================================
# STEP 2AK — TRUE LATEST NWDP RECORD TEST
# ======================================================================

import requests
import pandas as pd

URL = "https://www.nwdp.nwic.gov.in/api/action/datastore_search"

RESOURCE_ID = "847f5630-f231-46c0-922d-0f2f379a5cb8"

STATION = "NH15 Crossing Fakipara Tangni"

TIME_COL = "Data Acquisition Time"
WATER_COL = "River Water Level Telemetry Hourly (meter)"

print("=" * 70)
print("STEP 2AK — TRUE LATEST NWDP RECORD TEST")
print("=" * 70)

params = {
    "resource_id": RESOURCE_ID,
    "limit": 20,
    "q": STATION,
    "sort": f"{TIME_COL} desc"
}

response = requests.get(
    URL,
    params=params,
    timeout=30
)

print("HTTP Status:", response.status_code)

if response.status_code != 200:
    raise RuntimeError("NWDP API request failed")

data = response.json()

if not data.get("success"):
    raise RuntimeError("NWDP API returned success=False")

records = data["result"]["records"]

print("Records received:", len(records))

df = pd.DataFrame(records)

if df.empty:
    print("\nNO RECORDS FOUND")
else:

    df[TIME_COL] = pd.to_datetime(
        df[TIME_COL],
        errors="coerce",
        dayfirst=True
    )

    df = df.dropna(subset=[TIME_COL])
    df = df.sort_values(TIME_COL, ascending=False)

    print("\n" + "-" * 70)
    print("TRUE LATEST RECORDS")
    print("-" * 70)

    print(
        df[
            [TIME_COL, WATER_COL]
        ].head(10).to_string(index=False)
    )

    print("\nLatest timestamp:")
    print(df[TIME_COL].max())

    print("\nLatest water level:")
    print(
        df.loc[
            df[TIME_COL].idxmax(),
            WATER_COL
        ]
    )

print("\n" + "=" * 70)
print("STEP 2AK COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 2AL — NWDP LATEST RECORD WITHOUT STATION FILTER
# ======================================================================

import requests
import pandas as pd

URL = "https://www.nwdp.nwic.gov.in/api/action/datastore_search"

RESOURCE_ID = "847f5630-f231-46c0-922d-0f2f379a5cb8"

TIME_COL = "Data Acquisition Time"
STATION_COL = "Station"
WATER_COL = "River Water Level Telemetry Hourly (meter)"

print("=" * 70)
print("STEP 2AL — NWDP LATEST RECORD WITHOUT STATION FILTER")
print("=" * 70)

params = {
    "resource_id": RESOURCE_ID,
    "limit": 20,
    "sort": f"{TIME_COL} desc"
}

response = requests.get(
    URL,
    params=params,
    timeout=30
)

print("HTTP Status:", response.status_code)

if response.status_code != 200:
    raise RuntimeError("NWDP API request failed")

data = response.json()

if not data.get("success"):
    raise RuntimeError("NWDP API returned success=False")

records = data["result"]["records"]

print("Records received:", len(records))

if len(records) == 0:

    print("\nNO RECORDS RETURNED FROM RESOURCE")

else:

    df = pd.DataFrame(records)

    df[TIME_COL] = pd.to_datetime(
        df[TIME_COL],
        errors="coerce",
        dayfirst=True
    )

    df = df.dropna(subset=[TIME_COL])
    df = df.sort_values(
        TIME_COL,
        ascending=False
    )

    print("\n" + "-" * 70)
    print("LATEST RECORDS")
    print("-" * 70)

    cols = [
        TIME_COL,
        STATION_COL,
        WATER_COL
    ]

    available_cols = [
        col for col in cols
        if col in df.columns
    ]

    print(
        df[available_cols]
        .head(20)
        .to_string(index=False)
    )

    print("\nLatest timestamp:")
    print(df[TIME_COL].max())

    print("\nUnique stations in returned records:")
    print(df[STATION_COL].dropna().unique())

print("\n" + "=" * 70)
print("STEP 2AL COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 2AM — NWDP LIVE SOURCE FRESHNESS MONITOR
# ======================================================================

import requests
import pandas as pd
from datetime import datetime

URL = "https://www.nwdp.nwic.gov.in/api/action/datastore_search"

RESOURCE_ID = "847f5630-f231-46c0-922d-0f2f379a5cb8"

TIME_COL = "Data Acquisition Time"
STATION_COL = "Station"
WATER_COL = "River Water Level Telemetry Hourly (meter)"

print("=" * 70)
print("STEP 2AM — NWDP LIVE SOURCE FRESHNESS MONITOR")
print("=" * 70)

params = {
    "resource_id": RESOURCE_ID,
    "limit": 1,
    "sort": f"{TIME_COL} desc"
}

response = requests.get(
    URL,
    params=params,
    timeout=30
)

print("HTTP Status:", response.status_code)

if response.status_code != 200:
    raise RuntimeError("NWDP API request failed")

data = response.json()

if not data.get("success"):
    raise RuntimeError("NWDP API returned success=False")

records = data["result"]["records"]

if len(records) == 0:
    raise RuntimeError("No records returned")

df = pd.DataFrame(records)

df[TIME_COL] = pd.to_datetime(
    df[TIME_COL],
    errors="coerce",
    dayfirst=True
)

latest = df.iloc[0]

latest_time = latest[TIME_COL]
system_time = pd.Timestamp.now()

print("\n" + "-" * 70)
print("FRESHNESS RESULT")
print("-" * 70)

print("System time   :", system_time)
print("Latest API    :", latest_time)

if pd.notna(latest_time):

    age = system_time - latest_time

    print("Data age      :", age)

print("\nLatest station:")
print(latest[STATION_COL])

print("\nLatest water level:")
print(latest[WATER_COL])

print("\n" + "=" * 70)
print("STEP 2AM COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 2AN — NWDP STATION FRESHNESS AUDIT
# ======================================================================

import requests
import pandas as pd

URL = "https://www.nwdp.nwic.gov.in/api/action/datastore_search"

RESOURCE_ID = "847f5630-f231-46c0-922d-0f2f379a5cb8"

TIME_COL = "Data Acquisition Time"
STATION_COL = "Station"
WATER_COL = "River Water Level Telemetry Hourly (meter)"

TARGET_STATION = "NH15 Crossing Fakirpara Tangni"

print("=" * 70)
print("STEP 2AN — NWDP STATION FRESHNESS AUDIT")
print("=" * 70)

# ----------------------------------------------------------------------
# Fetch a larger recent block
# ----------------------------------------------------------------------

params = {
    "resource_id": RESOURCE_ID,
    "limit": 5000,
    "sort": f"{TIME_COL} desc"
}

response = requests.get(
    URL,
    params=params,
    timeout=60
)

print("HTTP Status:", response.status_code)

if response.status_code != 200:
    raise RuntimeError("NWDP API request failed")

data = response.json()

if not data.get("success"):
    raise RuntimeError("NWDP API returned success=False")

records = data["result"]["records"]

print("Records received:", len(records))

df = pd.DataFrame(records)

if df.empty:
    raise RuntimeError("No records returned")

# ----------------------------------------------------------------------
# Timestamp
# ----------------------------------------------------------------------

df[TIME_COL] = pd.to_datetime(
    df[TIME_COL],
    errors="coerce",
    dayfirst=True
)

df = df.dropna(subset=[TIME_COL])

# ----------------------------------------------------------------------
# Sort
# ----------------------------------------------------------------------

df = df.sort_values(
    TIME_COL,
    ascending=False
)

# ----------------------------------------------------------------------
# Overall latest
# ----------------------------------------------------------------------

print("\n" + "-" * 70)
print("OVERALL RESOURCE")
print("-" * 70)

print("Latest timestamp:", df[TIME_COL].max())

# ----------------------------------------------------------------------
# Station list
# ----------------------------------------------------------------------

print("\nUnique stations returned:", df[STATION_COL].nunique())

# ----------------------------------------------------------------------
# Target station
# ----------------------------------------------------------------------

station_df = df[
    df[STATION_COL].astype(str).str.strip().str.casefold()
    ==
    TARGET_STATION.casefold()
].copy()

print("\n" + "-" * 70)
print("TARGET STATION")
print("-" * 70)

print("Target station:", TARGET_STATION)

print("Matching records:", len(station_df))

if station_df.empty:

    print("TARGET STATION NOT FOUND IN THIS BLOCK")

else:

    station_df = station_df.sort_values(
        TIME_COL,
        ascending=False
    )

    print(
        station_df[
            [TIME_COL, STATION_COL, WATER_COL]
        ].head(20).to_string(index=False)
    )

    latest_station_time = station_df[TIME_COL].max()

    print("\nLatest target timestamp:")
    print(latest_station_time)

    print("\nLatest target water level:")
    print(
        station_df.iloc[0][WATER_COL]
    )

# ----------------------------------------------------------------------
# Latest timestamp by station
# ----------------------------------------------------------------------

print("\n" + "-" * 70)
print("LATEST TIMESTAMP BY STATION")
print("-" * 70)

station_latest = (
    df.groupby(STATION_COL)[TIME_COL]
      .max()
      .sort_values(ascending=False)
)

print(station_latest.head(20).to_string())

print("\n" + "=" * 70)
print("STEP 2AN COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 2AP — NWDP RESOURCE DATA RANGE AUDIT
# ======================================================================

import requests
import pandas as pd

URL = "https://www.nwdp.nwic.gov.in/api/action/datastore_search"

RESOURCE_ID = "847f5630-f231-46c0-922d-0f2f379a5cb8"

TIME_COL = "Data Acquisition Time"
STATION_COL = "Station"
WATER_COL = "River Water Level Telemetry Hourly (meter)"

TARGET_STATION = "NH15 Crossing Fakirpara Tangni"

print("=" * 70)
print("STEP 2AP — NWDP RESOURCE DATA RANGE AUDIT")
print("=" * 70)

params = {
    "resource_id": RESOURCE_ID,
    "limit": 5000,
    "sort": f"{TIME_COL} desc"
}

response = requests.get(
    URL,
    params=params,
    timeout=60
)

print("HTTP Status:", response.status_code)

if response.status_code != 200:
    raise RuntimeError("NWDP API request failed")

data = response.json()

if not data.get("success"):
    raise RuntimeError("NWDP API returned success=False")

records = data["result"]["records"]

print("Records received:", len(records))

df = pd.DataFrame(records)

df[TIME_COL] = pd.to_datetime(
    df[TIME_COL],
    errors="coerce",
    dayfirst=True
)

df = df.dropna(subset=[TIME_COL])

df = df[
    df[STATION_COL].astype(str).str.strip().str.casefold()
    == TARGET_STATION.casefold()
].copy()

df = df.sort_values(TIME_COL)

print("\n" + "-" * 70)
print("TARGET STATION DATA RANGE")
print("-" * 70)

print("Station:", TARGET_STATION)
print("Rows:", len(df))

print("Oldest timestamp:", df[TIME_COL].min())
print("Latest timestamp:", df[TIME_COL].max())

print("\nLatest 10 records:")
print(
    df[
        [TIME_COL, WATER_COL]
    ].tail(10).to_string(index=False)
)

# ----------------------------------------------------------------------
# Hourly gap check near latest data
# ----------------------------------------------------------------------

if len(df) >= 2:

    latest_times = df[TIME_COL].tail(50)

    gaps = latest_times.diff().dropna()

    print("\n" + "-" * 70)
    print("RECENT TIME GAP CHECK")
    print("-" * 70)

    print("Maximum recent gap:", gaps.max())
    print("Median recent gap :", gaps.median())

# ----------------------------------------------------------------------
# Freshness
# ----------------------------------------------------------------------

system_time = pd.Timestamp.now()
latest_time = df[TIME_COL].max()

age = system_time - latest_time

print("\n" + "-" * 70)
print("CURRENT FRESHNESS")
print("-" * 70)

print("System time :", system_time)
print("Latest API  :", latest_time)
print("Data age    :", age)

print("\n" + "=" * 70)
print("STEP 2AP COMPLETE")
print("=" * 70)
```

```python
# ======================================================================
# STEP 2AQ — OFFICIAL API RESOURCE DIRECT TEST
# ======================================================================

import requests
import pandas as pd

RESOURCE_ID = "847f5630-f231-46c0-922d-0f2f379a5cb8"

URL = "https://www.nwdp.nwic.gov.in/api/action/datastore_search"

TIME_COL = "Data Acquisition Time"
STATION_COL = "Station"
WATER_COL = "River Water Level Telemetry Hourly (meter)"

print("=" * 70)
print("STEP 2AQ — OFFICIAL API RESOURCE DIRECT TEST")
print("=" * 70)

params = {
    "resource_id": RESOURCE_ID,
    "limit": 100,
    "sort": f"{TIME_COL} desc"
}

response = requests.get(
    URL,
    params=params,
    timeout=30
)

print("HTTP Status:", response.status_code)

data = response.json()

print("API success:", data.get("success"))

records = data.get("result", {}).get("records", [])

print("Records received:", len(records))

if len(records) == 0:
    print("\nNO RECORDS FOUND")

else:

    df = pd.DataFrame(records)

    df[TIME_COL] = pd.to_datetime(
        df[TIME_COL],
        errors="coerce",
        dayfirst=True
    )

    df = df.dropna(subset=[TIME_COL])
    df = df.sort_values(TIME_COL, ascending=False)

    print("\n" + "-" * 70)
    print("LATEST API RECORDS")
    print("-" * 70)

    print(
        df[
            [TIME_COL, STATION_COL, WATER_COL]
        ].head(20).to_string(index=False)
    )

    print("\nLatest timestamp:")
    print(df[TIME_COL].max())

    print("\nStations returned:")
    print(df[STATION_COL].dropna().unique())

print("\n" + "=" * 70)
print("STEP 2AQ COMPLETE")
print("=" * 70)
```
